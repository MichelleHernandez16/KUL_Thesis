```python

# ========= CONFIGURATION =========

#first start a headless Chrome session (no visible window)
def setup_selenium_driver():
    """Setup Selenium WebDriver"""
    chrome_options = Options()
    chrome_options.add_argument("--headless") #to run browser in backgrownd
    chrome_options.add_argument("--no-sandbox") #
    chrome_options.add_argument("--disable-dev-shm-usage") #For cserver compatbility
    chrome_options.add_argument("--disable-blink-features=AutomationControlled")
    chrome_options.add_experimental_option("excludeSwitches", ["enable-automation"])
    chrome_options.add_experimental_option('useAutomationExtension', False)
    chrome_options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36")
    
    driver = webdriver.Chrome(options=chrome_options)
    driver.execute_script("Object.defineProperty(navigator, 'webdriver', {get: () => undefined})")
    return driver

# ========= MULTIPLE SEASONS CODE =========
#First get the list of urls to scrap
def get_la_liga_playing_time_seasons_2010_2015():
    """To create a list of URLs for each season from start year to end year according to the fbref urls"""
    base_url = "https://fbref.com/en/comps/12/{season}/playingtime/{season}-La-Liga-Stats"
    
    seasons = []
    # Seasons de 2010-2011 a 2014-2015 
    for year in range(2010, 2015):  
        season = f"{year}-{year+1}" #string year-year
        url = base_url.format(season=season)
        seasons.append((season, url))
    
    return seasons

# Core scrapping function
def extract_playing_time_table_selenium(driver, url, season):
    """Extracts the Playing Time tables with stats"""
    print(f"\n📥 Extracting PLAYING TIME stats from: {season}")
    
    try:
        driver.get(url) #Navigates to the url
        time.sleep(4) #Waits for the page to load (4secs)
        
        soup = BeautifulSoup(driver.page_source, 'html.parser') #Parses the HTML (grabs the Html that the browser loads after 4 s)
        
        #  Search for the table with stats we want and convert to a df
        stats_table = soup.find('table', {'id': 'stats_playing_time'})
        
        if not stats_table:
            print("❌ No stats_playing_time table found")
            return []
        
        # Read table with pandas
        df = pd.read_html(io.StringIO(str(stats_table)))[0]
        print(f"✅ Found PLAYING TIME table with {len(df)} rows")
        
        # I want to check the column structure (5 rows)
        print(f"🔍 Table structure: {df.shape}")
        print(f"📋 First 5 columns: {df.columns.tolist()[:5]}")
        
        # Process multi-index columns
        if isinstance(df.columns, pd.MultiIndex):
            print("🔍 Multi-index columns detected - processing...")
            
            # For flatten columns
            new_columns = []
            for col in df.columns:
                if pd.isna(col[0]) or col[0] == '':
                    new_columns.append(col[1])
                elif col[1] == '':
                    new_columns.append(col[0])
                else:
                    # Para columnas como ('Playing Time', 'MP'), usar solo el segundo nivel
                    new_columns.append(col[1])
            
            df.columns = new_columns
            print(f"📋 Processed columns: {new_columns[:10]}...")
        
        # Map data and extraction
        # Search for principal columns
        column_mapping = {
            'player': None,
            'squad': None, 
            'mp': None,     # Matches Played
            'min': None,    # Minutes
            'starts': None, #Games started
            'gls': None,    # Goals
            'ast': None,    # Assists
            'age': None,
            'pos': None,
            'nation': None
        }
        
        for col in df.columns:
            col_str = str(col).lower()
            
            if 'player' in col_str:
                column_mapping['player'] = col
            elif 'squad' in col_str:
                column_mapping['squad'] = col
            elif col_str == 'mp' or 'matches' in col_str:
                column_mapping['mp'] = col
            elif col_str == 'min' or 'minutes' in col_str:
                column_mapping['min'] = col
            elif 'start' in col_str:
                column_mapping['starts'] = col
            elif col_str == 'gls' or 'goals' in col_str:
                column_mapping['gls'] = col
            elif col_str == 'ast' or 'assists' in col_str:
                column_mapping['ast'] = col
            elif 'age' in col_str:
                column_mapping['age'] = col
            elif 'pos' in col_str:
                column_mapping['pos'] = col
            elif 'nation' in col_str:
                column_mapping['nation'] = col
        
        print(f"🎯 Column mapping: Player={column_mapping['player']}, MP={column_mapping['mp']}, Min={column_mapping['min']}")
        
        if not column_mapping['player']:
            print("❌ Could not find Player column")
            return []
        
        # Data cleaning and preparation
        players_data = []
        processed_count = 0
        
        for idx, row in df.iterrows():
            try:
                player_name = str(row[column_mapping['player']]) if column_mapping['player'] in row and pd.notna(row[column_mapping['player']]) else ''
                
                # Filter invalid rows
                if (not player_name or player_name == 'Player' or player_name == 'nan' or 
                    player_name == 'Rk' or any(keyword in player_name for keyword in ['Player', 'Squad', 'Nation'])):
                    continue
                
                # Extract player data
                player_data = {
                    'season': season,
                    'player_name': player_name.strip(),
                    'rank': idx + 1
                }
                
                # Función helper to get the values
                def get_value(col_key, default=''):
                    col = column_mapping[col_key]
                    if col and col in row and pd.notna(row[col]):
                        value = str(row[col])
                        # Filter non numerical values to extract the default
                        if value in ['Matches', 'Player', 'Squad', 'Nation']:
                            return default
                        return value
                    return default
                
                # Add mapped values
                if column_mapping['squad']:
                    player_data['squad'] = get_value('squad')
                
                if column_mapping['mp']:
                    mp_value = get_value('mp')
                    # Number?
                    if mp_value and mp_value.replace('.', '').isdigit():
                        player_data['matches_played'] = mp_value
                
                if column_mapping['min']:
                    min_value = get_value('min')
                    if min_value and min_value.replace(',', '').replace('.', '').isdigit():
                        player_data['minutes'] = min_value
                
                if column_mapping['starts']:
                    starts_value = get_value('starts')
                    if starts_value and starts_value.replace('.', '').isdigit():
                        player_data['starts'] = starts_value
                
                if column_mapping['gls']:
                    goals_value = get_value('gls')
                    if goals_value and goals_value.replace('.', '').isdigit():
                        player_data['goals'] = goals_value
                
                if column_mapping['ast']:
                    assists_value = get_value('ast')
                    if assists_value and assists_value.replace('.', '').isdigit():
                        player_data['assists'] = assists_value
                
                if column_mapping['age']:
                    player_data['age'] = get_value('age')
                
                if column_mapping['pos']:
                    player_data['position'] = get_value('pos')
                
                if column_mapping['nation']:
                    player_data['nation'] = get_value('nation')
                
                # Add if the name and team is available
                if player_data['player_name'] and player_data.get('squad'):
                    players_data.append(player_data)
                    processed_count += 1
                    
                    # Validate 3 first players
                    if processed_count <= 3:
                        mp_debug = player_data.get('matches_played', 'N/A')
                        min_debug = player_data.get('minutes', 'N/A')
                        print(f"   📝 Sample: {player_data['player_name']} - {player_data['squad']} - MP: {mp_debug}, Min: {min_debug}")
                
            except Exception as e:
                print(f"    Error processing row {idx}: {e}")
                continue
        
        print(f" Successfully processed {len(players_data)} valid player records")
        return players_data
        
    except Exception as e:
        print(f" Error extracting playing time stats: {e}")
        return []

## Main execution funcion
def main_playing_time_2010_2015():
    """Main funcion to extract stats for Playing Time"""
    print(" LA LIGA PLAYING TIME STATS EXTRACTION (2010-2015)")
    print("=" * 70)
    
    driver = setup_selenium_driver()
    
    try:
        all_players = []
        seasons = get_la_liga_playing_time_seasons_2010_2015()
        
        print(f" Processing {len(seasons)} seasons:")
        for season, url in seasons:
            print(f"   - {season}")
        
        for season, url in seasons:
            print(f"\n Processing {season}...")
            print(f"   URL: {url}")
            
            players = extract_playing_time_table_selenium(driver, url, season)
            all_players.extend(players)
            
            print(f" {season}: {len(players)} players extracted")
            
            time.sleep(random.uniform(3, 6))
        
        # Save results
        if all_players:
            df_players = pd.DataFrame(all_players)
            
            # Reorder columns
            preferred_order = ['season', 'player_name', 'squad', 'position', 'age', 'nation', 
                             'matches_played', 'minutes', 'starts', 'goals', 'assists', 'rank']
            
            existing_columns = [col for col in preferred_order if col in df_players.columns]
            other_columns = [col for col in df_players.columns if col not in preferred_order]
            final_columns = existing_columns + other_columns
            
            df_players = df_players[final_columns]
            
            # Save files
            output_file = "la_liga_stats_2010_2015.csv"
            #excel_file = "la_liga_stats_2015_2020.xlsx"
            
            df_players.to_csv(output_file, index=False, encoding='utf-8')
           # df_players.to_excel(excel_file, index=False, engine='openpyxl')
            
            print(f"\n" + "=" * 70)
            print(" EXTRACTION SUMMARY")
            print("=" * 70)
            print(f" Total players extracted: {len(all_players)}")
            print(f"Seasons processed: {len(set([p['season'] for p in all_players]))}")
            
            if 'squad' in df_players.columns:
                print(f"Unique clubs: {df_players['squad'].nunique()}")
            
            # Stats for numerical data
            if 'matches_played' in df_players.columns:
                try:
                    df_players['mp_numeric'] = pd.to_numeric(df_players['matches_played'], errors='coerce')
                    valid_mp = df_players['mp_numeric'].notna().sum()
                    print(f"📊 Valid matches played data: {valid_mp}/{len(df_players)} players")
                except:
                    pass
            
            if 'minutes' in df_players.columns:
                try:
                    df_players['min_numeric'] = pd.to_numeric(df_players['minutes'], errors='coerce')
                    valid_min = df_players['min_numeric'].notna().sum()
                    print(f" Valid minutes data: {valid_min}/{len(df_players)} players")
                except:
                    pass
            
            print(f" Saved to:")
            print(f"   - {output_file} (CSV)")
           # print(f"   - {excel_file} (Excel)") #not interested in excel rn
            
            # Show sample
            print(f"\n SAMPLE DATA (first 8 players with numeric stats):")
            sample_cols = ['season', 'player_name', 'squad', 'matches_played', 'minutes', 'goals']
            available_cols = [col for col in sample_cols if col in df_players.columns]
            
            if available_cols:
                # Filter just numerical cols
                sample_df = df_players[available_cols].head(10)
                print(sample_df.to_string(index=False))
            
            # Show top players (minutes)
            if 'minutes' in df_players.columns:
                try:
                    df_players['min_numeric'] = pd.to_numeric(df_players['minutes'], errors='coerce')
                    top_players = df_players.nlargest(5, 'min_numeric')[['player_name', 'squad', 'minutes']]
                    print(f"\n⏱️ TOP 5 PLAYERS BY MINUTES:")
                    print(top_players.to_string(index=False))
                except:
                    pass
            
        else:
            print(" No players were extracted")
    
    except Exception as e:
        print(f" Main function error: {e}")
    
    finally:
        driver.quit()
        print("\n Selenium driver closed")

if __name__ == "__main__":
    main_playing_time_2010_2015()
```

```python
import pandas as pd
from bs4 import BeautifulSoup
import re
import time
import io
import requests
from urllib.parse import quote
import concurrent.futures
from tqdm import tqdm

# ========= CONFIGURATION =========
INPUT_CSV = "la_liga_stats_2010_2015.csv"
OUTPUT_CSV = "la_liga_players_EFFICIENT.csv"
MAX_WORKERS = 3  # Be respectful to the server

def create_fbref_url(player_name):
    """Create FBRef URL directly from player name"""
    # Clean and create slug
    cleaned = re.sub(r'[^\w\s]', '', player_name)
    cleaned = re.sub(r'\s+', ' ', cleaned).strip()
    
    # Create slug: "Adriano Correia" -> "Adriano-Correia"
    slug = cleaned.lower()
    slug = re.sub(r'[áàäâ]', 'a', slug)
    slug = re.sub(r'[éèëê]', 'e', slug)
    slug = re.sub(r'[íìïî]', 'i', slug)
    slug = re.sub(r'[óòöô]', 'o', slug)
    slug = re.sub(r'[úùüû]', 'u', slug)
    slug = re.sub(r'[ñ]', 'n', slug)
    slug = re.sub(r'[ç]', 'c', slug)
    slug = re.sub(r'[^\w\s-]', '', slug)
    slug = re.sub(r'\s+', '-', slug)
    
    # For common names like Adriano, we'll try multiple URL patterns
    base_url = f"https://fbref.com/en/players/{slug}"
    
    return base_url

def get_player_info(player_name):
    """Get player info by trying multiple URL strategies"""
    print(f" Processing: {player_name}")
    
    # Initialize result
    result = {
        'original_name': player_name,
        'player_slug': create_fbref_url(player_name).split('/')[-1],
        'status': 'Not found',
        'player_url': 'Not found',
        'name_home_country': 'Not found',
        'position': 'Not found',
        'age': 'Not found',
        'place_birth': 'Not found',
        'height': 'Not found',
        'citizenship': 'Not found',
        'foot': 'Not found',
        'current_club': 'Not found',
        'debut_season': 'Not found',
        'debut_age': 'Not found',
        'first_club': 'Not found',
        'debut_year': 'Not found'
    }
    
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36',
        'Accept': 'text/html,application/xhtml+xml,application/xml;q=0.9,image/webp,*/*;q=0.8',
        'Accept-Language': 'en-US,en;q=0.5',
        'Accept-Encoding': 'gzip, deflate, br',
        'Connection': 'keep-alive',
        'Upgrade-Insecure-Requests': '1',
    }
    
    try:
        # Strategy 1: Try direct slug URL first
        base_url = create_fbref_url(player_name)
        test_urls = [base_url]
        
        # Strategy 2: If name has multiple words, try different combinations
        name_parts = player_name.split()
        if len(name_parts) >= 2:
            # Try "Firstname-Lastname"
            simple_slug = f"{name_parts[0].lower()}-{name_parts[1].lower()}"
            test_urls.append(f"https://fbref.com/en/players/{simple_slug}")
            
            # Try with first letter of first name + lastname
            if len(name_parts[0]) > 1:
                initial_slug = f"{name_parts[0][0].lower()}-{name_parts[1].lower()}"
                test_urls.append(f"https://fbref.com/en/players/{initial_slug}")
        
        # Try each URL pattern
        for test_url in test_urls:
            try:
                response = requests.get(test_url, headers=headers, timeout=10)
                
                if response.status_code == 200:
                    soup = BeautifulSoup(response.content, 'html.parser')
                    
                    # Check if this is a valid player page (has meta div)
                    meta_div = soup.find('div', {'id': 'meta'})
                    if meta_div:
                        # Extract player name from page to verify
                        name_h1 = meta_div.find('h1')
                        if name_h1:
                            page_name = name_h1.get_text(strip=True)
                            
                            # Basic verification - check if names are similar
                            if is_name_match(player_name, page_name):
                                result['player_url'] = test_url
                                result['status'] = 'Found'
                                result['name_home_country'] = page_name
                                
                                # Extract all other info
                                extract_all_info_from_page(soup, result)
                                print(f" Found: {player_name} -> {page_name}")
                                return result
                
                time.sleep(1)  # Brief delay between attempts
                
            except requests.RequestException:
                continue
        
        # If no direct URL worked, try search as last resort (but much slower)
        print(f"    Direct methods failed, trying search for: {player_name}")
        search_result = try_search_fallback(player_name, headers)
        if search_result:
            return search_result
        
        print(f" Not found: {player_name}")
        
    except Exception as e:
        print(f" Error processing {player_name}: {e}")
        result['status'] = f'Error: {str(e)[:50]}'
    
    return result

def is_name_match(search_name, found_name):
    """Check if names are reasonably similar"""
    search_clean = search_name.lower().replace('-', ' ').replace('.', '').strip()
    found_clean = found_name.lower().replace('-', ' ').replace('.', '').strip()
    
    # Exact match
    if search_clean == found_clean:
        return True
    
    # Contains the search name
    if search_clean in found_clean:
        return True
    
    # Shared words (for multi-word names)
    search_words = set(search_clean.split())
    found_words = set(found_clean.split())
    common_words = search_words.intersection(found_words)
    
    # If at least one word matches, consider it a potential match
    return len(common_words) >= 1

def extract_all_info_from_page(soup, result):
    """Extract all information from player page"""
    try:
        meta_div = soup.find('div', {'id': 'meta'})
        if not meta_div:
            return
        
        meta_text = meta_div.get_text()
        
        # POSITION
        position_match = re.search(r'Position:\s*([^\n▪]+)', meta_text)
        if position_match:
            result['position'] = position_match.group(1).strip()
        
        # FOOT
        foot_match = re.search(r'Footed:\s*([^\n]+)', meta_text)
        if foot_match:
            result['foot'] = foot_match.group(1).strip()
        
        # HEIGHT
        height_match = re.search(r'(\d+cm|\d+,\d+\s*m|\d\'\d+\")', meta_text)
        if height_match:
            result['height'] = height_match.group(1).strip()
        
        # BIRTH INFO
        born_match = re.search(r'Born:\s*([^\(]+)\s*\(([^)]+)\)\s*in\s*([^\n]+)', meta_text)
        if born_match:
            result['age'] = born_match.group(2).strip()
            result['place_birth'] = born_match.group(3).strip()
        
        # CITIZENSHIP
        national_match = re.search(r'National Team:\s*([^\n]+)', meta_text)
        if national_match:
            citizenship = national_match.group(1).strip()
            citizenship = re.sub(r'\s+[a-z]{2}$', '', citizenship)
            result['citizenship'] = citizenship
        
        # CURRENT CLUB
        club_match = re.search(r'Club:\s*([^\n]+)', meta_text)
        if club_match:
            result['current_club'] = club_match.group(1).strip()
        
        # DEBUT INFO (from stats tables)
        extract_debut_info(soup, result)
        
    except Exception as e:
        print(f"    Info extraction warning for {result['original_name']}: {e}")

def extract_debut_info(soup, result):
    """Extract debut information from stats tables"""
    try:
        tables = soup.find_all('table', class_=re.compile('stats_table'))
        
        for table in tables:
            try:
                df = pd.read_html(io.StringIO(str(table)))[0]
                
                if 'Season' in str(df.columns) and len(df) > 1:
                    # Find first valid season
                    for idx in range(len(df)):
                        season_val = str(df.iloc[idx, 0])
                        
                        if (season_val and season_val != 'nan' and 
                            'Season' not in season_val):
                            
                            result['debut_season'] = season_val
                            
                            if '-' in season_val:
                                result['debut_year'] = season_val.split('-')[0]
                            
                            # Try to find age
                            for col in df.columns:
                                if 'Age' in str(col):
                                    age_val = df.iloc[idx][col]
                                    if pd.notna(age_val):
                                        result['debut_age'] = str(age_val)
                                        break
                            
                            # Try to find first club
                            for col in df.columns:
                                if 'Squad' in str(col):
                                    club_val = df.iloc[idx][col]
                                    if pd.notna(club_val):
                                        result['first_club'] = str(club_val)
                                        break
                            
                            return
            except:
                continue
    except:
        pass

def try_search_fallback(player_name, headers):
    """Fallback to search if direct URLs fail (but much slower)"""
    try:
        search_url = f"https://fbref.com/en/search/search.fcgi?search={quote(player_name)}"
        response = requests.get(search_url, headers=headers, timeout=10)
        
        if response.status_code == 200:
            soup = BeautifulSoup(response.content, 'html.parser')
            
            # Look for player links in search results
            player_links = soup.find_all('a', href=re.compile(r'/en/players/[a-f0-9]+/'))
            
            for link in player_links[:3]:  # Check first 3 results
                player_url = "https://fbref.com" + link['href']
                found_name = link.get_text(strip=True)
                
                if is_name_match(player_name, found_name):
                    # Found a match, get the page
                    player_response = requests.get(player_url, headers=headers, timeout=10)
                    if player_response.status_code == 200:
                        player_soup = BeautifulSoup(player_response.content, 'html.parser')
                        
                        result = {
                            'original_name': player_name,
                            'player_slug': player_url.split('/')[-2],
                            'status': 'Found via search',
                            'player_url': player_url,
                            'name_home_country': found_name
                        }
                        
                        # Initialize other fields
                        other_fields = ['position', 'age', 'place_birth', 'height', 'citizenship', 
                                      'foot', 'current_club', 'debut_season', 'debut_age', 'first_club', 'debut_year']
                        for field in other_fields:
                            result[field] = 'Not found'
                        
                        extract_all_info_from_page(player_soup, result)
                        print(f"✅ Found via search: {player_name} -> {found_name}")
                        return result
        
        time.sleep(2)  # Be respectful
        
    except Exception as e:
        print(f"    Search fallback failed: {e}")
    
    return None

def process_batch_parallel(player_names, batch_size=50):
    """Process players in parallel batches"""
    all_results = []
    
    # Process in batches to avoid overwhelming the server
    for i in range(0, len(player_names), batch_size):
        batch = player_names[i:i + batch_size]
        print(f"\n Processing batch {i//batch_size + 1}/{(len(player_names)-1)//batch_size + 1}")
        print(f"📋 Batch size: {len(batch)} players")
        
        # Process batch with limited parallelism
        with concurrent.futures.ThreadPoolExecutor(max_workers=MAX_WORKERS) as executor:
            batch_results = list(tqdm(
                executor.map(get_player_info, batch),
                total=len(batch),
                desc="Processing"
            ))
        
        all_results.extend(batch_results)
        
        # Save progress after each batch
        df_batch = pd.DataFrame(all_results)
        df_batch.to_csv(f"progress_batch_{i//batch_size + 1}.csv", index=False)
        
        print(f" Saved progress batch {i//batch_size + 1}")
        
        # Be respectful between batches
        if i + batch_size < len(player_names):
            delay = 10
            print(f"⏳ Waiting {delay} seconds before next batch...")
            time.sleep(delay)
    
    return all_results

def main():
    """Main processing function"""
    print(" PLAYER DATA EXTRACTION")
    print("=" * 60)
    
    # Load player names
    try:
        df = pd.read_csv(INPUT_CSV)
        if 'original_name' in df.columns:
            player_names = df['original_name'].dropna().unique().tolist()
        elif 'player_name' in df.columns:
            player_names = df['player_name'].dropna().unique().tolist()
        else:
            player_names = df.iloc[:, 0].dropna().unique().tolist()
        
        print(f" Loaded {len(player_names)} players")
        
    except Exception as e:
        print(f" Error loading CSV: {e}")
        return
    
    # Test with small sample first
    if len(player_names) > 20:
        print(" Testing with first 20 players...")
        test_players = player_names[:20]
    else:
        test_players = player_names
    
    print(f" Test players: {len(test_players)}")
    
    # Process test batch
    test_results = process_batch_parallel(test_players, batch_size=10)
    
    # Analyze test results
    found_count = len([r for r in test_results if r['status'] in ['Found', 'Found via search']])
    success_rate = (found_count / len(test_results)) * 100
    
    print(f"\n TEST BATCH RESULTS:")
    print(f"   Total: {len(test_results)}")
    print(f"   Found: {found_count}")
    print(f"   Success rate: {success_rate:.1f}%")
    
    if success_rate >= 50:  # If decent success rate, process all
        print(f"\n Success rate good! Processing all {len(player_names)} players...")
        all_results = process_batch_parallel(player_names, batch_size=50)
        
        # Save final results
        df_final = pd.DataFrame(all_results)
        df_final.to_csv(OUTPUT_CSV, index=False, encoding='utf-8')
        
        # Final summary
        final_found = len([r for r in all_results if r['status'] in ['Found', 'Found via search']])
        final_rate = (final_found / len(all_results)) * 100
        
        print(f"\n COMPLETED!")
        print(f" Final results: {final_found}/{len(all_results)} found ({final_rate:.1f}%)")
        print(f" Saved to: {OUTPUT_CSV}")
        
    else:
        print(f"\n Low success rate ({success_rate:.1f}%). Check the test results before processing all players.")
        df_test = pd.DataFrame(test_results)
        df_test.to_csv("test_results.csv", index=False)
        print(" Test results saved to: test_results.csv")

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
from bs4 import BeautifulSoup
import re
import time
import io
import requests
from urllib.parse import urljoin
import concurrent.futures
from tqdm import tqdm

# ========= CONFIGURATION =========
INPUT_CSV = "la_liga_players_with_links.csv"  # table with player links
OUTPUT_CSV = "la_liga_players_COMPLETE_INFO.csv"
MAX_WORKERS = 5  # Can be higher since we have direct URLs

def extract_player_urls_from_existing_table(csv_file):
    """Extract player names and URLs from your existing table"""
    print("📁 Extracting player URLs from existing table...")
    
    try:
        df = pd.read_csv(csv_file)
        print(f"✅ Loaded table with {len(df)} rows")
        print(f"📋 Columns: {df.columns.tolist()}")
        
        players_data = []
        
        # Look for player name column and links
        name_column = None
        link_column = None
        
        # Common column names for player names
        name_columns = ['Player', 'player_name', 'Name', 'name']
        for col in name_columns:
            if col in df.columns:
                name_column = col
                break
        
        if not name_column:
            print(" No player name column found")
            return []
        
        print(f" Using name column: '{name_column}'")
        
        # Check if there are direct URLs in the table
        # Sometimes they're in the Player column as HTML links
        # Or there might be a separate URL column
        
        # Strategy 1: Check if Player column contains HTML links
        sample_value = str(df[name_column].iloc[0]) if len(df) > 0 else ""
        if '<a href=' in sample_value:
            print(" Found HTML links in Player column - extracting URLs...")
            players_data = extract_urls_from_html_column(df, name_column)
        
        # Strategy 2: Look for URL columns
        elif any('url' in col.lower() or 'link' in col.lower() for col in df.columns):
            url_columns = [col for col in df.columns if 'url' in col.lower() or 'link' in col.lower()]
            print(f" Found URL columns: {url_columns}")
            players_data = extract_urls_from_url_columns(df, name_column, url_columns[0])
        
        # Strategy 3: If no direct URLs, check for "Matches" links
        elif 'Matches' in df.columns:
            print(" Checking 'Matches' column for player links...")
            players_data = extract_urls_from_matches_column(df, name_column)
        
        else:
            print(" No obvious URL columns found. Checking all columns for links...")
            players_data = find_urls_in_any_column(df, name_column)
        
        print(f" Extracted {len(players_data)} players with URLs")
        return players_data
        
    except Exception as e:
        print(f" Error extracting URLs: {e}")
        return []

def extract_urls_from_html_column(df, name_column):
    """Extract URLs from Player column that contains HTML links"""
    players_data = []
    
    for idx, row in df.iterrows():
        player_html = str(row[name_column])
        
        # Parse HTML to extract link and text
        soup = BeautifulSoup(player_html, 'html.parser')
        link = soup.find('a')
        
        if link and link.get('href'):
            player_name = link.get_text(strip=True)
            player_url = link['href']
            
            # Make URL absolute if it's relative
            if player_url.startswith('/'):
                player_url = "https://fbref.com" + player_url
            
            players_data.append({
                'original_name': player_name,
                'player_url': player_url,
                'source_row': idx
            })
        else:
            # Fallback: just use the text content
            player_name = re.sub(r'<[^>]+>', '', player_html).strip()
            players_data.append({
                'original_name': player_name,
                'player_url': 'Not found in HTML',
                'source_row': idx
            })
    
    return players_data

def extract_urls_from_url_columns(df, name_column, url_column):
    """Extract URLs from dedicated URL columns"""
    players_data = []
    
    for idx, row in df.iterrows():
        player_name = str(row[name_column])
        player_url = str(row[url_column]) if pd.notna(row[url_column]) else 'Not found'
        
        # Clean the URL
        if player_url != 'Not found' and player_url.startswith('/'):
            player_url = "https://fbref.com" + player_url
        
        players_data.append({
            'original_name': player_name,
            'player_url': player_url,
            'source_row': idx
        })
    
    return players_data

def extract_urls_from_matches_column(df, name_column):
    """Extract URLs from 'Matches' column which often contains player links"""
    players_data = []
    
    for idx, row in df.iterrows():
        player_name = str(row[name_column])
        matches_html = str(row['Matches']) if 'Matches' in row and pd.notna(row['Matches']) else ''
        
        if matches_html and '<a href=' in matches_html:
            soup = BeautifulSoup(matches_html, 'html.parser')
            link = soup.find('a')
            
            if link and link.get('href'):
                player_url = link['href']
                if player_url.startswith('/'):
                    player_url = "https://fbref.com" + player_url
                
                players_data.append({
                    'original_name': player_name,
                    'player_url': player_url,
                    'source_row': idx
                })
            else:
                players_data.append({
                    'original_name': player_name,
                    'player_url': 'Not found in Matches',
                    'source_row': idx
                })
        else:
            players_data.append({
                'original_name': player_name,
                'player_url': 'No Matches link',
                'source_row': idx
            })
    
    return players_data

def find_urls_in_any_column(df, name_column):
    """Search all columns for player URLs"""
    players_data = []
    
    for idx, row in df.iterrows():
        player_name = str(row[name_column])
        player_url = 'Not found'
        
        # Check every column for this row
        for col_name, value in row.items():
            if pd.notna(value) and isinstance(value, str):
                # Look for FBRef player URLs
                if '/players/' in value and '/en/players/' in value:
                    # Extract URL from the string
                    url_match = re.search(r'(https://fbref\.com/en/players/[^\s<>"]+|/en/players/[^\s<>"]+)', value)
                    if url_match:
                        player_url = url_match.group(1)
                        if player_url.startswith('/'):
                            player_url = "https://fbref.com" + player_url
                        break
        
        players_data.append({
            'original_name': player_name,
            'player_url': player_url,
            'source_row': idx
        })
    
    return players_data

def get_player_details(player_data):
    """Get detailed player information using the direct URL"""
    player_name = player_data['original_name']
    player_url = player_data['player_url']
    
    # Initialize result with basic info
    result = player_data.copy()
    
    # Add all the fields we want to extract
    info_fields = [
        'name_home_country', 'position', 'age', 'birth_date', 'place_birth', 
        'height', 'citizenship', 'foot', 'current_club', 'debut_season', 
        'debut_age', 'first_club', 'debut_year', 'status', 'scraping_notes',
        'player_slug'
    ]
    
    for field in info_fields:
        if field not in result:
            result[field] = 'Not found'
    
    # Extract player slug from URL
    if player_url != 'Not found':
        slug_match = re.search(r'/players/[a-f0-9]+/([^/]+)', player_url)
        if slug_match:
            result['player_slug'] = slug_match.group(1)
    
    if player_url == 'Not found' or not player_url.startswith('http'):
        result['status'] = 'No valid URL'
        return result
    
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36',
        'Accept': 'text/html,application/xhtml+xml,application/xml;q=0.9,image/webp,*/*;q=0.8',
    }
    
    try:
        print(f"🔍 Extracting: {player_name}")
        response = requests.get(player_url, headers=headers, timeout=10)
        
        if response.status_code == 200:
            soup = BeautifulSoup(response.content, 'html.parser')
            
            # Verify this is a valid player page
            meta_div = soup.find('div', {'id': 'meta'})
            if meta_div:
                result['status'] = 'Success'
                
                # Extract player name from page
                name_h1 = meta_div.find('h1')
                if name_h1:
                    result['name_home_country'] = name_h1.get_text(strip=True)
                
                # Extract all other information
                extract_all_player_info(soup, result)
                print(f" Success: {player_name}")
            else:
                result['status'] = 'Error: Not a player page'
                print(f" Not a player page: {player_name}")
        
        else:
            result['status'] = f'Error: HTTP {response.status_code}'
            print(f" HTTP Error {response.status_code}: {player_name}")
    
    except Exception as e:
        result['status'] = f'Error: {str(e)[:50]}'
        print(f" Exception: {player_name} - {e}")
    
    return result

def extract_all_player_info(soup, result):
    """Extract all information from player page"""
    try:
        meta_div = soup.find('div', {'id': 'meta'})
        if not meta_div:
            return
        
        meta_text = meta_div.get_text()
        
        # POSITION
        position_match = re.search(r'Position:\s*([^\n▪]+)', meta_text)
        if position_match:
            result['position'] = position_match.group(1).strip()
        
        # FOOT
        foot_match = re.search(r'Footed:\s*([^\n]+)', meta_text)
        if foot_match:
            result['foot'] = foot_match.group(1).strip()
        
        # HEIGHT
        height_match = re.search(r'(\d+cm|\d+,\d+\s*m|\d\'\d+\")', meta_text)
        if height_match:
            result['height'] = height_match.group(1).strip()
        
        # BIRTH INFO
        born_match = re.search(r'Born:\s*([^\(]+)\s*\(([^)]+)\)\s*in\s*([^\n]+)', meta_text)
        if born_match:
            result['birth_date'] = born_match.group(1).strip()
            result['age'] = born_match.group(2).strip()
            result['place_birth'] = born_match.group(3).strip()
        
        # CITIZENSHIP
        national_match = re.search(r'National Team:\s*([^\n]+)', meta_text)
        if national_match:
            citizenship = national_match.group(1).strip()
            citizenship = re.sub(r'\s+[a-z]{2}$', '', citizenship)
            result['citizenship'] = citizenship
        
        # CURRENT CLUB
        club_match = re.search(r'Club:\s*([^\n]+)', meta_text)
        if club_match:
            result['current_club'] = club_match.group(1).strip()
        
        # DEBUT INFO
        extract_debut_info_from_page(soup, result)
        
    except Exception as e:
        print(f"    Info extraction warning: {e}")

def extract_debut_info_from_page(soup, result):
    """Extract debut information from stats tables"""
    try:
        tables = soup.find_all('table', class_=re.compile('stats_table'))
        
        for table in tables:
            try:
                df = pd.read_html(io.StringIO(str(table)))[0]
                
                if 'Season' in str(df.columns) and len(df) > 1:
                    # Find first valid season row
                    for idx in range(len(df)):
                        season_val = str(df.iloc[idx, 0]) if len(df.columns) > 0 else ''
                        
                        if (season_val and season_val != 'nan' and 
                            season_val != '' and 'Season' not in season_val):
                            
                            result['debut_season'] = season_val
                            
                            if '-' in season_val:
                                result['debut_year'] = season_val.split('-')[0]
                            
                            # Find age
                            for col in df.columns:
                                if 'Age' in str(col):
                                    age_val = df.iloc[idx][col]
                                    if pd.notna(age_val):
                                        result['debut_age'] = str(age_val)
                                        break
                            
                            # Find first club
                            for col in df.columns:
                                if 'Squad' in str(col):
                                    club_val = df.iloc[idx][col]
                                    if pd.notna(club_val):
                                        result['first_club'] = str(club_val)
                                        break
                            
                            return
            except:
                continue
    except:
        pass

def process_players_parallel(players_with_urls, batch_size=50):
    """Process players in parallel batches"""
    all_results = []
    
    # Filter out players without URLs
    valid_players = [p for p in players_with_urls if p['player_url'] not in ['Not found', 'No Matches link', 'Not found in HTML', 'Not found in Matches']]
    invalid_players = [p for p in players_with_urls if p['player_url'] in ['Not found', 'No Matches link', 'Not found in HTML', 'Not found in Matches']]
    
    print(f" Players with valid URLs: {len(valid_players)}/{len(players_with_urls)}")
    print(f" Players without URLs: {len(invalid_players)}")
    
    if not valid_players:
        print(" No valid URLs found to process")
        return all_results + invalid_players
    
    # Process in batches
    for i in range(0, len(valid_players), batch_size):
        batch = valid_players[i:i + batch_size]
        print(f"\nProcessing batch {i//batch_size + 1}/{(len(valid_players)-1)//batch_size + 1}")
        print(f"Batch size: {len(batch)} players")
        
        # Process batch with parallelism
        with concurrent.futures.ThreadPoolExecutor(max_workers=MAX_WORKERS) as executor:
            batch_results = list(tqdm(
                executor.map(get_player_details, batch),
                total=len(batch),
                desc="Extracting player info"
            ))
        
        all_results.extend(batch_results)
        
        # Save progress after each batch
        progress_df = pd.DataFrame(all_results + invalid_players)
        progress_df.to_csv(f"progress_batch_{i//batch_size + 1}.csv", index=False)
        
        print(f" Saved progress batch {i//batch_size + 1}")
        
        # Respectful delay between batches
        if i + batch_size < len(valid_players):
            delay = 5
            print(f" Waiting {delay} seconds before next batch...")
            time.sleep(delay)
    
    return all_results + invalid_players

def main():
    """Main function using existing player URLs"""
    print(" DIRECT URL PLAYER DATA EXTRACTION")
    print("=" * 60)
    print(" Using existing player URLs from your table")
    print("=" * 60)
    
    # Extract player URLs from existing table
    players_with_urls = extract_player_urls_from_existing_table(INPUT_CSV)
    
    if not players_with_urls:
        print(" No player data extracted from table")
        return
    
    # Show sample of what we found
    print(f"\n URL EXTRACTION RESULTS:")
    url_stats = {}
    for player in players_with_urls:
        url_status = player['player_url']
        url_stats[url_status] = url_stats.get(url_status, 0) + 1
    
    for status, count in url_stats.items():
        print(f"   {status}: {count} players")
    
    # Process players
    print(f"\n Starting data extraction for {len(players_with_urls)} players...")
    all_results = process_players_parallel(players_with_urls, batch_size=50)
    
    # Save final results
    df_final = pd.DataFrame(all_results)
    df_final.to_csv(OUTPUT_CSV, index=False, encoding='utf-8')
    
    # Final summary
    successful = len([r for r in all_results if r.get('status') == 'Success'])
    no_urls = len([r for r in all_results if 'No valid URL' in str(r.get('status', '')) or r.get('player_url') in ['Not found', 'No Matches link']])
    errors = len([r for r in all_results if 'Error' in str(r.get('status', ''))])
    
    print(f"\n EXTRACTION COMPLETED!")
    print(f" FINAL RESULTS:")
    print(f"    Successful: {successful}/{len(all_results)}")
    print(f"    No URLs: {no_urls}/{len(all_results)}")
    print(f"    Errors: {errors}/{len(all_results)}")
    print(f"    Success rate: {(successful/len(all_results))*100:.1f}%")
    print(f" Saved to: {OUTPUT_CSV}")
    
    # Show sample of successful extractions
    successful_players = [r for r in all_results if r.get('status') == 'Success']
    if successful_players:
        print(f"\n SAMPLE OF EXTRACTED DATA:")
        for player in successful_players[:3]:
            print(f"   • {player.get('name_home_country', player['original_name'])}")
            print(f"     Position: {player.get('position', 'N/A')}")
            print(f"     Club: {player.get('current_club', 'N/A')}")
            print(f"     Age: {player.get('age', 'N/A')}")

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
import time
import re
from urllib.parse import urljoin
import os

# ========= CONFIGURATION =========
SEASONS = [
    "2010-2011", "2011-2012", "2012-2013", "2013-2014", "2014-2015",
    "2015-2016", "2016-2017", "2017-2018", "2018-2019", "2019-2020"
]

OUTPUT_CSV = "all_la_liga_players_urls.csv"

def setup_selenium_driver():
    """Setup Selenium WebDriver"""
    chrome_options = Options()
    chrome_options.add_argument("--headless=new")
    chrome_options.add_argument("--no-sandbox")
    chrome_options.add_argument("--disable-dev-shm-usage")
    chrome_options.add_argument("--disable-blink-features=AutomationControlled")
    chrome_options.add_experimental_option("excludeSwitches", ["enable-automation"])
    chrome_options.add_experimental_option('useAutomationExtension', False)
    chrome_options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36")
    
    driver = webdriver.Chrome(options=chrome_options)
    driver.execute_script("Object.defineProperty(navigator, 'webdriver', {get: () => undefined})")
    return driver

def extract_player_urls_from_page(driver, url, season):
    """Extract ALL player URLs from a FBRef stats page"""
    print(f"Extracting from: {season}")
    
    all_players = []
    
    try:
        driver.get(url)
        time.sleep(5)  # Wait for page to load
        
        # Save the page for debugging
        with open(f"debug_{season}.html", "w", encoding="utf-8") as f:
            f.write(driver.page_source)
        print(f" Saved page to debug_{season}.html")
        
        soup = BeautifulSoup(driver.page_source, 'html.parser')
        
        # STRATEGY 1: Look for the main stats table
        table = soup.find('table', {'id': 'stats_playing_time'})
        
        if table:
            print(" Found stats_playing_time table")
            players_from_table = extract_from_stats_table(table, season)
            all_players.extend(players_from_table)
        
        # STRATEGY 2: Look for ANY player links on the page
        all_player_links = soup.find_all('a', href=re.compile(r'/en/players/[a-f0-9]+/'))
        print(f" Found {len(all_player_links)} total player links on page")
        
        # Filter unique players from all links
        unique_players = extract_unique_players_from_links(all_player_links, season)
        all_players.extend(unique_players)
        
        # Remove duplicates based on player_url
        seen_urls = set()
        unique_players = []
        for player in all_players:
            if player['player_url'] not in seen_urls:
                seen_urls.add(player['player_url'])
                unique_players.append(player)
        
        print(f" Season {season}: Found {len(unique_players)} unique players")
        return unique_players
        
    except Exception as e:
        print(f" Error extracting from {season}: {e}")
        return []

def extract_from_stats_table(table, season):
    """Extract players from the stats table"""
    players = []
    
    try:
        # Find all rows in the table
        rows = table.find_all('tr')
        
        for row in rows:
            try:
                # Look for player cells - they can be th or td with player links
                player_cells = row.find_all(['th', 'td'])
                
                for cell in player_cells:
                    player_link = cell.find('a', href=re.compile(r'/en/players/[a-f0-9]+/'))
                    if player_link:
                        player_name = player_link.get_text(strip=True)
                        player_url = urljoin("https://fbref.com", player_link['href'])
                        
                        # Extract player ID
                        player_id_match = re.search(r'/players/([a-f0-9]+)/', player_url)
                        player_id = player_id_match.group(1) if player_id_match else "unknown"
                        
                        player_data = {
                            'player_name': player_name,
                            'player_url': player_url,
                            'player_id': player_id,
                            'season': season,
                            'source': 'stats_table'
                        }
                        
                        players.append(player_data)
                        break  # Only take one player per row
                        
            except Exception as e:
                continue
                
    except Exception as e:
        print(f"    Error parsing table: {e}")
    
    return players

def extract_unique_players_from_links(player_links, season):
    """Extract unique players from all player links on page"""
    players = []
    seen_urls = set()
    
    for link in player_links:
        try:
            player_url = urljoin("https://fbref.com", link['href'])
            
            # Skip if we've already seen this URL
            if player_url in seen_urls:
                continue
                
            player_name = link.get_text(strip=True)
            
            # Extract player ID
            player_id_match = re.search(r'/players/([a-f0-9]+)/', player_url)
            player_id = player_id_match.group(1) if player_id_match else "unknown"
            
            player_data = {
                'player_name': player_name,
                'player_url': player_url,
                'player_id': player_id,
                'season': season,
                'source': 'page_links'
            }
            
            players.append(player_data)
            seen_urls.add(player_url)
            
        except Exception as e:
            continue
    
    return players

def generate_season_urls():
    """Generate URLs for all seasons"""
    base_url = "https://fbref.com/en/comps/12/{season}/playingtime/{season}-La-Liga-Stats"
    
    season_urls = []
    for season in SEASONS:
        url = base_url.format(season=season)
        season_urls.append((season, url))
    
    return season_urls

def main():
    """Main function to extract ALL player URLs from multiple seasons"""
    print("🚀 EXTRACTING ALL LA LIGA PLAYER URLs")
    print("=" * 60)
    print(f"📅 Processing {len(SEASONS)} seasons: {SEASONS}")
    print("=" * 60)
    
    # Setup driver
    driver = setup_selenium_driver()
    all_players = []
    
    try:
        # Generate season URLs
        season_urls = generate_season_urls()
        
        for season, url in season_urls:
            print(f"\n Processing {season}...")
            print(f"   URL: {url}")
            
            # Extract player URLs from this season
            season_players = extract_player_urls_from_page(driver, url, season)
            all_players.extend(season_players)
            
            print(f" {season}: {len(season_players)} players found")
            
            # Progress update
            total_so_far = len(all_players)
            unique_so_far = len(set(p['player_url'] for p in all_players))
            print(f"Total so far: {total_so_far} records, {unique_so_far} unique players")
            
            # Save progress after each season
            df_progress = pd.DataFrame(all_players)
            df_progress.to_csv(f"progress_{season}.csv", index=False)
            print(f" Progress saved: progress_{season}.csv")
            
            # Delay between seasons
            if season != season_urls[-1][0]:
                delay = 10
                print(f" Waiting {delay} seconds before next season...")
                time.sleep(delay)
    
    except Exception as e:
        print(f" Main function error: {e}")
    finally:
        driver.quit()
        print("\n Selenium driver closed")
    
    # Final processing and saving
    if all_players:
        # Create final DataFrame
        df_final = pd.DataFrame(all_players)
        
        # Remove duplicates (keep first occurrence)
        df_final = df_final.drop_duplicates(subset=['player_url'], keep='first')
        
        # Add additional information
        df_final['seasons_appeared'] = df_final.groupby('player_url')['season'].transform('count')
        
        # Reorder columns
        column_order = ['player_name', 'player_url', 'player_id', 'season', 'seasons_appeared', 'source']
        df_final = df_final[column_order]
        
        # Save final results
        df_final.to_csv(OUTPUT_CSV, index=False, encoding='utf-8')
        
        # Generate summary
        generate_summary(df_final)
    else:
        print(" No players found!")

def generate_summary(df):
    """Generate extraction summary"""
    print("\n" + "=" * 60)
    print(" EXTRACTION SUMMARY")
    print("=" * 60)
    
    total_players = len(df)
    unique_players = df['player_url'].nunique()
    seasons_covered = df['season'].nunique()
    
    print(f" Total player records: {total_players}")
    print(f" Unique players: {unique_players}")
    print(f" Seasons covered: {seasons_covered}")
    
    # Players by season
    print(f"\n Players per season:")
    season_counts = df['season'].value_counts().sort_index()
    for season, count in season_counts.items():
        print(f"   {season}: {count} players")
    
    # Players with multiple seasons
    multi_season = df[df['seasons_appeared'] > 1]
    if not multi_season.empty:
        multi_season_count = multi_season['player_url'].nunique()
        print(f" Players in multiple seasons: {multi_season_count}")
    
    print(f"\n Final results saved to: {OUTPUT_CSV}")
    
    # Show sample of players
    print(f"\n SAMPLE PLAYERS (first 10):")
    sample_players = df.head(10)[['player_name', 'season', 'player_url']]
    for idx, row in sample_players.iterrows():
        print(f"   • {row['player_name']} ({row['season']})")
        print(f"     {row['player_url']}")

def analyze_debug_files():
    """Analyze the saved debug HTML files to understand the structure"""
    print("\n ANALYZING DEBUG FILES...")
    
    for season in SEASONS[:1]:  # Just analyze first season
        debug_file = f"debug_{season}.html"
        if os.path.exists(debug_file):
            print(f"\n📄 Analyzing {debug_file}")
            
            with open(debug_file, 'r', encoding='utf-8') as f:
                content = f.read()
            
            soup = BeautifulSoup(content, 'html.parser')
            
            # Count player links
            player_links = soup.find_all('a', href=re.compile(r'/en/players/[a-f0-9]+/'))
            print(f"   Total player links found: {len(player_links)}")
            
            # Check for tables
            tables = soup.find_all('table')
            print(f"   Total tables found: {len(tables)}")
            
            # Check specific table
            stats_table = soup.find('table', {'id': 'stats_playing_time'})
            if stats_table:
                print("    stats_playing_time table exists")
                rows = stats_table.find_all('tr')
                print(f"   Table has {len(rows)} rows")
            else:
                print("    stats_playing_time table not found")
            
            # Show first few player links
            print(f"   First 5 player links:")
            for link in player_links[:5]:
                print(f"     - {link.get_text(strip=True)}: {link['href']}")

if __name__ == "__main__":
    main()
    
    # After main execution, analyze debug files if needed
    print("\n" + "=" * 60)
    print("🔧 DEBUG ANALYSIS")
    print("=" * 60)
    analyze_debug_files()
```

```python
# ========= CONFIGURATION =========
INPUT_CSV = "players_liga3.csv"
OUTPUT_CSV = "players_liga_scrap3.csv"

def setup_selenium_driver():
    """Setup Selenium WebDriver optimizado"""
    chrome_options = Options()
    chrome_options.add_argument("--headless")
    chrome_options.add_argument("--no-sandbox")
    chrome_options.add_argument("--disable-dev-shm-usage")
    chrome_options.add_argument("--disable-blink-features=AutomationControlled")
    chrome_options.add_experimental_option("excludeSwitches", ["enable-automation"])
    chrome_options.add_experimental_option('useAutomationExtension', False)
    chrome_options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36")
    
    driver = webdriver.Chrome(options=chrome_options)
    driver.execute_script("Object.defineProperty(navigator, 'webdriver', {get: () => undefined})")
    return driver

def extract_all_player_info(driver, player_url, player_name, player_slug):
    """Extrae TODA la información del jugador de una sola vez"""
    print(f" Extracting ALL info: {player_name}")
    
    player_info = {
        'player_name': player_name,
        'player_slug': player_slug,
        'player_url': player_url,
        'status': 'Success'
    }
    
    # Inicializar todos los campos
    fields = [
        'name_home_country', 'dob_age', 'place_birth', 'height', 
        'citizenship', 'position', 'foot', 'agent', 'current_club',
        'joined', 'contract_expires', 'last_contract_ext', 'outfitter',
        'debut_season', 'debut_age', 'first_club', 'debut_year'
    ]
    
    for field in fields:
        player_info[field] = 'Not found'
    
    try:
        if player_url == 'Not found' or not player_url:
            player_info['status'] = 'No URL'
            return player_info
        
        # Cargar página con Selenium
        driver.get(player_url)
        time.sleep(3)  # Esperar a que cargue
        
        soup = BeautifulSoup(driver.page_source, 'html.parser')
        
        # ========= EXTRACT META INFORMATION =========
        meta_div = soup.find('div', {'id': 'meta'})
        if meta_div:
            extract_meta_info_corrected(meta_div, player_info)
        
        # ========= EXTRACT DEBUT INFO =========
        extract_debut_info_complete(soup, player_info)
        
        print(f"   ✅ Completed: {player_name}")
        return player_info
        
    except Exception as e:
        print(f"    Error: {player_name} - {str(e)[:100]}")
        player_info['status'] = f'Error: {str(e)[:100]}'
        return player_info

def extract_meta_info_corrected(meta_div, player_info):
    """Extrae información CORREGIDA del div meta"""
    meta_text = meta_div.get_text()
    
    print(f"    Meta text preview: {meta_text[:500]}...")
    
    # ========= EXTRACT POSITION =========
    position_match = re.search(r'Position:\s*([^\n▪]+)', meta_text)
    if position_match:
        player_info['position'] = position_match.group(1).strip()
        print(f"    position: {player_info['position']}")
    
    # ========= EXTRACT FOOT =========
    foot_match = re.search(r'Footed:\s*([^\n]+)', meta_text)
    if foot_match:
        player_info['foot'] = foot_match.group(1).strip()
        print(f"    foot: {player_info['foot']}")
    
    # ========= EXTRACT HEIGHT =========
    # Buscar patrones de altura: "180cm" o "1,80 m"
    height_match = re.search(r'(\d+cm|\d+,\d+\s*m)', meta_text)
    if height_match:
        player_info['height'] = height_match.group(1).strip()
        print(f"    height: {player_info['height']}")
    
    # ========= EXTRACT BIRTH INFO (CORREGIDO) =========
    # Buscar "Born: July 13, 2007 (Age: 18-092d) in Mataró, Spain"
    born_match = re.search(r'Born:\s*([^\(]+)\s*\(([^)]+)\)\s*in\s*([^\n]+)', meta_text)
    if born_match:
        player_info['dob_age'] = f"{born_match.group(1).strip()} ({born_match.group(2).strip()})"
        player_info['place_birth'] = born_match.group(3).strip()
        print(f"    dob_age: {player_info['dob_age']}")
        print(f"    place_birth: {player_info['place_birth']}")
    else:
        # Fallback: buscar solo fecha de nacimiento
        born_simple = re.search(r'Born:\s*([^\n\(]+)', meta_text)
        if born_simple:
            player_info['dob_age'] = born_simple.group(1).strip()
            print(f"    dob_age (simple): {player_info['dob_age']}")
    
    # ========= EXTRACT CITIZENSHIP/NATIONAL TEAM (CORREGIDO) =========
    # Buscar "National Team: Spain es"
    national_match = re.search(r'National Team:\s*([^\n]+)', meta_text)
    if national_match:
        citizenship_text = national_match.group(1).strip()
        # Limpiar el texto (remover códigos de país como "es")
        citizenship_clean = re.sub(r'\s+[a-z]{2}$', '', citizenship_text)
        player_info['citizenship'] = citizenship_clean
        print(f"    citizenship: {player_info['citizenship']}")
    
    # ========= EXTRACT CURRENT CLUB =========
    club_match = re.search(r'Club:\s*([^\n]+)', meta_text)
    if club_match:
        player_info['current_club'] = club_match.group(1).strip()
        print(f"    current_club: {player_info['current_club']}")
    
    # ========= EXTRACT CONTRACT INFO =========
    # Buscar información de contrato como "Expires June 2031"
    contract_match = re.search(r'Expires\s*([^\n]+)', meta_text)
    if contract_match:
        player_info['contract_expires'] = contract_match.group(1).strip()
        print(f"    contract_expires: {player_info['contract_expires']}")
    
    # ========= EXTRACT NAME_HOME_COUNTRY =========
    # Buscar el nombre completo con país: "Lamine Yamal Nasraoui Ebana" + país
    name_match = re.search(r'^([A-Z][a-z]+(?:\s+[A-Z][a-z]+))\s\(([^)]+)\)', meta_text)
    if not name_match:
        # Fallback: buscar cualquier texto que parezca nombre completo al inicio
        name_match = re.search(r'^([A-Z][a-z]+(?:\s+[A-Z][a-z]+)+)', meta_text)
    
    if name_match:
        player_info['name_home_country'] = name_match.group(1).strip()
        print(f"    name_home_country: {player_info['name_home_country']}")
    
    # ========= EXTRACT AGENT (si está disponible) =========
    agent_match = re.search(r'Agent:\s*([^\n]+)', meta_text)
    if agent_match:
        player_info['agent'] = agent_match.group(1).strip()
        print(f"    agent: {player_info['agent']}")

def extract_debut_info_complete(soup, player_info):
    """Extrae información de debut de las tablas"""
    try:
        # Buscar todas las tablas
        tables = soup.find_all('table')
        
        for table in tables:
            try:
                table_html = str(table)
                df = pd.read_html(io.StringIO(table_html))[0]
                
                # Buscar tabla con Season y Age
                has_season = any('Season' in str(col) for col in df.columns)
                has_age = any('Age' in str(col) for col in df.columns)
                
                if has_season and has_age and len(df) > 1:
                    season_col = next((col for col in df.columns if 'Season' in str(col)), None)
                    age_col = next((col for col in df.columns if 'Age' in str(col)), None)
                    squad_col = next((col for col in df.columns if 'Squad' in str(col)), None)
                    
                    if season_col:
                        # Tomar la primera fila válida
                        for idx in range(len(df)):
                            row = df.iloc[idx]
                            season_val = str(row[season_col])
                            
                            # Saltar filas de encabezado
                            if (season_val and season_val != 'nan' and 
                                season_val != '' and 'Season' not in season_val):
                                
                                player_info['debut_season'] = season_val
                                if isinstance(season_val, str) and '-' in season_val:
                                    player_info['debut_year'] = season_val.split('-')[0]
                                
                                if age_col and age_col in row:
                                    age_val = row[age_col]
                                    if pd.notna(age_val) and age_val != '':
                                        player_info['debut_age'] = f"{age_val} years"
                                
                                if squad_col and squad_col in row:
                                    club_val = row[squad_col]
                                    if pd.notna(club_val) and club_val != '':
                                        player_info['first_club'] = club_val
                                
                                print(f"   🎯 Debut: {player_info['debut_season']} at {player_info['debut_age']}")
                                break
                        break
            except:
                continue
    except:
        pass

def main():
    """Procesa todos los jugadores con Selenium"""
    print(" COMPLETE PLAYER INFO EXTRACTION WITH SELENIUM")
    print("=" * 60)
    
    try:
        df_existing = pd.read_csv(INPUT_CSV)
        print(f" Loaded {len(df_existing)} players with URLs")
        
        players_to_process = df_existing[
            (df_existing['player_url'] != 'Not found') & 
            (df_existing['player_url'].notna())
        ]
        
        print(f"🔧 Players to process: {len(players_to_process)}")
        
    except Exception as e:
        print(f" Error reading CSV: {e}")
        return
    
    # Setup Selenium driver
    driver = setup_selenium_driver()
    all_results = []
    
    try:
        # Procesar todos los jugadores
        for index, row in players_to_process.iterrows():
            player_name = str(row.get('player_name', '')).strip()
            player_slug = str(row.get('player_slug', '')).strip()
            player_url = str(row.get('player_url', '')).strip()
            
            print(f"\n{index + 1}/{len(players_to_process)}: {player_name}")
            
            # Extraer toda la información
            player_data = extract_all_player_info(driver, player_url, player_name, player_slug)
            all_results.append(player_data)
            
            # Progreso cada 10 jugadores
            if (index + 1) % 10 == 0:
                successful = len([r for r in all_results if r['status'] == 'Success'])
                print(f"\n Progress: {index + 1}/{len(players_to_process)}, {successful} successful")
            
            # Delay aleatorio
            time.sleep(random.uniform(2, 4))
    
    except KeyboardInterrupt:
        print("\n Process interrupted by user")
    except Exception as e:
        print(f"\n Unexpected error: {e}")
    finally:
        driver.quit()
        print("\n Selenium driver closed")
    
    # Guardar resultados
    df_final = pd.DataFrame(all_results)
    df_final.to_csv(OUTPUT_CSV, index=False, encoding='utf-8')
    
    # Resumen final
    print("\n" + "=" * 60)
    print(" FINAL RESULTS")
    print("=" * 60)
    
    total = len(all_results)
    successful = len([r for r in all_results if r['status'] == 'Success'])
    
    # Estadísticas por campo
    fields_to_check = ['position', 'dob_age', 'height', 'current_club', 'citizenship', 'place_birth', 'debut_season']
    field_stats = {}
    
    for field in fields_to_check:
        count = len([r for r in all_results if r.get(field, 'Not found') != 'Not found'])
        field_stats[field] = count
    
    print(f" Total processed: {total}")
    print(f" Successful: {successful}")
    print(f"\n Fields extracted:")
    for field, count in field_stats.items():
        print(f"   {field}: {count}/{total} ({count/total*100:.1f}%)")
    
    print(f"💾 Saved to: {OUTPUT_CSV}")
    
    # Mostrar ejemplos exitosos
    successful_players = [r for r in all_results if r['status'] == 'Success']
    if successful_players:
        sample_df = pd.DataFrame(successful_players[:3])
        columns_to_show = ['player_name', 'position', 'dob_age', 'citizenship', 'place_birth', 'current_club']
        available_columns = [col for col in columns_to_show if col in sample_df.columns]
        print(f"\n Sample results:")
        print(sample_df[available_columns].to_string(index=False))

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import re
import time
import random

# ========= CONFIGURATION =========
URLS_CSV = "players_transfermarkt_urls.csv"
DATA_CSV = "players_transfermarkt_complete_data.csv"
BATCH_SIZE = 5  # Process in small batches

HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/141.0.0.0 Safari/537.36"
}

def extract_detailed_player_data(player_url, player_name):
    """Extract all detailed player information from Transfermarkt"""
    print(f" Extracting data: {player_name}")
    
    player_info = {
        'player_name': player_name,
        'player_url': player_url,
        'status': 'Success'
    }
    
    # Initialize all target fields
    target_fields = [
        'player_id', 'player_slug', 'home_name', 'dob_age', 'place_birth',
        'height', 'citizenship', 'position', 'foot', 'agent', 'current_club',
        'joined', 'contract_expires', 'outfitter', 'injury_count', 'days_missed',
        'retired', 'retirement_date', 'market_value', 'shirt_number'
    ]
    
    for field in target_fields:
        player_info[field] = 'Not found'
    
    try:
        if player_url == 'Not found':
            player_info['status'] = 'No URL'
            return player_info
        
        # Extract player ID from URL
        player_id = player_url.split('/')[-1]
        player_info['player_id'] = player_id
        player_info['player_slug'] = player_url.split('/')[-2]
        
        response = requests.get(player_url, headers=HEADERS)
        if response.status_code != 200:
            player_info['status'] = f'HTTP Error: {response.status_code}'
            return player_info
        
        soup = BeautifulSoup(response.content, "html.parser")
        
        # ========= BASIC INFORMATION =========
        
        # Full name
        name_element = soup.select_one('h1[class="data-header__headline-wrapper"]')
        if name_element:
            full_name = name_element.text.split('\n')[-1].strip()
            player_info['home_name'] = full_name
            print(f"    Name: {full_name}")
        
        # Shirt number
        number_element = soup.select_one('span[class="data-header__shirt-number"]')
        if number_element:
            player_info['shirt_number'] = number_element.text.replace('#', '').strip()
            print(f"    Number: {player_info['shirt_number']}")
        
        # Current club
        club_element = soup.select_one('span[class="data-header__club"]')
        if club_element:
            player_info['current_club'] = club_element.text.split('\n')[-1].strip()
            print(f"    Club: {player_info['current_club']}")
        
        # ========= DETAILED DATA GRID =========
        
        data_grid = soup.find('div', class_='info-table')
        if data_grid:
            grid_text = data_grid.get_text()
            
            # Extract all fields using regex patterns
            field_patterns = {
                'dob_age': r'Age:\s*([^\n]+)',
                'place_birth': r'Place of birth:\s*([^\n]+)',
                'height': r'Height:\s*([^\n]+)',
                'citizenship': r'Citizenship:\s*([^\n]+)',
                'position': r'Position:\s*([^\n]+)',
                'foot': r'Foot:\s*([^\n]+)',
                'agent': r'Agent:\s*([^\n]+)',
                'joined': r'Joined:\s*([^\n]+)',
                'contract_expires': r'Contract expires:\s*([^\n]+)',
                'outfitter': r'Outfitter:\s*([^\n]+)'
            }
            
            for field, pattern in field_patterns.items():
                match = re.search(pattern, grid_text)
                if match:
                    player_info[field] = match.group(1).strip()
                    print(f"    {field}: {player_info[field]}")
        
        # ========= MARKET VALUE =========
        value_element = soup.find('div', class_='tm-player-market-value-development__current-value')
        if value_element:
            player_info['market_value'] = value_element.text.strip()
            print(f"    Market Value: {player_info['market_value']}")
        
        # ========= RETIREMENT STATUS =========
        page_text = soup.get_text()
        if 'Retired' in page_text or 'retired' in page_text:
            player_info['retired'] = 'Yes'
            retirement_match = re.search(r'Retired[^.]*?(\d{1,2}\s+\w+\s+\d{4})', page_text, re.IGNORECASE)
            if retirement_match:
                player_info['retirement_date'] = retirement_match.group(1)
            print(f"    Retired: {player_info['retired']}")
        else:
            player_info['retired'] = 'No'
            print(f"    Active player")
        
        print(f"    Completed: {player_name}")
        return player_info
        
    except Exception as e:
        print(f"    Error: {str(e)[:100]}")
        player_info['status'] = f'Error: {str(e)[:100]}'
        return player_info

def extract_data_from_urls():
    """Second pass: Extract detailed data from collected URLs"""
    print(" STEP 2: EXTRACTING DETAILED PLAYER DATA")
    print("=" * 60)
    
    try:
        df_urls = pd.read_csv(URLS_CSV)
        print(f" Loaded {len(df_urls)} players with URLs")
        
        # Filter only players with valid URLs
        valid_players = df_urls[df_urls['transfermarkt_url'] != 'Not found']
        print(f" Processing {len(valid_players)} players with valid URLs")
        
        # Process in batches
        all_results = []
        
        for index, row in valid_players.iterrows():
            player_name = row['player_name']
            player_url = row['transfermarkt_url']
            
            print(f"\n[{index + 1}/{len(valid_players)}] {player_name}")
            print(f"   URL: {player_url}")
            
            # Extract detailed data
            player_data = extract_detailed_player_data(player_url, player_name)
            all_results.append(player_data)
            
            # Delay between requests
            if index + 1 < len(valid_players):
                delay = random.uniform(3, 6)
                print(f"⏳ Waiting {delay:.1f} seconds...")
                time.sleep(delay)
            
            # Save progress every BATCH_SIZE players
            if (index + 1) % BATCH_SIZE == 0:
                df_temp = pd.DataFrame(all_results)
                df_temp.to_csv(f"progress_batch_{(index + 1)//BATCH_SIZE}.csv", index=False)
                print(f"💾 Progress saved: batch {(index + 1)//BATCH_SIZE}")
        
        # Save final results
        df_final = pd.DataFrame(all_results)
        df_final.to_csv(DATA_CSV, index=False, encoding='utf-8')
        
        # Summary
        successful = len([r for r in all_results if r['status'] == 'Success'])
        print(f"\n📊 DATA EXTRACTION SUMMARY:")
        print(f"   Total processed: {len(all_results)}")
        print(f"   Successful: {successful}")
        print(f"   Success rate: {(successful/len(all_results))*100:.1f}%")
        print(f"💾 Data saved to: {DATA_CSV}")
        
    except Exception as e:
        print(f"❌ Error in data extraction: {e}")

if __name__ == "__main__":
    extract_data_from_urls()
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import re
import time
import random

# ========= CONFIGURATION =========
URLS_CSV = "players_transfermarkt_urls.csv"
DATA_CSV = "players_transfermarkt_cleaned.csv"
PROCESS_SIZE = 500

HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/141.0.0.0 Safari/537.36"
}

def fix_encoding(text):
    """Fix encoding issues in player names"""
    if not text or text == 'Not found':
        return text
    
    # Common encoding fixes
    encoding_fixes = {
        'Ãƒ': '', 'â€': '', 'Ã³': 'ó', 'Ã©': 'é', 'Ã': 'í', 
        'Ã±': 'ñ', 'Ãº': 'ú', 'Ã¡': 'á', 'Ã¨': 'è', 'Ãª': 'ê',
        'Ã§': 'ç', 'Ã£': 'ã', 'Ãµ': 'õ', 'Ã¶': 'ö', 'Ã¼': 'ü',
        'Ã¯': 'ï', 'Ã«': 'ë', 'Ã¬': 'ì', 'Ã': 'à'
    }
    
    fixed_text = text
    for wrong, correct in encoding_fixes.items():
        fixed_text = fixed_text.replace(wrong, correct)
    
    return fixed_text

def extract_injury_data_correct_table(player_url, player_name):
    """Extract injury data from the SEASON SUMMARY table"""
    print(f"🏥 Extracting injury data: {player_name}")
    
    injury_data = {
        'injury_count_total': 0,
        'days_missed_total': 0,
        'games_missed_total': 0,
        'injury_seasons': 0
    }
    
    try:
        # Construct injury URL from player URL
        injury_url = player_url.replace('/profil/', '/verletzungen/')
        
        response = requests.get(injury_url, headers=HEADERS)
        if response.status_code != 200:
            return injury_data
        
        soup = BeautifulSoup(response.content, "html.parser")
        
        # Look for ALL tables with class 'items'
        all_tables = soup.find_all('table', {'class': 'items'})
        
        # Find the summary table
        summary_table = None
        for table in all_tables:
            header = table.get_text()
            if 'Season' in header and 'Days' in header and 'Injuries' in header and 'Games missed' in header:
                summary_table = table
                break
        
        if not summary_table:
            return injury_data
        
        # Find the tbody within the summary table
        tbody = summary_table.find('tbody')
        if not tbody:
            return injury_data
        
        # Extract all rows from tbody
        rows = tbody.find_all('tr')
        
        total_injuries = 0
        total_days = 0
        total_games_missed = 0
        
        for row in rows:
            cells = row.find_all('td')
            if len(cells) >= 4:
                days_text = cells[1].get_text(strip=True)
                injuries = cells[2].get_text(strip=True)
                games_missed = cells[3].get_text(strip=True)
                
                # Parse injuries count
                injuries_num = int(injuries) if injuries.isdigit() else 0
                
                # Parse games missed
                games_missed_num = int(games_missed) if games_missed.isdigit() else 0
                
                # Parse days
                days_num = 0
                if days_text:
                    days_match = re.search(r'(\d+)', days_text)
                    if days_match:
                        days_num = int(days_match.group(1))
                
                total_injuries += injuries_num
                total_days += days_num
                total_games_missed += games_missed_num
        
        injury_data.update({
            'injury_count_total': total_injuries,
            'days_missed_total': total_days,
            'games_missed_total': total_games_missed,
            'injury_seasons': len(rows)
        })
        
        print(f"   💊 Injuries: {total_injuries} total, {total_days} days, {total_games_missed} games")
        
        return injury_data
        
    except Exception as e:
        print(f"   ⚠️ Injury extraction error: {e}")
        return injury_data

def extract_retirement_info(soup, player_info):
    """Extract retirement information for retired players - FIXED VERSION"""
    try:
        # Method 1: Check if the club name is "Retired"
        club_element = soup.select_one('span[class="data-header__club"]')
        if club_element:
            club_text = club_element.get_text(strip=True)
            if 'Retired' in club_text:
                player_info['retired'] = 'Yes'
                print(f"   🏁 Player is retired (Method 1)")
                
                # Extract retirement date using the label
                retirement_labels = soup.find_all('span', class_='data-header__label')
                for label in retirement_labels:
                    label_text = label.get_text(strip=True)
                    if 'Retired since' in label_text:
                        retirement_date_span = label.find_next('span', class_='data-header__content')
                        if retirement_date_span:
                            player_info['retirement_date'] = retirement_date_span.get_text(strip=True)
                            print(f"   📅 Retirement Date: {player_info['retirement_date']}")
                        break
                
                # Extract last club
                for label in retirement_labels:
                    label_text = label.get_text(strip=True)
                    if 'Last club' in label_text:
                        last_club_link = label.find_next('a')
                        if last_club_link:
                            player_info['current_club'] = last_club_link.get_text(strip=True)
                            print(f"   ⚽ Last Club: {player_info['current_club']}")
                        break
                return player_info
        
        # Method 2: Look for "Retired since" directly in all labels
        retirement_labels = soup.find_all('span', class_='data-header__label')
        for label in retirement_labels:
            label_text = label.get_text(strip=True)
            if 'Retired since' in label_text:
                player_info['retired'] = 'Yes'
                print(f"   🏁 Player is retired (Method 2)")
                
                # Extract retirement date
                retirement_date_span = label.find_next('span', class_='data-header__content')
                if retirement_date_span:
                    player_info['retirement_date'] = retirement_date_span.get_text(strip=True)
                    print(f"   📅 Retirement Date: {player_info['retirement_date']}")
                
                # Extract last club
                for label2 in retirement_labels:
                    label_text2 = label2.get_text(strip=True)
                    if 'Last club' in label_text2:
                        last_club_link = label2.find_next('a')
                        if last_club_link:
                            player_info['current_club'] = last_club_link.get_text(strip=True)
                            print(f"   ⚽ Last Club: {player_info['current_club']}")
                        break
                return player_info
        
        return player_info
        
    except Exception as e:
        print(f"   ⚠️ Retirement info error: {e}")
        return player_info

def extract_detailed_player_data(player_url, player_name):
    """Extract all detailed player information from Transfermarkt"""
    print(f"📊 Extracting data: {player_name}")
    
    player_info = {
        'player_name': fix_encoding(player_name),  # Fix encoding for player name
        'player_url': player_url,
        'status': 'Success'
    }
    
    # Initialize only the fields we want
    target_fields = [
        'player_id', 'player_slug', 'home_name', 'dob_age', 'place_birth',
        'height', 'citizenship', 'position', 'foot', 'agent', 'current_club',
        'joined', 'contract_expires', 'outfitter', 'injury_count_total', 
        'days_missed_total', 'games_missed_total', 'retired', 'retirement_date',
        'shirt_number'
    ]
    
    for field in target_fields:
        player_info[field] = 'Not found'
    
    try:
        if player_url == 'Not found':
            player_info['status'] = 'No URL'
            return player_info
        
        # Extract player ID from URL
        player_id = player_url.split('/')[-1]
        player_info['player_id'] = player_id
        player_info['player_slug'] = player_url.split('/')[-2]
        
        response = requests.get(player_url, headers=HEADERS)
        if response.status_code != 200:
            player_info['status'] = f'HTTP Error: {response.status_code}'
            return player_info
        
        soup = BeautifulSoup(response.content, "html.parser")
        
        # ========= BASIC INFORMATION =========
        
        # Full name (with encoding fix)
        name_element = soup.select_one('h1[class="data-header__headline-wrapper"]')
        if name_element:
            full_name = name_element.text.split('\n')[-1].strip()
            player_info['home_name'] = fix_encoding(full_name)
            print(f"   👤 Name: {player_info['home_name']}")
        
        # Shirt number
        number_element = soup.select_one('span[class="data-header__shirt-number"]')
        if number_element:
            player_info['shirt_number'] = number_element.text.replace('#', '').strip()
            print(f"   🔢 Number: {player_info['shirt_number']}")
        
        # ========= RETIREMENT INFORMATION =========
        # Check retirement status first, as it affects current_club
        player_info = extract_retirement_info(soup, player_info)
        
        # ========= CURRENT CLUB (only if not retired) =========
        if player_info.get('retired') != 'Yes':
            club_element = soup.select_one('span[class="data-header__club"]')
            if club_element:
                club_text = club_element.text.split('\n')[-1].strip()
                if club_text and club_text != 'Retired':
                    player_info['current_club'] = fix_encoding(club_text)
                    print(f"   ⚽ Club: {player_info['current_club']}")
        
        # ========= DETAILED DATA GRID =========
        
        data_grid = soup.find('div', class_='info-table')
        if data_grid:
            grid_text = data_grid.get_text()
            
            # Extract all fields using regex patterns
            field_patterns = {
                'dob_age': r'Age:\s*([^\n]+)',
                'place_birth': r'Place of birth:\s*([^\n]+)',
                'height': r'Height:\s*([^\n]+)',
                'citizenship': r'Citizenship:\s*([^\n]+)',
                'position': r'Position:\s*([^\n]+)',
                'foot': r'Foot:\s*([^\n]+)',
                'agent': r'Agent:\s*([^\n]+)',
                'joined': r'Joined:\s*([^\n]+)',
                'contract_expires': r'Contract expires:\s*([^\n]+)',
                'outfitter': r'Outfitter:\s*([^\n]+)'
            }
            
            for field, pattern in field_patterns.items():
                match = re.search(pattern, grid_text)
                if match:
                    player_info[field] = fix_encoding(match.group(1).strip())
                    print(f"   📍 {field}: {player_info[field]}")
        
        # Set default retirement status if not found
        if player_info['retired'] == 'Not found':
            player_info['retired'] = 'No'
            print(f"   🔄 Active player")
        
        # ========= INJURY DATA =========
        injury_data = extract_injury_data_correct_table(player_url, player_name)
        player_info.update(injury_data)
        
        print(f"   ✅ Completed: {player_info['home_name']}")
        return player_info
        
    except Exception as e:
        print(f"   ❌ Error: {str(e)[:100]}")
        player_info['status'] = f'Error: {str(e)[:100]}'
        return player_info

def extract_data_sample():
    """Test with a few players"""
    print("🚀 EXTRACTING PLAYER DATA (FIXED RETIREMENT DETECTION)")
    print("=" * 60)
    
    try:
        df_urls = pd.read_csv(URLS_CSV)
        print(f"📁 Loaded {len(df_urls)} players with URLs")
        
        # Filter only players with valid URLs
        valid_players = df_urls[df_urls['transfermarkt_url'] != 'Not found']
        print(f"🔧 Found {len(valid_players)} players with valid URLs")
        
        # Take only PROCESS_SIZE players
        sample_players = valid_players.head(PROCESS_SIZE)
        print(f"🎯 Processing {len(sample_players)} sample players:")
        
        for i, (idx, row) in enumerate(sample_players.iterrows(), 1):
            print(f"   {i}. {row['player_name']}")
        
        all_results = []
        
        for index, row in sample_players.iterrows():
            player_name = row['player_name']
            player_url = row['transfermarkt_url']
            
            print(f"\n{'='*50}")
            print(f"[{index + 1}/{len(sample_players)}] {player_name}")
            print(f"{'='*50}")
            
            # Extract detailed data
            player_data = extract_detailed_player_data(player_url, player_name)
            all_results.append(player_data)
            
            # Delay between requests
            if index + 1 < len(sample_players):
                delay = random.uniform(4, 7)
                print(f"⏳ Waiting {delay:.1f} seconds...")
                time.sleep(delay)
        
        # Save results
        df_final = pd.DataFrame(all_results)
        df_final.to_csv(DATA_CSV, index=False, encoding='utf-8')
        
        # Summary
        successful = len([r for r in all_results if r['status'] == 'Success'])
        retired_count = len([r for r in all_results if r.get('retired') == 'Yes'])
        
        print(f"\n📊 EXTRACTION SUMMARY:")
        print(f"   Total processed: {len(all_results)}")
        print(f"   Successful: {successful}")
        print(f"   Retired players found: {retired_count}")
        
        # Show cleaned data
        print(f"\n🎯 CLEANED DATA EXTRACTED:")
        for player in all_results:
            if player['status'] == 'Success':
                print(f"\n• {player['home_name']}")
                print(f"  Position: {player.get('position', 'N/A')}")
                print(f"  Club/Status: {player.get('current_club', 'N/A')}")
                print(f"  Retired: {player.get('retired', 'N/A')}")
                if player.get('retired') == 'Yes':
                    print(f"  Retirement Date: {player.get('retirement_date', 'N/A')}")
                print(f"  Injuries: {player.get('injury_count_total', 'N/A')}")
                print(f"  Days Missed: {player.get('days_missed_total', 'N/A')}")
                print(f"  Games Missed: {player.get('games_missed_total', 'N/A')}")
        
        print(f"\n💾 Cleaned data saved to: {DATA_CSV}")
        
    except Exception as e:
        print(f"❌ Error: {e}")

if __name__ == "__main__":
    extract_data_sample()
```

```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from bs4 import BeautifulSoup
import re
import time
import random
import io
import os

# ========= CONFIGURATION =========
INPUT_CSV = "scrap detalle carreras/National Selection/player_liga1.csv"
OUTPUT_CSV = "players_national1.csv"

def setup_stealth_driver():
    """Setup Chrome driver with stealth options"""
    options = Options()
    
    # Remove headless for debugging
    # options.add_argument("--headless")
    
    # Basic options
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    
    # Stealth options
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_experimental_option("excludeSwitches", ["enable-automation"])
    options.add_experimental_option('useAutomationExtension', False)
    
    # Realistic settings
    options.add_argument("--window-size=1920,1080")
    options.add_argument("--disable-extensions")
    options.add_argument("--disable-plugins")
    
    # User agent
    user_agent = "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
    options.add_argument(f"user-agent={user_agent}")
    
    try:
        driver = webdriver.Chrome(options=options)
        
        # Remove webdriver property
        driver.execute_script("Object.defineProperty(navigator, 'webdriver', {get: () => undefined})")
        
        return driver
    except Exception as e:
        print(f"❌ Error setting up driver: {e}")
        return None

def convert_to_national_team_url(player_url):
    """Convierte la URL normal a URL de selección nacional"""
    if 'Not found' in player_url or not player_url:
        return player_url
    
    if '/nat_tm/' in player_url:
        return player_url
    
    try:
        parts = player_url.split('/')
        player_id = parts[5]
        player_slug = parts[6]
        
        # Remove any existing suffix and add national team suffix
        if '-Stats' in player_slug:
            base_slug = player_slug.split('-Stats')[0]
        else:
            base_slug = player_slug
        
        national_team_url = f"https://fbref.com/en/players/{player_id}/nat_tm/{base_slug}-National-Team-Stats"
        return national_team_url
    except Exception as e:
        print(f"   Error converting to national team URL: {e}")
        return player_url

def extract_national_team_data(driver, player_url, player_name, player_slug):
    """Extrae datos de selección nacional"""
    print(f" Extracting national team history: {player_name}")
    
    career_data = {
        'player_name': player_name,
        'player_slug': player_slug,
        'player_url': player_url,
        'has_national_team': 'No',
        'total_seasons_national': 0,
        'total_matches_national': 0,
        'total_goals_national': 0,
        'total_assists_national': 0,
        'national_team': 'None',
        'national_team_years': '',
        'status': 'Success'
    }
    
    all_seasons = []
    
    try:
        if player_url == 'Not found' or not player_url:
            career_data['status'] = 'No URL'
            return career_data, all_seasons
        
        national_team_url = convert_to_national_team_url(player_url)
        print(f"    Using national team URL: {national_team_url}")
        
        # Navigate to the page
        driver.get(national_team_url)
        time.sleep(random.uniform(3, 6))
        
        # Check if we got blocked
        page_source = driver.page_source
        if "Just a moment" in page_source or "Cloudflare" in page_source:
            print("    Cloudflare detected, waiting longer...")
            time.sleep(15)
            page_source = driver.page_source
        
        # Check if page loaded successfully
        if "Page Not Found" in page_source or "404" in driver.title:
            print(f"    Page not found (404)")
            career_data['status'] = 'Page not found'
            return career_data, all_seasons
        
        # Parse the page
        soup = BeautifulSoup(page_source, 'html.parser')
        
        # Look for national team tables
        national_table = find_national_team_table_only(soup)
        
        if national_table:
            seasons_processed = process_national_team_table(national_table, player_name, player_url, all_seasons)
            if seasons_processed:
                return update_career_data(career_data, seasons_processed, all_seasons)
            else:
                career_data['status'] = 'No valid seasons'
        else:
            print(f"    No national team table found")
            career_data['status'] = 'No table found'
        
        return career_data, all_seasons
        
    except Exception as e:
        print(f"    Error: {player_name} - {str(e)}")
        career_data['status'] = f'Error: {str(e)[:100]}'
        return career_data, []

def find_national_team_table_only(soup):
    """Find ONLY national team tables"""
    
    # Method 1: Look for standard stats national team table by exact ID
    table = soup.find('table', id='stats_standard_nat_tm')
    if table:
        print("    Found NATIONAL TEAM table by exact ID: stats_standard_nat_tm")
        return table
    
    # Method 2: Look for tables with national team in parent context
    all_tables = soup.find_all('table')
    for table in all_tables:
        parent = table.find_parent()
        while parent:
            parent_html = str(parent)
            if 'National Team' in parent_html and 'Standard Stats' in parent_html:
                print("    Found NATIONAL TEAM table by parent context")
                return table
            parent = parent.find_parent()
    
    return None

def update_career_data(career_data, seasons_processed, all_seasons):
    """Update career data with national team information"""
    career_data['has_national_team'] = 'Yes'
    career_data['total_seasons_national'] = len(seasons_processed)
    
    # Calculate totals
    total_matches = 0
    total_goals = 0
    total_assists = 0
    for season in seasons_processed:
        if season['matches_played'] and season['matches_played'].isdigit():
            total_matches += int(season['matches_played'])
        if season['goals'] and season['goals'].isdigit():
            total_goals += int(season['goals'])
        if season['assists'] and season['assists'].isdigit():
            total_assists += int(season['assists'])
    
    career_data['total_matches_national'] = total_matches
    career_data['total_goals_national'] = total_goals
    career_data['total_assists_national'] = total_assists
    
    # Get national team info
    if seasons_processed:
        first_team = seasons_processed[0]['squad']
        career_data['national_team'] = first_team
        
        first_season = seasons_processed[0]['season']
        last_season = seasons_processed[-1]['season']
        career_data['national_team_years'] = f"{first_season} to {last_season}"
    
    print(f"    National Team: {career_data['national_team']}")
    print(f"    Seasons: {career_data['total_seasons_national']}, Matches: {career_data['total_matches_national']}, Goals: {career_data['total_goals_national']}, Assists: {career_data['total_assists_national']}")
    
    return career_data, all_seasons

def process_national_team_table(national_table, player_name, player_url, all_seasons):
    """Procesa SOLO la tabla de selección nacional con mapeo CORREGIDO"""
    seasons_processed = []
    
    try:
        # Convert table to DataFrame
        df = pd.read_html(io.StringIO(str(national_table)))[0]
        print(f"    Processing NATIONAL TEAM table with {len(df)} rows")
        
        # Handle multi-index columns
        if isinstance(df.columns, pd.MultiIndex):
            new_columns = []
            for col in df.columns:
                if pd.isna(col[0]) or col[0] == '':
                    new_columns.append(col[1])
                else:
                    new_columns.append(f"{col[0]}_{col[1]}")
            df.columns = new_columns
        
        columns_list = list(df.columns)
        print(f"    NATIONAL TEAM Columns: {columns_list}")
        
        # Find the correct column indices for national team tables
        # CORRECTED: We need the TOTAL goals/assists, not per 90
        column_mapping = find_national_team_columns_corrected(columns_list)
        print(f"    Corrected column mapping: {column_mapping}")
        
        # Process each row
        for idx, row in df.iterrows():
            season_data = extract_from_national_team_row(row, column_mapping, player_url)
            
            if season_data and season_data['season'] and is_national_team_data(season_data):
                season_data['player_name'] = player_name
                seasons_processed.append(season_data)
                all_seasons.append(season_data)
        
        print(f"    Processed {len(seasons_processed)} valid NATIONAL TEAM seasons")
        return seasons_processed
        
    except Exception as e:
        print(f"    Error processing NATIONAL TEAM table: {e}")
        return []

def find_national_team_columns_corrected(columns_list):
    """Find the correct columns for national team tables - CORRECTED FOR TOTALS"""
    mapping = {
        'season': None, 'squad': None, 'competition': None, 'age': None,
        'matches_played': None, 'minutes': None, 'goals': None, 'assists': None
    }
    
    # National team table structure:
    # Season, Age, Squad, Comp, LgRank, MP, Starts, Min, 90s, Gls, Ast, G+A, G-PK, PK, PKatt, CrdY, CrdR, Gls/90, Ast/90, ...
    # We want the TOTAL goals (column 9) and TOTAL assists (column 10), not the per90 ones
    
    for i, col in enumerate(columns_list):
        col_str = str(col).lower()
        
        # Basic info columns
        if 'season' in col_str:
            mapping['season'] = i
        elif 'squad' in col_str or 'team' in col_str:
            mapping['squad'] = i
        elif 'comp' in col_str:
            mapping['competition'] = i
        elif 'age' in col_str:
            mapping['age'] = i
        
        # Stats columns - we want the TOTALS, not per90
        elif col_str == 'mp' or col_str.endswith('_mp'):
            mapping['matches_played'] = i
        elif col_str == 'min' or col_str.endswith('_min'):
            mapping['minutes'] = i
        elif (col_str == 'gls' or col_str.endswith('_gls')) and '90' not in col_str:
            # Make sure it's not the per90 column
            mapping['goals'] = i
        elif (col_str == 'ast' or col_str.endswith('_ast')) and '90' not in col_str:
            # Make sure it's not the per90 column
            mapping['assists'] = i
    
    # Fallback: Use typical positions in national team tables for TOTALS
    # Season(0), Age(1), Squad(2), Comp(3), MP(5), Min(7), Gls(9), Ast(10)
    if mapping['season'] is None and len(columns_list) > 0:
        mapping['season'] = 0
    if mapping['age'] is None and len(columns_list) > 1:
        mapping['age'] = 1
    if mapping['squad'] is None and len(columns_list) > 2:
        mapping['squad'] = 2
    if mapping['competition'] is None and len(columns_list) > 3:
        mapping['competition'] = 3
    if mapping['matches_played'] is None and len(columns_list) > 5:
        mapping['matches_played'] = 5
    if mapping['minutes'] is None and len(columns_list) > 7:
        mapping['minutes'] = 7
    if mapping['goals'] is None and len(columns_list) > 9:
        mapping['goals'] = 9
    if mapping['assists'] is None and len(columns_list) > 10:
        mapping['assists'] = 10
    
    return mapping

def is_national_team_data(season_data):
    """Verify that this is actually national team data"""
    squad = season_data.get('squad', '').lower()
    competition = season_data.get('competition', '').lower()
    
    # National team indicators
    national_indicators = [
        'world cup', 'euro', 'copa america', 'africa cup of nations',
        'asian cup', 'gold cup', 'confederations cup', 'wcq', 'qualifying',
        'friendlies', 'nation', 'international'
    ]
    
    # Check if it looks like national team data
    is_national = any(indicator in squad or indicator in competition 
                     for indicator in national_indicators)
    
    return is_national

def extract_from_national_team_row(row, column_mapping, player_url):
    """Extrae datos de una fila de selección nacional con mapeo correcto"""
    try:
        def get_value(col_idx):
            if col_idx is None or col_idx >= len(row):
                return ''
            value = row[col_idx]
            if pd.isna(value):
                return ''
            # Clean the value - remove commas from numbers
            clean_value = str(value).replace(',', '').strip()
            return clean_value
        
        season_val = get_value(column_mapping['season'])
        
        # Skip header and summary rows
        if (not season_val or season_val == 'nan' or 
            'Season' in season_val or 'Career' in season_val or
            'Seasons' in season_val):
            return None
        
        # Only process rows with valid season values (years)
        if not (season_val.isdigit() or '-' in season_val):
            return None
        
        season_data = {
            'season': season_val,
            'squad': get_value(column_mapping['squad']),
            'competition': get_value(column_mapping['competition']),
            'age': get_value(column_mapping['age']),
            'matches_played': get_value(column_mapping['matches_played']),
            'minutes': get_value(column_mapping['minutes']),
            'goals': get_value(column_mapping['goals']),
            'assists': get_value(column_mapping['assists']),
            'player_url': player_url,
            'player_name': ''
        }
        
        # Basic validation
        if not season_data['squad']:
            return None
            
        return season_data
        
    except Exception as e:
        return None

def safe_save_checkpoint(data, filename):
    """Safely save checkpoint with error handling"""
    try:
        df_temp = pd.DataFrame(data)
        df_temp.to_csv(filename, index=False, encoding='utf-8')
        print(f" Checkpoint saved: {filename}")
        return True
    except PermissionError:
        print(f" Permission denied for {filename}, trying alternative name...")
        try:
            # Try with a different filename
            alt_filename = f"temp_{filename}"
            df_temp = pd.DataFrame(data)
            df_temp.to_csv(alt_filename, index=False, encoding='utf-8')
            print(f" Checkpoint saved as: {alt_filename}")
            return True
        except Exception as e:
            print(f" Failed to save checkpoint: {e}")
            return False
    except Exception as e:
        print(f" Failed to save checkpoint: {e}")
        return False

def main():
    """Extrae el historial de selección nacional de TODOS los jugadores"""
    print(" EXTRACTING PLAYER NATIONAL TEAM HISTORY - ALL PLAYERS")
    print("=" * 70)
    
    try:
        df_existing = pd.read_csv(INPUT_CSV)
        print(f" Loaded {len(df_existing)} players with URLs")
        
        players_to_process = df_existing[
            (df_existing['player_url'] != 'Not found') & 
            (df_existing['player_url'].notna())
        ]
        
        print(f" Players to process: {len(players_to_process)}")
        
    except Exception as e:
        print(f"Error reading CSV: {e}")
        return
    
    driver = setup_stealth_driver()
    if not driver:
        print(" Failed to setup browser driver")
        return
    
    all_national_summaries = []
    all_national_seasons = []
    
    try:
        # Process ALL players - start from beginning for new file
        total_players = len(players_to_process)
        
        for index, row in players_to_process.iterrows():
            player_name = str(row.get('player_name', '')).strip()
            player_slug = str(row.get('player_slug', '')).strip()
            player_url = str(row.get('player_url', '')).strip()
            
            print(f"\n{index + 1}/{total_players}: {player_name}")
            
            # Extract national team history
            national_summary, national_seasons = extract_national_team_data(driver, player_url, player_name, player_slug)
            
            all_national_summaries.append(national_summary)
            all_national_seasons.extend(national_seasons)
            
            # Progress update every 10 players
            if (index + 1) % 10 == 0:
                with_national_team = len([n for n in all_national_summaries if n['has_national_team'] == 'Yes'])
                print(f"\n📊 Progress: {index + 1}/{total_players}, {with_national_team} with national team data")
                
                # Save checkpoint with error handling
                if all_national_seasons:
                    safe_save_checkpoint(all_national_seasons, f"national_team_checkpoint_{index+1}.csv")
            
            # Random delay to avoid blocking
            time.sleep(random.uniform(5, 8))
    
    except KeyboardInterrupt:
        print("\n Process interrupted by user")
        print(" Saving collected data so far...")
    except Exception as e:
        print(f"\n Unexpected error: {e}")
        print(" Saving collected data so far...")
    finally:
        if driver:
            driver.quit()
            print("\n Browser driver closed")
    
    # Save final results
    if all_national_summaries:
        df_summary = pd.DataFrame(all_national_summaries)
        df_summary.to_csv("players_national_team_summary.csv", index=False, encoding='utf-8')
        
        if all_national_seasons:
            df_seasons = pd.DataFrame(all_national_seasons)
            df_seasons.to_csv(OUTPUT_CSV, index=False, encoding='utf-8')
            
            print(f"\n FINAL NATIONAL TEAM SAMPLE DATA (First 10 rows):")
            print(df_seasons.head(10).to_string(index=False))
        
        # Final summary
        total_processed = len(all_national_summaries)
        with_national_team = len([n for n in all_national_summaries if n['has_national_team'] == 'Yes'])
        
        print(f"\n FINAL RESULTS:")
        print(f"    Total players processed: {total_processed}")
        print(f"    Players with national team data: {with_national_team}")
        print(f"    Players without national team data: {total_processed - with_national_team}")
        print(f"    Total national team seasons extracted: {len(all_national_seasons)}")
        
        if total_processed > 0:
            success_rate = (with_national_team / total_processed) * 100
            print(f"    Success rate: {success_rate:.1f}%")
        
        print(f"\n Files saved:")
        print(f"   - players_national_team_summary.csv")
        print(f"   - {OUTPUT_CSV}")
    else:
        print(" No data extracted")

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import re
import time
import random

# ========= CONFIGURATION =========
INPUT_CSV = "death/tran_url.csv"  
OUTPUT_CSV = "death/players_with_deceased_status.csv"

HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/141.0.0.0 Safari/537.36"
}

def check_deceased_status(player_url, player_name):
    """Check if a player is deceased and extract death date if available"""
    print(f"🔍 Checking: {player_name}")
    
    deceased_data = {
        'deceased': 'No',
        'death_date': 'Not found',
        'death_age': 'Not found'
    }
    
    try:
        response = requests.get(player_url, headers=HEADERS, timeout=10)
        if response.status_code != 200:
            print(f"    HTTP Error {response.status_code}")
            return deceased_data
        
        soup = BeautifulSoup(response.content, "html.parser")
        
        # Method 1: Look for death information in the data grid
        data_grid = soup.find('div', class_='info-table')
        if data_grid:
            grid_text = data_grid.get_text()
            
            # Look for death date patterns
            death_match = re.search(r'Date of death:\s*([^\n]+)', grid_text)
            if death_match:
                deceased_data['deceased'] = 'Yes'
                death_info = death_match.group(1).strip()
                
                # Extract date and age if available
                date_age_match = re.search(r'(\d{2}\.\d{2}\.\d{4})\s*\((\d+)\)', death_info)
                if date_age_match:
                    deceased_data['death_date'] = date_age_match.group(1)  # 01.06.2019
                    deceased_data['death_age'] = date_age_match.group(2)   # 35
                    print(f"    Deceased: {deceased_data['death_date']} (age {deceased_data['death_age']})")
                else:
                    deceased_data['death_date'] = death_info
                    print(f"    Deceased: {death_info}")
                
                return deceased_data
        
        # Method 2: Look for "Date of death" in labels
        data_labels = soup.find_all('span', class_='data-header__label')
        for label in data_labels:
            label_text = label.get_text(strip=True)
            if 'Date of death' in label_text:
                deceased_data['deceased'] = 'Yes'
                death_content = label.find_next('span', class_='data-header__content')
                if death_content:
                    death_info = death_content.get_text(strip=True)
                    
                    # Extract date and age
                    date_age_match = re.search(r'(\d{2}\.\d{2}\.\d{4})\s*\((\d+)\)', death_info)
                    if date_age_match:
                        deceased_data['death_date'] = date_age_match.group(1)
                        deceased_data['death_age'] = date_age_match.group(2)
                        print(f"    Deceased: {deceased_data['death_date']} (age {deceased_data['death_age']})")
                    else:
                        deceased_data['death_date'] = death_info
                        print(f"    Deceased: {death_info}")
                
                return deceased_data
        
        # Method 3: Check for memorial indicators in the page
        page_text = soup.get_text()
        death_indicators = ['date of death', 'passed away', 'deceased', 'died', 'in memoriam']
        if any(indicator in page_text.lower() for indicator in death_indicators):
            deceased_data['deceased'] = 'Yes'
            print(f"    Deceased (found memorial text)")
        
        if deceased_data['deceased'] == 'No':
            print(f"    Alive")
        
        return deceased_data
        
    except Exception as e:
        print(f"   Error: {str(e)[:50]}")
        return deceased_data

def add_deceased_status_to_csv():
    """Add deceased status to your existing CSV"""
    print("ADDING DECEASED STATUS TO PLAYERS")
    print("=" * 60)
    
    try:
        df = pd.read_csv(INPUT_CSV)
        print(f" Loaded {len(df)} players")
        
        # Add new columns
        df['deceased'] = 'No'
        df['death_date'] = 'Not found'
        df['death_age'] = 'Not found'
        
        deceased_count = 0
        
        for index, row in df.iterrows():
            player_name = str(row.get('player_name', '')).strip()
            player_url = str(row.get('transfermarkt_url', '')).strip()
            
            print(f"\n[{index + 1}/{len(df)}] {player_name}")
            
            # Skip if no valid URL
            if player_url == 'Not found' or 'transfermarkt.com' not in player_url:
                print(f"     No valid URL, skipping")
                continue
            
            # Check deceased status
            deceased_data = check_deceased_status(player_url, player_name)
            
            # Update the row
            df.at[index, 'deceased'] = deceased_data['deceased']
            df.at[index, 'death_date'] = deceased_data['death_date']
            df.at[index, 'death_age'] = deceased_data['death_age']
            
            if deceased_data['deceased'] == 'Yes':
                deceased_count += 1
            
            # Save progress every 20 players
            if (index + 1) % 20 == 0:
                df.to_csv(f"progress_deceased_{index + 1}.csv", index=False)
                print(f" Progress saved: {index + 1} players")
            
            # Delay between requests
            if index + 1 < len(df):
                delay = random.uniform(2, 4)
                time.sleep(delay)
        
        # Save final results
        df.to_csv(OUTPUT_CSV, index=False, encoding='utf-8')
        
        print(f"\n COMPLETED!")
        print(f" DECEASED PLAYERS FOUND: {deceased_count}")
        print(f" Results saved to: {OUTPUT_CSV}")
        
        # Show deceased players
        if deceased_count > 0:
            print(f"\n DECEASED PLAYERS:")
            deceased_players = df[df['deceased'] == 'Yes']
            for _, player in deceased_players.iterrows():
                print(f"    {player['player_name']} - {player['death_date']} (age {player['death_age']})")
        
        return df
        
    except Exception as e:
        print(f" Error: {e}")
        return None

# ========= RUN THE CODE =========
if __name__ == "__main__":
    add_deceased_status_to_csv()
```
