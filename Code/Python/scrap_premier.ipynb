```python
import pandas as pd
import re

# ========= CONFIGURATION =========
INPUT_CSV = "premier_players.csv"

OUTPUT_CSV = "premier_players_with_urls.csv"

def create_transfermarkt_url_from_id(player_name, transfermarkt_id):
    """Create Transfermarkt URL from player name and ID"""
    # Clean player name for URL slug
    slug = player_name.lower()
    
    # Replace spaces with hyphens
    slug = slug.replace(' ', '-')
    
    # Remove special characters, keep only alphanumeric and hyphens
    slug = re.sub(r'[^a-z0-9-]', '', slug)
    
    # Construct URL
    url = f"https://www.transfermarkt.com/{slug}/profil/spieler/{transfermarkt_id}"
    return url

def add_urls_to_premier_players():
    """Add Transfermarkt URLs to premier_players.csv"""
    print("CREATING TRANSFERMARKT URLS FOR PREMIER LEAGUE PLAYERS")
    print("=" * 60)
    
    try:
        # Read CSV
        df = pd.read_csv(INPUT_CSV)
        print(f" Loaded {len(df)} players from {INPUT_CSV}")
        
        # Check required columns
        print(f" Columns in your CSV: {list(df.columns)}")
        
        # Check if we have the right column names
        name_column = None
        id_column = None
        
        # Try to find the correct column names
        possible_name_columns = ['player_name', 'name', 'Name', 'Player', 'player']
        possible_id_columns = ['transfermarkt_id', 'tm_id', 'id', 'ID', 'player_id']
        
        for col in possible_name_columns:
            if col in df.columns:
                name_column = col
                break
                
        for col in possible_id_columns:
            if col in df.columns:
                id_column = col
                break
        
        if not name_column or not id_column:
            print(" Could not find player name and ID columns automatically")
            print("  Make sure your CSV has columns for player names and Transfermarkt IDs")
            return
        
        print(f" Using columns: '{name_column}' for names, '{id_column}' for IDs")
        
        # Add Transfermarkt URL column
        df['transfermarkt_url'] = 'Not found'
        
        url_count = 0
        total_players = len(df)
        
        print(f"\n Creating URLs for {total_players} players...")
        
        for index, row in df.iterrows():
            player_name = str(row[name_column]).strip()
            transfermarkt_id = str(row[id_column]).strip()
            
            # Skip if missing data
            if not player_name or player_name == 'nan' or not transfermarkt_id or transfermarkt_id == 'nan':
                continue
            
            # Create URL
            transfermarkt_url = create_transfermarkt_url_from_id(player_name, transfermarkt_id)
            df.at[index, 'transfermarkt_url'] = transfermarkt_url
            url_count += 1
            
            # Progress indicator
            if (index + 1) % 100 == 0:
                print(f"   ✅ Processed {index + 1}/{total_players} players...")
        
        # Save results
        df.to_csv(OUTPUT_CSV, index=False)
        
        print(f"\n URL CREATION COMPLETED!")
        print(f" RESULTS:")
        print(f"   Total players: {total_players}")
        print(f"   URLs created: {url_count}")
        print(f"   Players without URLs: {total_players - url_count}")
        
        # Show sample of created URLs
        print(f"\n SAMPLE CREATED URLS:")
        sample_df = df[df['transfermarkt_url'] != 'Not found'].head(5)
        for _, player in sample_df.iterrows():
            print(f"   • {player[name_column]} → {player['transfermarkt_url']}")
        
        print(f"\n Results saved to: {OUTPUT_CSV}")
        
        return df
        
    except FileNotFoundError:
        print(f" File '{INPUT_CSV}' not found")
        print("   Please make sure the file is in the same directory as this script")
        return None
    except Exception as e:
        print(f" Error: {e}")
        return None

# Quick function to check your CSV structure
def check_csv_structure():
    """Check the structure of your premier_players.csv file"""
    print(" CHECKING CSV STRUCTURE")
    print("=" * 60)
    
    try:
        df = pd.read_csv(INPUT_CSV)
        print(f" File '{INPUT_CSV}' loaded successfully")
        print(f" Total rows: {len(df)}")
        print(f" Columns: {list(df.columns)}")
        print(f"\n First 3 rows:")
        print(df.head(3))
        
        # Check for missing data
        name_col = None
        id_col = None
        
        for col in df.columns:
            if any(word in col.lower() for word in ['name', 'player']):
                name_col = col
            if any(word in col.lower() for word in ['id', 'tm_id', 'transfermarkt']):
                id_col = col
        
        if name_col and id_col:
            print(f"\n Detected columns:")
            print(f"   Name column: '{name_col}'")
            print(f"   ID column: '{id_col}'")
            
            # Check for missing values
            missing_names = df[name_col].isna().sum()
            missing_ids = df[id_col].isna().sum()
            
            print(f"\n Data quality:")
            print(f"   Missing names: {missing_names}")
            print(f"   Missing IDs: {missing_ids}")
            
        return df
        
    except FileNotFoundError:
        print(f" File '{INPUT_CSV}' not found")
        return None
    except Exception as e:
        print(f" Error: {e}")
        return None

# ========= RUN THE CODE =========
if __name__ == "__main__":
    print("Choose an option:")
    print("1. Check CSV structure first")
    print("2. Create URLs directly")
    
    choice = input("Enter 1 or 2: ").strip()
    
    if choice == "1":
        check_csv_structure()
        print("\n" + "="*60)
        proceed = input("Proceed with URL creation? (y/n): ").strip().lower()
        if proceed == 'y':
            add_urls_to_premier_players()
    else:
        add_urls_to_premier_players()
```

# Get debut information from transfermarkt
debut age, debut club and date



```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import re
import time
import random

# ========= CONFIGURATION =========
#INPUT_CSV = "premier_players_with_urls.csv"  
## SECOND SCRAP IS WITH THE FOLLOWING INPUT< TO GET ONLY THE INFO FOR PLAYERS THAT HAD INCORRECT IDS
INPUT_CSV = "Second_scrap/premier_players_with_urls_revised.csv"  
OUTPUT_CSV = "Second_scrap/premier_players_debut_2.csv"

HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/141.0.0.0 Safari/537.36"
}

def extract_debut_info_final(player_url, player_name):
    """Final debut information extraction with comprehensive handling"""
    print(f"🔍 Extracting debut info: {player_name}")
    
    debut_data = {
        'first_tier_debut_date': 'Not found',
        'first_tier_debut_age': 'Not found',
        'first_tier_debut_club': 'Not found',
        'first_tier_debut_competition': 'Not found',
        'second_tier_debut_date': 'No second tier debut',
        'second_tier_debut_age': 'No second tier debut',
        'second_tier_debut_club': 'No second tier debut',
        'second_tier_debut_competition': 'No second tier debut'
    }
    
    try:
        player_id_match = re.search(r'/spieler/(\d+)', player_url)
        if not player_id_match:
            return debut_data
        
        player_id = player_id_match.group(1)
        debut_url = f"https://www.transfermarkt.com/-/debuets/spieler/{player_id}"
        
        response = requests.get(debut_url, headers=HEADERS, timeout=10)
        if response.status_code != 200:
            return debut_data
        
        soup = BeautifulSoup(response.content, "html.parser")
        tables = soup.find_all('table')
        
        first_tier_found = False
        second_tier_found = False
        found_second_tier_section = False
        
        for table in tables:
            rows = table.find_all('tr')
            if len(rows) > 10:  # This should be our debut table
                current_section = None
                
                for row in rows:
                    cells = row.find_all('td')
                    
                    # Section headers
                    if len(cells) == 1:
                        section_text = cells[0].get_text(strip=True)
                        
                        if section_text == 'First Tier':
                            current_section = 'first_tier'
                            print(f"   🏷️  First Tier section found")
                        elif section_text == 'Second Tier':
                            current_section = 'second_tier'
                            found_second_tier_section = True
                            print(f"   🏷️  Second Tier section found")
                        else:
                            current_section = None
                        continue
                    
                    # Data rows
                    if len(cells) >= 7 and current_section:
                        competition = cells[1].get_text(strip=True)
                        club = cells[2].get_text(strip=True)
                        date = cells[3].get_text(strip=True)
                        age = cells[6].get_text(strip=True)
                        
                        # Validate we have a proper date
                        if date and re.match(r'\d{2}/\d{2}/\d{4}', date):
                            
                            if current_section == 'first_tier' and not first_tier_found:
                                debut_data.update({
                                    'first_tier_debut_date': date,
                                    'first_tier_debut_age': age,
                                    'first_tier_debut_club': club,
                                    'first_tier_debut_competition': competition
                                })
                                first_tier_found = True
                                print(f"    First Tier: {date} ({age}) - {club}")
                            
                            elif current_section == 'second_tier' and not second_tier_found:
                                debut_data.update({
                                    'second_tier_debut_date': date,
                                    'second_tier_debut_age': age,
                                    'second_tier_debut_club': club,
                                    'second_tier_debut_competition': competition
                                })
                                second_tier_found = True
                                print(f"   ✅ Second Tier: {date} ({age}) - {club}")
                
                # If we found both, we can stop
                if first_tier_found and second_tier_found:
                    break
        
        # Update status for players without second tier debuts
        if not found_second_tier_section:
            print(f"     No Second Tier section found for this player")
            debut_data.update({
                'second_tier_debut_date': 'No second tier career',
                'second_tier_debut_age': 'No second tier career',
                'second_tier_debut_club': 'No second tier career',
                'second_tier_debut_competition': 'No second tier career'
            })
        elif found_second_tier_section and not second_tier_found:
            print(f"     Second Tier section exists but no valid debut found")
            debut_data.update({
                'second_tier_debut_date': 'Second tier section empty',
                'second_tier_debut_age': 'Second tier section empty',
                'second_tier_debut_club': 'Second tier section empty',
                'second_tier_debut_competition': 'Second tier section empty'
            })
        
        return debut_data
        
    except Exception as e:
        print(f"    Error: {str(e)[:50]}")
        return debut_data

def process_all_players_with_debuts():
    """Process all players with comprehensive debut extraction"""
    print(" COMPREHENSIVE DEBUT EXTRACTION FOR ALL PLAYERS")
    print("=" * 60)
    
    try:
        df = pd.read_csv(INPUT_CSV)
        print(f" Loaded {len(df)} players")
        
        # Add debut columns with meaningful default values
        debut_columns = [
            'first_tier_debut_date', 'first_tier_debut_age', 'first_tier_debut_club', 'first_tier_debut_competition',
            'second_tier_debut_date', 'second_tier_debut_age', 'second_tier_debut_club', 'second_tier_debut_competition'
        ]
        
        for col in debut_columns:
            if col not in df.columns:
                if 'first_tier' in col:
                    df[col] = 'Not found'
                else:
                    df[col] = 'Not processed'
        
        total_players = len(df)
        processed_count = 0
        first_tier_count = 0
        second_tier_count = 0
        
        for index, row in df.iterrows():
            player_name = str(row.get('player_name', '')).strip()
            player_url = str(row.get('transfermarkt_url', '')).strip()
            
            print(f"\n[{index + 1}/{total_players}] {player_name}")
            
            if player_url == 'Not found' or 'transfermarkt.com' not in player_url:
                print(f"     No valid URL, skipping")
                continue
            
            # Extract debut info
            debut_data = extract_debut_info_final(player_url, player_name)
            
            # Update the row
            for key, value in debut_data.items():
                df.at[index, key] = value
            
            # Update counters
            processed_count += 1
            if debut_data['first_tier_debut_date'] not in ['Not found', 'Not processed']:
                first_tier_count += 1
            if debut_data['second_tier_debut_date'] not in ['No second tier career', 'No second tier debut', 'Second tier section empty', 'Not processed', 'Not found']:
                second_tier_count += 1
            
            # Save progress every 20 players
            if (index + 1) % 20 == 0:
                progress_csv = f"progress_debuts_{index + 1}.csv"
                df.to_csv(progress_csv, index=False)
                print(f" Progress saved: {progress_csv}")
                
                # Interim summary
                print(f" INTERIM: {first_tier_count}/{processed_count} with First Tier, {second_tier_count}/{processed_count} with Second Tier")
            
            # Delay
            if index + 1 < total_players:
                delay = random.uniform(2, 4)
                time.sleep(delay)
        
        # Save final results
        df.to_csv(OUTPUT_CSV, index=False, encoding='utf-8')
        
        # Final summary
        print(f"\n EXTRACTION COMPLETED!")
        print(f" FINAL RESULTS:")
        print(f"   Total players processed: {processed_count}")
        print(f"   Players with First Tier debut: {first_tier_count} ({first_tier_count/processed_count*100:.1f}%)")
        print(f"   Players with Second Tier debut: {second_tier_count} ({second_tier_count/processed_count*100:.1f}%)")
        
        # Show breakdown of second tier status
        second_tier_breakdown = df['second_tier_debut_date'].value_counts()
        print(f"\n SECOND TIER BREAKDOWN:")
        for status, count in second_tier_breakdown.items():
            print(f"   {status}: {count} players")
        
        # Show sample of players with both tiers
        players_with_both = df[
            (df['first_tier_debut_date'] != 'Not found') & 
            (df['second_tier_debut_date'].isin(['No second tier career', 'No second tier debut', 'Second tier section empty', 'Not processed']))
        ]
        if len(players_with_both) > 0:
            print(f"\n SAMPLE PLAYERS WITH BOTH TIERS:")
            sample = players_with_both.head(3)
            for _, player in sample.iterrows():
                print(f"   • {player['player_name']}")
                print(f"     First: {player['first_tier_debut_date']} | Second: {player['second_tier_debut_date']}")
        
        print(f" Final results saved to: {OUTPUT_CSV}")
        
        return df
        
    except Exception as e:
        print(f" Error: {e}")
        return None

# Quick test to verify the final version
def quick_test_final():
    """Quick test of the final extraction method"""
    test_players = [
        ("Antonio Adán", "https://www.transfermarkt.com/antonio-adan/profil/spieler/44062"),
        ("Sergio Ramos", "https://www.transfermarkt.com/sergio-ramos/profil/spieler/25557"),
        ("Lionel Messi", "https://www.transfermarkt.com/lionel-messi/profil/spieler/28003")
    ]
    
    print(" FINAL EXTRACTION TEST")
    print("=" * 60)
    
    for player_name, player_url in test_players:
        print(f"\n {player_name}:")
        debut_data = extract_debut_info_final(player_url, player_name)
        print(f"   First Tier: {debut_data['first_tier_debut_date']}")
        print(f"   Second Tier: {debut_data['second_tier_debut_date']}")
        time.sleep(2)

# ========= RUN THE CODE =========
if __name__ == "__main__":
    print("Choose an option:")
    print("1. Quick test of final extraction")
    print("2. Process all players with debut extraction")
    
    choice = input("Enter 1 or 2: ").strip()
    
    if choice == "1":
        quick_test_final()
    else:
        process_all_players_with_debuts()
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import re
import time
import random

# ========= CONFIGURATION =========
#URLS_CSV = "premier_players_with_urls.csv"
## SECOND SCRAP IS WITH THE FOLLOWING INPUT< TO GET ONLY THE INFO FOR PLAYERS THAT HAD INCORRECT IDS
URLS_CSV = "Second_scrap/premier_players_with_urls_revised.csv" 

DATA_CSV = "Second_scrap/players_transfermarkt_2.csv"
PROCESS_SIZE = 1500 #to run all players

HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/141.0.0.0 Safari/537.36"
}

def fix_encoding(text):
    """Fix encoding issues in player names"""
    if not text or text == 'Not found':
        return text
    
    # Common encoding fixes
    encoding_fixes = {
        'Ãƒ': '', 'â€': '', 'Ã³': 'ó', 'Ã©': 'é', 'Ã': 'í', 
        'Ã±': 'ñ', 'Ãº': 'ú', 'Ã¡': 'á', 'Ã¨': 'è', 'Ãª': 'ê',
        'Ã§': 'ç', 'Ã£': 'ã', 'Ãµ': 'õ', 'Ã¶': 'ö', 'Ã¼': 'ü',
        'Ã¯': 'ï', 'Ã«': 'ë', 'Ã¬': 'ì', 'Ã': 'à'
    }
    
    fixed_text = text
    for wrong, correct in encoding_fixes.items():
        fixed_text = fixed_text.replace(wrong, correct)
    
    return fixed_text

def extract_injury_data_correct_table(player_url, player_name):
    """Extract injury data from the SEASON SUMMARY table"""
    print(f" Extracting injury data: {player_name}")
    
    injury_data = {
        'injury_count_total': 0,
        'days_missed_total': 0,
        'games_missed_total': 0,
        'injury_seasons': 0
    }
    
    try:
        # Construct injury URL from player URL
        injury_url = player_url.replace('/profil/', '/verletzungen/')
        
        response = requests.get(injury_url, headers=HEADERS)
        if response.status_code != 200:
            return injury_data
        
        soup = BeautifulSoup(response.content, "html.parser")
        
        # Look for ALL tables with class 'items'
        all_tables = soup.find_all('table', {'class': 'items'})
        
        # Find the summary table
        summary_table = None
        for table in all_tables:
            header = table.get_text()
            if 'Season' in header and 'Days' in header and 'Injuries' in header and 'Games missed' in header:
                summary_table = table
                break
        
        if not summary_table:
            return injury_data
        
        # Find the tbody within the summary table
        tbody = summary_table.find('tbody')
        if not tbody:
            return injury_data
        
        # Extract all rows from tbody
        rows = tbody.find_all('tr')
        
        total_injuries = 0
        total_days = 0
        total_games_missed = 0
        
        for row in rows:
            cells = row.find_all('td')
            if len(cells) >= 4:
                days_text = cells[1].get_text(strip=True)
                injuries = cells[2].get_text(strip=True)
                games_missed = cells[3].get_text(strip=True)
                
                # Parse injuries count
                injuries_num = int(injuries) if injuries.isdigit() else 0
                
                # Parse games missed
                games_missed_num = int(games_missed) if games_missed.isdigit() else 0
                
                # Parse days
                days_num = 0
                if days_text:
                    days_match = re.search(r'(\d+)', days_text)
                    if days_match:
                        days_num = int(days_match.group(1))
                
                total_injuries += injuries_num
                total_days += days_num
                total_games_missed += games_missed_num
        
        injury_data.update({
            'injury_count_total': total_injuries,
            'days_missed_total': total_days,
            'games_missed_total': total_games_missed,
            'injury_seasons': len(rows)
        })
        
        print(f"    Injuries: {total_injuries} total, {total_days} days, {total_games_missed} games")
        
        return injury_data
        
    except Exception as e:
        print(f"    Injury extraction error: {e}")
        return injury_data

def extract_retirement_info(soup, player_info):
    """Extract retirement information for retired players - FIXED!!"""
    try:
        # Method 1: Check if the club name is "Retired"
        club_element = soup.select_one('span[class="data-header__club"]')
        if club_element:
            club_text = club_element.get_text(strip=True)
            if 'Retired' in club_text:
                player_info['retired'] = 'Yes'
                print(f"    Player is retired (Method 1)")
                
                # Extract retirement date using the label
                retirement_labels = soup.find_all('span', class_='data-header__label')
                for label in retirement_labels:
                    label_text = label.get_text(strip=True)
                    if 'Retired since' in label_text:
                        retirement_date_span = label.find_next('span', class_='data-header__content')
                        if retirement_date_span:
                            player_info['retirement_date'] = retirement_date_span.get_text(strip=True)
                            print(f"    Retirement Date: {player_info['retirement_date']}")
                        break
                
                # Extract last club
                for label in retirement_labels:
                    label_text = label.get_text(strip=True)
                    if 'Last club' in label_text:
                        last_club_link = label.find_next('a')
                        if last_club_link:
                            player_info['current_club'] = last_club_link.get_text(strip=True)
                            print(f"    Last Club: {player_info['current_club']}")
                        break
                return player_info
        
        # Method 2: Look for "Retired since" directly in all labels
        retirement_labels = soup.find_all('span', class_='data-header__label')
        for label in retirement_labels:
            label_text = label.get_text(strip=True)
            if 'Retired since' in label_text:
                player_info['retired'] = 'Yes'
                print(f"   🏁 Player is retired (Method 2)")
                
                # Extract retirement date
                retirement_date_span = label.find_next('span', class_='data-header__content')
                if retirement_date_span:
                    player_info['retirement_date'] = retirement_date_span.get_text(strip=True)
                    print(f"    Retirement Date: {player_info['retirement_date']}")
                
                # Extract last club
                for label2 in retirement_labels:
                    label_text2 = label2.get_text(strip=True)
                    if 'Last club' in label_text2:
                        last_club_link = label2.find_next('a')
                        if last_club_link:
                            player_info['current_club'] = last_club_link.get_text(strip=True)
                            print(f"    Last Club: {player_info['current_club']}")
                        break
                return player_info
        
        return player_info
        
    except Exception as e:
        print(f"    Retirement info error: {e}")
        return player_info

def extract_detailed_player_data(player_url, player_name):
    """Extract all detailed player information from Transfermarkt"""
    print(f" Extracting data: {player_name}")
    
    player_info = {
        'player_name': fix_encoding(player_name),  # Fix encoding for player name
        'player_url': player_url,
        'status': 'Success'
    }
    
    # Initialize only the fields we want
    target_fields = [
        'player_id', 'player_slug', 'home_name', 'dob_age', 'place_birth',
        'height', 'citizenship', 'position', 'foot', 'agent', 'current_club',
        'joined', 'contract_expires', 'outfitter', 'injury_count_total', 
        'days_missed_total', 'games_missed_total', 'retired', 'retirement_date',
        'shirt_number'
    ]
    
    for field in target_fields:
        player_info[field] = 'Not found'
    
    try:
        if player_url == 'Not found':
            player_info['status'] = 'No URL'
            return player_info
        
        # Extract player ID from URL
        player_id = player_url.split('/')[-1]
        player_info['player_id'] = player_id
        player_info['player_slug'] = player_url.split('/')[-2]
        
        response = requests.get(player_url, headers=HEADERS)
        if response.status_code != 200:
            player_info['status'] = f'HTTP Error: {response.status_code}'
            return player_info
        
        soup = BeautifulSoup(response.content, "html.parser")
        
        # ========= BASIC INFORMATION =========
        
        # Full name (with encoding fix)
        name_element = soup.select_one('h1[class="data-header__headline-wrapper"]')
        if name_element:
            full_name = name_element.text.split('\n')[-1].strip()
            player_info['home_name'] = fix_encoding(full_name)
            print(f"   👤 Name: {player_info['home_name']}")
        
        # Shirt number
        number_element = soup.select_one('span[class="data-header__shirt-number"]')
        if number_element:
            player_info['shirt_number'] = number_element.text.replace('#', '').strip()
            print(f"   🔢 Number: {player_info['shirt_number']}")
        
        # ========= RETIREMENT INFORMATION =========
        # Check retirement status first, as it affects current_club
        player_info = extract_retirement_info(soup, player_info)
        
        # ========= CURRENT CLUB (only if not retired) =========
        if player_info.get('retired') != 'Yes':
            club_element = soup.select_one('span[class="data-header__club"]')
            if club_element:
                club_text = club_element.text.split('\n')[-1].strip()
                if club_text and club_text != 'Retired':
                    player_info['current_club'] = fix_encoding(club_text)
                    print(f"   ⚽ Club: {player_info['current_club']}")
        
        # ========= DETAILED DATA GRID =========
        
        data_grid = soup.find('div', class_='info-table')
        if data_grid:
            grid_text = data_grid.get_text()
            
            # Extract all fields using regex patterns
            field_patterns = {
                'dob_age': r'Age:\s*([^\n]+)',
                'place_birth': r'Place of birth:\s*([^\n]+)',
                'height': r'Height:\s*([^\n]+)',
                'citizenship': r'Citizenship:\s*([^\n]+)',
                'position': r'Position:\s*([^\n]+)',
                'foot': r'Foot:\s*([^\n]+)',
                'agent': r'Agent:\s*([^\n]+)',
                'joined': r'Joined:\s*([^\n]+)',
                'contract_expires': r'Contract expires:\s*([^\n]+)',
                'outfitter': r'Outfitter:\s*([^\n]+)'
            }
            
            for field, pattern in field_patterns.items():
                match = re.search(pattern, grid_text)
                if match:
                    player_info[field] = fix_encoding(match.group(1).strip())
                    print(f"    {field}: {player_info[field]}")
        
        # Set default retirement status if not found
        if player_info['retired'] == 'Not found':
            player_info['retired'] = 'No'
            print(f"    Active player")
        
        # ========= INJURY DATA =========
        injury_data = extract_injury_data_correct_table(player_url, player_name)
        player_info.update(injury_data)
        
        print(f"    Completed: {player_info['home_name']}")
        return player_info
        
    except Exception as e:
        print(f"    Error: {str(e)[:100]}")
        player_info['status'] = f'Error: {str(e)[:100]}'
        return player_info

def extract_data_sample():
    """Test with a few players"""
    print(" EXTRACTING PLAYER DATA (FIXED RETIREMENT DETECTION)")
    print("=" * 60)
    
    try:
        df_urls = pd.read_csv(URLS_CSV)
        print(f" Loaded {len(df_urls)} players with URLs")
        
        # Filter only players with valid URLs
        valid_players = df_urls[df_urls['transfermarkt_url'] != 'Not found']
        print(f" Found {len(valid_players)} players with valid URLs")
        
        # Take only PROCESS_SIZE players
        sample_players = valid_players.head(PROCESS_SIZE)
        print(f" Processing {len(sample_players)} sample players:")
        
        for i, (idx, row) in enumerate(sample_players.iterrows(), 1):
            print(f"   {i}. {row['player_name']}")
        
        all_results = []
        
        for index, row in sample_players.iterrows():
            player_name = row['player_name']
            player_url = row['transfermarkt_url']
            
            print(f"\n{'='*50}")
            print(f"[{index + 1}/{len(sample_players)}] {player_name}")
            print(f"{'='*50}")
            
            # Extract detailed data
            player_data = extract_detailed_player_data(player_url, player_name)
            all_results.append(player_data)
            
            # Delay between requests
            if index + 1 < len(sample_players):
                delay = random.uniform(4, 7)
                print(f"⏳ Waiting {delay:.1f} seconds...")
                time.sleep(delay)
        
        # Save results
        df_final = pd.DataFrame(all_results)
        df_final.to_csv(DATA_CSV, index=False, encoding='utf-8')
        
        # Summary
        successful = len([r for r in all_results if r['status'] == 'Success'])
        retired_count = len([r for r in all_results if r.get('retired') == 'Yes'])
        
        print(f"\n EXTRACTION SUMMARY:")
        print(f"   Total processed: {len(all_results)}")
        print(f"   Successful: {successful}")
        print(f"   Retired players found: {retired_count}")
        
        # Show cleaned data
        print(f"\n CLEANED DATA EXTRACTED:")
        for player in all_results:
            if player['status'] == 'Success':
                print(f"\n• {player['home_name']}")
                print(f"  Position: {player.get('position', 'N/A')}")
                print(f"  Club/Status: {player.get('current_club', 'N/A')}")
                print(f"  Retired: {player.get('retired', 'N/A')}")
                if player.get('retired') == 'Yes':
                    print(f"  Retirement Date: {player.get('retirement_date', 'N/A')}")
                print(f"  Injuries: {player.get('injury_count_total', 'N/A')}")
                print(f"  Days Missed: {player.get('days_missed_total', 'N/A')}")
                print(f"  Games Missed: {player.get('games_missed_total', 'N/A')}")
        
        print(f"\n Cleaned data saved to: {DATA_CSV}")
        
    except Exception as e:
        print(f" Error: {e}")

if __name__ == "__main__":
    extract_data_sample()
```

```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from bs4 import BeautifulSoup
import re
import time
import random
import io

# ========= CONFIGURATION =========


#INPUT_CSV = "Second_scrap/players_fbref.csv"
INPUT_CSV = "Second_scrap/players_fbref2.csv"

OUTPUT_CSV = "Second_scrap/players_premier_history2.csv"



def setup_selenium_driver():
    """Setup Selenium WebDriver"""
    chrome_options = Options()
    chrome_options.add_argument("--headless")
    chrome_options.add_argument("--no-sandbox")
    chrome_options.add_argument("--disable-dev-shm-usage")
    chrome_options.add_argument("--disable-blink-features=AutomationControlled")
    chrome_options.add_experimental_option("excludeSwitches", ["enable-automation"])
    chrome_options.add_experimental_option('useAutomationExtension', False)
    chrome_options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36")
    
    driver = webdriver.Chrome(options=chrome_options)
    driver.execute_script("Object.defineProperty(navigator, 'webdriver', {get: () => undefined})")
    return driver

def convert_to_all_comps_url(player_url):
    """Convierte la URL normal a URL de todas las competiciones"""
    if 'Not found' in player_url or not player_url:
        return player_url
    
    if '/all_comps/' in player_url:
        return player_url
    
    try:
        parts = player_url.split('/')
        player_id = parts[5]
        player_slug = parts[6].split('-Stats')[0] if '-Stats' in parts[6] else parts[6]
        all_comps_url = f"https://fbref.com/en/players/{player_id}/all_comps/{player_slug}-Stats---All-Competitions"
        return all_comps_url
    except:
        return player_url

def extract_career_history(driver, player_url, player_name, player_slug):
    """Extrae el historial completo de carrera del jugador"""
    print(f"🔍 Extracting career history: {player_name}")
    
    career_data = {
        'player_name': player_name,
        'player_slug': player_slug,
        'player_url': player_url,
        'total_seasons': 0,
        'total_clubs': 0,
        'career_years': '',
        'status': 'Success'
    }
    
    all_seasons = []
    
    try:
        if player_url == 'Not found' or not player_url:
            career_data['status'] = 'No URL'
            return career_data, all_seasons
        
        all_comps_url = convert_to_all_comps_url(player_url)
        print(f"   📊 Using all competitions URL: {all_comps_url}")
        
        driver.get(all_comps_url)
        time.sleep(4)
        
        soup = BeautifulSoup(driver.page_source, 'html.parser')
        
        # ========= EXTRACT MAIN CAREER TABLE =========
        career_table = None
        
        stats_tables = soup.find_all('table', id=re.compile(r'stats'))
        
        for table in stats_tables:
            try:
                table_html = str(table)
                df = pd.read_html(io.StringIO(table_html))[0]
                
                has_season = any('Season' in str(col) for col in df.columns)
                has_squad = any('Squad' in str(col) for col in df.columns)
                has_comp = any('Comp' in str(col) for col in df.columns)
                
                if has_season and has_squad and has_comp:
                    career_table = df
                    print(f"    Found ALL COMPETITIONS career table with {len(df)} rows")
                    break
                    
            except Exception as e:
                continue
        
        if career_table is not None:
            seasons_processed = process_complex_career_table(career_table, player_name, all_seasons)
            
            career_data['total_seasons'] = len(seasons_processed)
            career_data['total_clubs'] = len(set([s['squad'] for s in seasons_processed if s['squad']]))
            
            if seasons_processed:
                first_season = seasons_processed[0]['season']
                last_season = seasons_processed[-1]['season']
                career_data['career_years'] = f"{first_season} to {last_season}"
            
            print(f"   Career: {career_data['total_seasons']} seasons, {career_data['total_clubs']} clubs")
            
        else:
            print("    No career table found")
            career_data['status'] = 'No career table'
        
        return career_data, all_seasons
        
    except Exception as e:
        print(f"    Error: {player_name} - {str(e)[:100]}")
        career_data['status'] = f'Error: {str(e)[:100]}'
        return career_data, []

def process_complex_career_table(career_table, player_name, all_seasons):
    """Procesa la tabla compleja con multi-index"""
    seasons_processed = []
    
    try:
        print(f"    Processing complex multi-index table")
        print(f"   Table shape: {career_table.shape}")
        
        # Estrategia: Flatten the multi-index para hacerlo más manejable
        if isinstance(career_table.columns, pd.MultiIndex):
            # Crear nombres de columnas más simples
            new_columns = []
            for col in career_table.columns:
                if pd.isna(col[0]) or col[0] == '':
                    new_columns.append(col[1])
                else:
                    new_columns.append(f"{col[0]}_{col[1]}")
            
            career_table.columns = new_columns
            print(f"    Flattened columns: {new_columns[:10]}...")  # Mostrar primeras 10
        
        # Buscar las columnas clave por patrones
        column_mapping = find_columns_by_patterns(career_table)
        
        print(f"    Column mapping:")
        for key, value in column_mapping.items():
            print(f"      {key}: {value}")
        
        # Procesar cada fila
        for idx, row in career_table.iterrows():
            season_data = extract_from_complex_row(row, column_mapping)
            
            if season_data and season_data['season'] and season_data['season'] != 'Career' and 'Season' not in season_data['season']:
                season_data['player_name'] = player_name
                seasons_processed.append(season_data)
                all_seasons.append(season_data)
        
        return seasons_processed
        
    except Exception as e:
        print(f"    Error processing complex table: {e}")
        import traceback
        traceback.print_exc()
        return []

def find_columns_by_patterns(career_table):
    """Encuentra columnas por patrones específicos"""
    mapping = {
        'season': None, 'squad': None, 'competition': None, 'age': None,
        'matches_played': None, 'minutes': None, 'goals': None, 'assists': None
    }
    
    columns_list = list(career_table.columns)
    
    # Estrategia 1: Buscar por nombres específicos
    for i, col in enumerate(columns_list):
        col_str = str(col).lower()
        
        # Columnas básicas
        if 'season' in col_str:
            mapping['season'] = col
        elif 'squad' in col_str:
            mapping['squad'] = col
        elif 'comp' in col_str and 'country' not in col_str:
            mapping['competition'] = col
        elif 'age' in col_str:
            mapping['age'] = col
        
        # Playing Time stats
        elif 'mp' == col_str or ('playing' in col_str and 'mp' in col_str):
            mapping['matches_played'] = col
        elif 'min' == col_str or ('playing' in col_str and 'min' in col_str and '90s' not in col_str):
            mapping['minutes'] = col
        
        # Performance stats
        elif 'gls' == col_str or ('performance' in col_str and 'gls' in col_str):
            mapping['goals'] = col
        elif 'ast' == col_str or ('performance' in col_str and 'ast' in col_str):
            mapping['assists'] = col
    
    # Estrategia 2: Si no encontramos por nombre, usar posiciones fijas basadas en la estructura que vimos
    if mapping['season'] is None and len(columns_list) > 0:
        mapping['season'] = columns_list[0]
    if mapping['age'] is None and len(columns_list) > 1:
        mapping['age'] = columns_list[1]
    if mapping['squad'] is None and len(columns_list) > 2:
        mapping['squad'] = columns_list[2]
    if mapping['competition'] is None and len(columns_list) > 4:
        mapping['competition'] = columns_list[4]
    
    # Para MP y Min - posiciones basadas en la estructura mostrada
    if mapping['matches_played'] is None:
        # Buscar en las primeras columnas después de las básicas
        for i, col in enumerate(columns_list):
            col_str = str(col).lower()
            if col_str == 'mp' or col_str.endswith('_mp'):
                mapping['matches_played'] = col
                break
        if mapping['matches_played'] is None and len(columns_list) > 5:
            mapping['matches_played'] = columns_list[5]  # Posición típica para MP
    
    if mapping['minutes'] is None:
        for i, col in enumerate(columns_list):
            col_str = str(col).lower()
            if col_str == 'min' or col_str.endswith('_min'):
                mapping['minutes'] = col
                break
        if mapping['minutes'] is None and len(columns_list) > 6:
            mapping['minutes'] = columns_list[6]  # Posición típica para Min
    
    if mapping['goals'] is None:
        for i, col in enumerate(columns_list):
            col_str = str(col).lower()
            if col_str == 'gls' or col_str.endswith('_gls'):
                mapping['goals'] = col
                break
    
    if mapping['assists'] is None:
        for i, col in enumerate(columns_list):
            col_str = str(col).lower()
            if col_str == 'ast' or col_str.endswith('_ast'):
                mapping['assists'] = col
                break
    
    return mapping

def extract_from_complex_row(row, column_mapping):
    """Extrae datos de una fila compleja"""
    try:
        # Obtener valores básicos
        season_val = str(row[column_mapping['season']]) if column_mapping['season'] in row and pd.notna(row[column_mapping['season']]) else ''
        
        # Filtrar filas no deseadas
        if (not season_val or season_val == 'nan' or season_val == '' or 
            'Season' in season_val or 'Career' in season_val or 'Seasons' in season_val):
            return None
        
        def get_value(col_key):
            col = column_mapping[col_key]
            if col is None or col not in row:
                return ''
            try:
                value = row[col]
                if pd.isna(value):
                    return ''
                # Para minutos, mantener las comas
                if col_key == 'minutes':
                    return str(value)
                # Para otros valores numéricos, limpiar
                return str(value).replace(',', '').strip()
            except:
                return ''
        
        season_data = {
            'season': get_value('season'),
            'squad': get_value('squad'),
            'competition': get_value('competition'),
            'age': get_value('age'),
            'matches_played': get_value('matches_played'),
            'minutes': get_value('minutes'),
            'goals': get_value('goals'),
            'assists': get_value('assists')
        }
        
        # Validar que tenemos datos básicos
        if not season_data['season'] or not season_data['competition']:
            return None
        
        return season_data
        
    except Exception as e:
        print(f"    Error extracting from complex row: {e}")
        return None

def extract_additional_player_info(driver, player_url, player_name):
    """Extrae información adicional del jugador"""
    try:
        driver.get(player_url)
        time.sleep(2)
        
        soup = BeautifulSoup(driver.page_source, 'html.parser')
        
        info = {
            'player_name': player_name,
            'current_club': 'Not found',
            'position': 'Not found',
            'age': 'Not found',
            'nationality': 'Not found'
        }
        
        meta_div = soup.find('div', {'id': 'meta'})
        if meta_div:
            meta_text = meta_div.get_text()
            
            club_match = re.search(r'Club:\s*([^\n]+)', meta_text)
            if club_match:
                info['current_club'] = club_match.group(1).strip()
            
            position_match = re.search(r'Position:\s*([^\n▪]+)', meta_text)
            if position_match:
                info['position'] = position_match.group(1).strip()
            
            age_match = re.search(r'Age:\s*([^\s\)]+)', meta_text)
            if age_match:
                info['age'] = age_match.group(1).strip()
            
            national_match = re.search(r'National Team:\s*([^\n]+)', meta_text)
            if national_match:
                nationality_text = national_match.group(1).strip()
                info['nationality'] = re.sub(r'\s+[a-z]{2}$', '', nationality_text)
        
        return info
        
    except:
        return {
            'player_name': player_name,
            'current_club': 'Not found', 
            'position': 'Not found', 
            'age': 'Not found',
            'nationality': 'Not found'
        }

def main():
    """Extrae el historial de carrera de todos los jugadores"""
    print(" EXTRACTING PLAYER CAREER HISTORY (ALL COMPETITIONS)")
    print("=" * 70)
    
    try:
        df_existing = pd.read_csv(INPUT_CSV)
        print(f" Loaded {len(df_existing)} players with URLs")
        
        players_to_process = df_existing[
            (df_existing['player_url'] != 'Not found') & 
            (df_existing['player_url'].notna())
        ]
        
        print(f" Players to process: {len(players_to_process)}")
        
    except Exception as e:
        print(f" Error reading CSV: {e}")
        return
    
    driver = setup_selenium_driver()
    
    all_career_summaries = []
    all_seasons_detailed = []
    all_player_info = []
    
    try:
        # 
        test_players = players_to_process  
        
        total_players = len(test_players)
        
        for index, row in test_players.iterrows():
            player_name = str(row.get('player_name', '')).strip()
            player_slug = str(row.get('player_slug', '')).strip()
            player_url = str(row.get('player_url', '')).strip()
            
            print(f"\n{index + 1}/{total_players}: {player_name}")
            
           
            career_summary, seasons_data = extract_career_history(driver, player_url, player_name, player_slug)
            
           
            additional_info = extract_additional_player_info(driver, player_url, player_name)
            
           
            career_summary.update(additional_info)
            all_career_summaries.append(career_summary)
            all_player_info.append(additional_info)
            all_seasons_detailed.extend(seasons_data)
            
            
            if (index + 1) % 10 == 0:
                successful = len([c for c in all_career_summaries if c['status'] == 'Success'])
                print(f"\n Progress: {index + 1}/{total_players}, {successful} successful")
                
                
                if all_seasons_detailed:
                    df_temp = pd.DataFrame(all_seasons_detailed)
                    df_temp.to_csv(f"players_career_history_CHECKPOINT_{index+1}.csv", index=False, encoding='utf-8')
                    print(f" Checkpoint saved: players_career_history_CHECKPOINT_{index+1}.csv")
            
            
            time.sleep(random.uniform(2, 4))
    
    except KeyboardInterrupt:
        print("\n Process interrupted by user")
        print(" Saving collected data so far...")
    except Exception as e:
        print(f"\n Unexpected error: {e}")
        print(" Saving collected data so far...")
    finally:
        driver.quit()
        print("\n Selenium driver closed")
    
    # Guardar resultados finales
    if all_career_summaries:
        # Guardar resumen de carreras
        df_summary = pd.DataFrame(all_career_summaries)
        df_summary.to_csv("players_career_summary_ALLCOMPS.csv", index=False, encoding='utf-8')
        
        # Guardar información adicional de jugadores
        df_player_info = pd.DataFrame(all_player_info)
        df_player_info.to_csv("players_additional_info.csv", index=False, encoding='utf-8')
        
        # Guardar detalle de temporadas
        if all_seasons_detailed:
            df_seasons = pd.DataFrame(all_seasons_detailed)
            df_seasons.to_csv(OUTPUT_CSV, index=False, encoding='utf-8')
            
            # Mostrar preview final
            print(f"\n FINAL EXTRACTION SAMPLE (first 10 rows):")
            preview_columns = ['player_name', 'season', 'squad', 'competition', 'matches_played', 'minutes', 'goals', 'assists']
            available_preview = [col for col in preview_columns if col in df_seasons.columns]
            print(df_seasons[available_preview].head(10).to_string(index=False))
        
        # Resumen final detallado
        print("\n" + "=" * 70)
        print(" FINAL EXTRACTION SUMMARY")
        print("=" * 70)
        
        total_processed = len(all_career_summaries)
        successful = len([c for c in all_career_summaries if c['status'] == 'Success'])
        failed = total_processed - successful
        
        print(f" Total players processed: {total_processed}")
        print(f" Successful extractions: {successful}")
        print(f"Failed extractions: {failed}")
        print(f" Total seasons extracted: {len(all_seasons_detailed)}")
        
        
        if total_processed > 0:
            success_rate = (successful / total_processed) * 100
            print(f" Success rate: {success_rate:.1f}%")
        
        print(f"\n Files saved:")
        print(f"   - players_career_summary_ALLCOMPS.csv")
        print(f"   - players_additional_info.csv") 
        print(f"   - {OUTPUT_CSV}")
        
        
        if successful > 0:
            successful_players = [c for c in all_career_summaries if c['status'] == 'Success']
            print(f"\n Sample of successful extractions:")
            for i, player in enumerate(successful_players[:5]):
                print(f"   {i+1}. {player['player_name']}: {player['total_seasons']} seasons, {player['total_clubs']} clubs")

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from bs4 import BeautifulSoup
import re
import time
import random
import io

# ========= CONFIGURATION =========
INPUT_CSV = "Second_scrap/players_fbref.csv"
OUTPUT_CSV = "Second_scrap/players_premier_history2.csv"

def setup_selenium_driver():
    """Setup Selenium WebDriver"""
    chrome_options = Options()
    chrome_options.add_argument("--headless")
    chrome_options.add_argument("--no-sandbox")
    chrome_options.add_argument("--disable-dev-shm-usage")
    chrome_options.add_argument("--disable-blink-features=AutomationControlled")
    chrome_options.add_experimental_option("excludeSwitches", ["enable-automation"])
    chrome_options.add_experimental_option('useAutomationExtension', False)
    chrome_options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36")
    
    driver = webdriver.Chrome(options=chrome_options)
    driver.execute_script("Object.defineProperty(navigator, 'webdriver', {get: () => undefined})")
    return driver

def convert_to_all_comps_url(player_url):
    """Converts normal URL to all competitions URL"""
    if 'Not found' in player_url or not player_url:
        return player_url
    
    if '/all_comps/' in player_url:
        return player_url
    
    try:
        parts = player_url.split('/')
        player_id = parts[5]
        player_slug = parts[6].split('-Stats')[0] if '-Stats' in parts[6] else parts[6]
        all_comps_url = f"https://fbref.com/en/players/{player_id}/all_comps/{player_slug}-Stats---All-Competitions"
        return all_comps_url
    except:
        return player_url

def extract_career_history(driver, player_url, player_name, player_slug):
    """Extracts complete career history of the player"""
    print(f"🔍 Extracting career history: {player_name}")
    
    career_data = {
        'player_name': player_name,
        'player_slug': player_slug,
        'player_url': player_url,
        'total_seasons': 0,
        'total_clubs': 0,
        'career_years': '',
        'status': 'Success'
    }
    
    all_seasons = []
    
    try:
        if player_url == 'Not found' or not player_url:
            career_data['status'] = 'No URL'
            return career_data, all_seasons
        
        all_comps_url = convert_to_all_comps_url(player_url)
        print(f"   📊 Using all competitions URL: {all_comps_url}")
        
        driver.get(all_comps_url)
        time.sleep(4)
        
        soup = BeautifulSoup(driver.page_source, 'html.parser')
        
        # ========= EXTRACT MAIN CAREER TABLE =========
        career_table = None
        
        stats_tables = soup.find_all('table', id=re.compile(r'stats'))
        
        for table in stats_tables:
            try:
                table_html = str(table)
                df = pd.read_html(io.StringIO(table_html))[0]
                
                has_season = any('Season' in str(col) for col in df.columns)
                has_squad = any('Squad' in str(col) for col in df.columns)
                has_comp = any('Comp' in str(col) for col in df.columns)
                
                if has_season and has_squad and has_comp:
                    career_table = df
                    print(f"    Found ALL COMPETITIONS career table with {len(df)} rows")
                    break
                    
            except Exception as e:
                continue
        
        if career_table is not None:
            seasons_processed = process_complex_career_table(career_table, player_name, player_url, all_seasons)
            
            career_data['total_seasons'] = len(seasons_processed)
            career_data['total_clubs'] = len(set([s['squad'] for s in seasons_processed if s['squad']]))
            
            if seasons_processed:
                first_season = seasons_processed[0]['season']
                last_season = seasons_processed[-1]['season']
                career_data['career_years'] = f"{first_season} to {last_season}"
            
            print(f"   Career: {career_data['total_seasons']} seasons, {career_data['total_clubs']} clubs")
            
        else:
            print("    No career table found")
            career_data['status'] = 'No career table'
        
        return career_data, all_seasons
        
    except Exception as e:
        print(f"    Error: {player_name} - {str(e)[:100]}")
        career_data['status'] = f'Error: {str(e)[:100]}'
        return career_data, []

def process_complex_career_table(career_table, player_name, player_url, all_seasons):
    """Processes complex multi-index table"""
    seasons_processed = []
    
    try:
        print(f"    Processing complex multi-index table")
        print(f"   Table shape: {career_table.shape}")
        
        # Strategy: Flatten the multi-index to make it more manageable
        if isinstance(career_table.columns, pd.MultiIndex):
            # Create simpler column names
            new_columns = []
            for col in career_table.columns:
                if pd.isna(col[0]) or col[0] == '':
                    new_columns.append(col[1])
                else:
                    new_columns.append(f"{col[0]}_{col[1]}")
            
            career_table.columns = new_columns
            print(f"    Flattened columns: {new_columns[:10]}...")  # Show first 10
        
        # Find key columns by patterns
        column_mapping = find_columns_by_patterns(career_table)
        
        print(f"    Column mapping:")
        for key, value in column_mapping.items():
            print(f"      {key}: {value}")
        
        # Process each row
        for idx, row in career_table.iterrows():
            season_data = extract_from_complex_row(row, column_mapping)
            
            if season_data and season_data['season'] and season_data['season'] != 'Career' and 'Season' not in season_data['season']:
                season_data['player_name'] = player_name
                season_data['player_url'] = player_url  # ADDED: Include player URL
                seasons_processed.append(season_data)
                all_seasons.append(season_data)
        
        return seasons_processed
        
    except Exception as e:
        print(f"    Error processing complex table: {e}")
        import traceback
        traceback.print_exc()
        return []

def find_columns_by_patterns(career_table):
    """Finds columns by specific patterns"""
    mapping = {
        'season': None, 'squad': None, 'competition': None, 'age': None,
        'matches_played': None, 'minutes': None, 'goals': None, 'assists': None
    }
    
    columns_list = list(career_table.columns)
    
    # Strategy 1: Search by specific names
    for i, col in enumerate(columns_list):
        col_str = str(col).lower()
        
        # Basic columns
        if 'season' in col_str:
            mapping['season'] = col
        elif 'squad' in col_str:
            mapping['squad'] = col
        elif 'comp' in col_str and 'country' not in col_str:
            mapping['competition'] = col
        elif 'age' in col_str:
            mapping['age'] = col
        
        # Playing Time stats
        elif 'mp' == col_str or ('playing' in col_str and 'mp' in col_str):
            mapping['matches_played'] = col
        elif 'min' == col_str or ('playing' in col_str and 'min' in col_str and '90s' not in col_str):
            mapping['minutes'] = col
        
        # Performance stats
        elif 'gls' == col_str or ('performance' in col_str and 'gls' in col_str):
            mapping['goals'] = col
        elif 'ast' == col_str or ('performance' in col_str and 'ast' in col_str):
            mapping['assists'] = col
    
    # Strategy 2: If not found by name, use fixed positions based on the structure we saw
    if mapping['season'] is None and len(columns_list) > 0:
        mapping['season'] = columns_list[0]
    if mapping['age'] is None and len(columns_list) > 1:
        mapping['age'] = columns_list[1]
    if mapping['squad'] is None and len(columns_list) > 2:
        mapping['squad'] = columns_list[2]
    if mapping['competition'] is None and len(columns_list) > 4:
        mapping['competition'] = columns_list[4]
    
    # For MP and Min - positions based on the shown structure
    if mapping['matches_played'] is None:
        # Search in the first columns after the basics
        for i, col in enumerate(columns_list):
            col_str = str(col).lower()
            if col_str == 'mp' or col_str.endswith('_mp'):
                mapping['matches_played'] = col
                break
        if mapping['matches_played'] is None and len(columns_list) > 5:
            mapping['matches_played'] = columns_list[5]  # Typical position for MP
    
    if mapping['minutes'] is None:
        for i, col in enumerate(columns_list):
            col_str = str(col).lower()
            if col_str == 'min' or col_str.endswith('_min'):
                mapping['minutes'] = col
                break
        if mapping['minutes'] is None and len(columns_list) > 6:
            mapping['minutes'] = columns_list[6]  # Typical position for Min
    
    if mapping['goals'] is None:
        for i, col in enumerate(columns_list):
            col_str = str(col).lower()
            if col_str == 'gls' or col_str.endswith('_gls'):
                mapping['goals'] = col
                break
    
    if mapping['assists'] is None:
        for i, col in enumerate(columns_list):
            col_str = str(col).lower()
            if col_str == 'ast' or col_str.endswith('_ast'):
                mapping['assists'] = col
                break
    
    return mapping

def extract_from_complex_row(row, column_mapping):
    """Extracts data from a complex row"""
    try:
        # Get basic values
        season_val = str(row[column_mapping['season']]) if column_mapping['season'] in row and pd.notna(row[column_mapping['season']]) else ''
        
        # Filter unwanted rows
        if (not season_val or season_val == 'nan' or season_val == '' or 
            'Season' in season_val or 'Career' in season_val or 'Seasons' in season_val):
            return None
        
        def get_value(col_key):
            col = column_mapping[col_key]
            if col is None or col not in row:
                return ''
            try:
                value = row[col]
                if pd.isna(value):
                    return ''
                # For minutes, keep commas
                if col_key == 'minutes':
                    return str(value)
                # For other numeric values, clean
                return str(value).replace(',', '').strip()
            except:
                return ''
        
        season_data = {
            'season': get_value('season'),
            'squad': get_value('squad'),
            'competition': get_value('competition'),
            'age': get_value('age'),
            'matches_played': get_value('matches_played'),
            'minutes': get_value('minutes'),
            'goals': get_value('goals'),
            'assists': get_value('assists'),
            # Note: player_name and player_url will be added in the calling function
        }
        
        # Validate that we have basic data
        if not season_data['season'] or not season_data['competition']:
            return None
        
        return season_data
        
    except Exception as e:
        print(f"    Error extracting from complex row: {e}")
        return None

def extract_additional_player_info(driver, player_url, player_name):
    """Extracts additional player information"""
    try:
        driver.get(player_url)
        time.sleep(2)
        
        soup = BeautifulSoup(driver.page_source, 'html.parser')
        
        info = {
            'player_name': player_name,
            'current_club': 'Not found',
            'position': 'Not found',
            'age': 'Not found',
            'nationality': 'Not found'
        }
        
        meta_div = soup.find('div', {'id': 'meta'})
        if meta_div:
            meta_text = meta_div.get_text()
            
            club_match = re.search(r'Club:\s*([^\n]+)', meta_text)
            if club_match:
                info['current_club'] = club_match.group(1).strip()
            
            position_match = re.search(r'Position:\s*([^\n▪]+)', meta_text)
            if position_match:
                info['position'] = position_match.group(1).strip()
            
            age_match = re.search(r'Age:\s*([^\s\)]+)', meta_text)
            if age_match:
                info['age'] = age_match.group(1).strip()
            
            national_match = re.search(r'National Team:\s*([^\n]+)', meta_text)
            if national_match:
                nationality_text = national_match.group(1).strip()
                info['nationality'] = re.sub(r'\s+[a-z]{2}$', '', nationality_text)
        
        return info
        
    except:
        return {
            'player_name': player_name,
            'current_club': 'Not found', 
            'position': 'Not found', 
            'age': 'Not found',
            'nationality': 'Not found'
        }

def main():
    """Extracts career history of all players"""
    print(" EXTRACTING PLAYER CAREER HISTORY (ALL COMPETITIONS)")
    print("=" * 70)
    
    try:
        df_existing = pd.read_csv(INPUT_CSV)
        print(f" Loaded {len(df_existing)} players with URLs")
        
        players_to_process = df_existing[
            (df_existing['player_url'] != 'Not found') & 
            (df_existing['player_url'].notna())
        ]
        
        print(f" Players to process: {len(players_to_process)}")
        
    except Exception as e:
        print(f"Error reading CSV: {e}")
        return
    
    driver = setup_selenium_driver()
    
    all_career_summaries = []
    all_seasons_detailed = []
    all_player_info = []
    
    try:
        # Process ALL players
        test_players = players_to_process
        
        total_players = len(test_players)
        
        for index, row in test_players.iterrows():
            player_name = str(row.get('player_name', '')).strip()
            player_slug = str(row.get('player_slug', '')).strip()
            player_url = str(row.get('player_url', '')).strip()
            
            print(f"\n{index + 1}/{total_players}: {player_name}")
            
            # Extract career history
            career_summary, seasons_data = extract_career_history(driver, player_url, player_name, player_slug)
            
            # Extract additional information
            additional_info = extract_additional_player_info(driver, player_url, player_name)
            
            # Combine information
            career_summary.update(additional_info)
            all_career_summaries.append(career_summary)
            all_player_info.append(additional_info)
            all_seasons_detailed.extend(seasons_data)
            
            # Show progress every 10 players
            if (index + 1) % 10 == 0:
                successful = len([c for c in all_career_summaries if c['status'] == 'Success'])
                print(f"\n Progress: {index + 1}/{total_players}, {successful} successful")
                
                # Save temporary checkpoint
                if all_seasons_detailed:
                    df_temp = pd.DataFrame(all_seasons_detailed)
                    df_temp.to_csv(f"players_career_history_CHECKPOINT_{index+1}.csv", index=False, encoding='utf-8')
                    print(f" Checkpoint saved: players_career_history_CHECKPOINT_{index+1}.csv")
            
            # Random delay to avoid blocking
            time.sleep(random.uniform(2, 4))
    
    except KeyboardInterrupt:
        print("\n Process interrupted by user")
        print(" Saving collected data so far...")
    except Exception as e:
        print(f"\n Unexpected error: {e}")
        print(" Saving collected data so far...")
    finally:
        driver.quit()
        print("\n Selenium driver closed")
    
    # Save final results
    if all_career_summaries:
        # Save career summaries
        df_summary = pd.DataFrame(all_career_summaries)
        df_summary.to_csv("players_career_summary_ALLCOMPS.csv", index=False, encoding='utf-8')
        
        # Save additional player information
        df_player_info = pd.DataFrame(all_player_info)
        df_player_info.to_csv("players_additional_info.csv", index=False, encoding='utf-8')
        
        # Save season details (with player_url included)
        if all_seasons_detailed:
            df_seasons = pd.DataFrame(all_seasons_detailed)
            df_seasons.to_csv(OUTPUT_CSV, index=False, encoding='utf-8')
            
            # Show final preview - NOW INCLUDES player_url
            print(f"\n FINAL EXTRACTION SAMPLE (first 10 rows):")
            preview_columns = ['player_name', 'season', 'squad', 'competition', 'matches_played', 'minutes', 'goals', 'assists', 'player_url']
            available_preview = [col for col in preview_columns if col in df_seasons.columns]
            print(df_seasons[available_preview].head(10).to_string(index=False))
        
        # Detailed final summary
        print("\n" + "=" * 70)
        print(" FINAL EXTRACTION SUMMARY")
        print("=" * 70)
        
        total_processed = len(all_career_summaries)
        successful = len([c for c in all_career_summaries if c['status'] == 'Success'])
        failed = total_processed - successful
        
        print(f" Total players processed: {total_processed}")
        print(f" Successful extractions: {successful}")
        print(f" Failed extractions: {failed}")
        print(f" Total seasons extracted: {len(all_seasons_detailed)}")
        
        # Success statistics
        if total_processed > 0:
            success_rate = (successful / total_processed) * 100
            print(f" Success rate: {success_rate:.1f}%")
        
        print(f"\n Files saved:")
        print(f"   - players_career_summary_ALLCOMPS.csv")
        print(f"   - players_additional_info.csv") 
        print(f"   - {OUTPUT_CSV}")
        
        # Show some successful examples
        if successful > 0:
            successful_players = [c for c in all_career_summaries if c['status'] == 'Success']
            print(f"\n Sample of successful extractions:")
            for i, player in enumerate(successful_players[:5]):
                print(f"   {i+1}. {player['player_name']}: {player['total_seasons']} seasons, {player['total_clubs']} clubs")

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import re
import time
import random

# ========= CONFIGURATION =========

INPUT_CSV = "Second_scrap/tran_url_premier.csv"  
OUTPUT_CSV = "Second_scrap/players_with_deceased_status.csv"

HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/141.0.0.0 Safari/537.36"
}

def check_deceased_status(player_url, player_name):
    """Check if a player is deceased and extract death date if available"""
    print(f"🔍 Checking: {player_name}")
    
    deceased_data = {
        'deceased': 'No',
        'death_date': 'Not found',
        'death_age': 'Not found'
    }
    
    try:
        response = requests.get(player_url, headers=HEADERS, timeout=10)
        if response.status_code != 200:
            print(f"    HTTP Error {response.status_code}")
            return deceased_data
        
        soup = BeautifulSoup(response.content, "html.parser")
        
        # Method 1: Look for death information in the data grid
        data_grid = soup.find('div', class_='info-table')
        if data_grid:
            grid_text = data_grid.get_text()
            
            # Look for death date patterns
            death_match = re.search(r'Date of death:\s*([^\n]+)', grid_text)
            if death_match:
                deceased_data['deceased'] = 'Yes'
                death_info = death_match.group(1).strip()
                
                # Extract date and age if available
                date_age_match = re.search(r'(\d{2}\.\d{2}\.\d{4})\s*\((\d+)\)', death_info)
                if date_age_match:
                    deceased_data['death_date'] = date_age_match.group(1)  # 01.06.2019
                    deceased_data['death_age'] = date_age_match.group(2)   # 35
                    print(f"    Deceased: {deceased_data['death_date']} (age {deceased_data['death_age']})")
                else:
                    deceased_data['death_date'] = death_info
                    print(f"    Deceased: {death_info}")
                
                return deceased_data
        
        # Method 2: Look for "Date of death" in labels
        data_labels = soup.find_all('span', class_='data-header__label')
        for label in data_labels:
            label_text = label.get_text(strip=True)
            if 'Date of death' in label_text:
                deceased_data['deceased'] = 'Yes'
                death_content = label.find_next('span', class_='data-header__content')
                if death_content:
                    death_info = death_content.get_text(strip=True)
                    
                    # Extract date and age
                    date_age_match = re.search(r'(\d{2}\.\d{2}\.\d{4})\s*\((\d+)\)', death_info)
                    if date_age_match:
                        deceased_data['death_date'] = date_age_match.group(1)
                        deceased_data['death_age'] = date_age_match.group(2)
                        print(f"    Deceased: {deceased_data['death_date']} (age {deceased_data['death_age']})")
                    else:
                        deceased_data['death_date'] = death_info
                        print(f"    Deceased: {death_info}")
                
                return deceased_data
        
        # Method 3: Check for memorial indicators in the page
        page_text = soup.get_text()
        death_indicators = ['date of death', 'passed away', 'deceased', 'died', 'in memoriam']
        if any(indicator in page_text.lower() for indicator in death_indicators):
            deceased_data['deceased'] = 'Yes'
            print(f"    Deceased (found memorial text)")
        
        if deceased_data['deceased'] == 'No':
            print(f"    Alive")
        
        return deceased_data
        
    except Exception as e:
        print(f"    Error: {str(e)[:50]}")
        return deceased_data

def add_deceased_status_to_csv():
    """Add deceased status to your existing CSV"""
    print(" ADDING DECEASED STATUS TO PLAYERS")
    print("=" * 60)
    
    try:
        df = pd.read_csv(INPUT_CSV)
        print(f" Loaded {len(df)} players")
        
        # Add new columns
        df['deceased'] = 'No'
        df['death_date'] = 'Not found'
        df['death_age'] = 'Not found'
        
        deceased_count = 0
        
        for index, row in df.iterrows():
            player_name = str(row.get('player_name', '')).strip()
            player_url = str(row.get('transfermarkt_url', '')).strip()
            
            print(f"\n[{index + 1}/{len(df)}] {player_name}")
            
            # Skip if no valid URL
            if player_url == 'Not found' or 'transfermarkt.com' not in player_url:
                print(f"     No valid URL, skipping")
                continue
            
            # Check deceased status
            deceased_data = check_deceased_status(player_url, player_name)
            
            # Update the row
            df.at[index, 'deceased'] = deceased_data['deceased']
            df.at[index, 'death_date'] = deceased_data['death_date']
            df.at[index, 'death_age'] = deceased_data['death_age']
            
            if deceased_data['deceased'] == 'Yes':
                deceased_count += 1
            
            # Save progress every 20 players
            if (index + 1) % 20 == 0:
                df.to_csv(f"progress_deceased_{index + 1}.csv", index=False)
                print(f" Progress saved: {index + 1} players")
            
            # Delay between requests
            if index + 1 < len(df):
                delay = random.uniform(2, 4)
                time.sleep(delay)
        
        # Save final results
        df.to_csv(OUTPUT_CSV, index=False, encoding='utf-8')
        
        print(f"\n COMPLETED!")
        print(f" DECEASED PLAYERS FOUND: {deceased_count}")
        print(f" Results saved to: {OUTPUT_CSV}")
        
        # Show deceased players
        if deceased_count > 0:
            print(f"\n DECEASED PLAYERS:")
            deceased_players = df[df['deceased'] == 'Yes']
            for _, player in deceased_players.iterrows():
                print(f"    {player['player_name']} - {player['death_date']} (age {player['death_age']})")
        
        return df
        
    except Exception as e:
        print(f" Error: {e}")
        return None

# ========= RUN THE CODE =========
if __name__ == "__main__":
    add_deceased_status_to_csv()
```

```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from bs4 import BeautifulSoup
import re
import time
import random
import io

# ========= CONFIGURATION =========
INPUT_CSV = "Second_scrap/LIGA/players_fbref_liga.csv"

OUTPUT_CSV = "Second_scrap/LIGA/players_LIGA_history2.csv"

def setup_selenium_driver():
    """Setup Selenium WebDriver"""
    chrome_options = Options()
    chrome_options.add_argument("--headless")
    chrome_options.add_argument("--no-sandbox")
    chrome_options.add_argument("--disable-dev-shm-usage")
    chrome_options.add_argument("--disable-blink-features=AutomationControlled")
    chrome_options.add_experimental_option("excludeSwitches", ["enable-automation"])
    chrome_options.add_experimental_option('useAutomationExtension', False)
    chrome_options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36")
    
    driver = webdriver.Chrome(options=chrome_options)
    driver.execute_script("Object.defineProperty(navigator, 'webdriver', {get: () => undefined})")
    return driver

def convert_to_all_comps_url(player_url):
    """Converts normal URL to all competitions URL"""
    if 'Not found' in player_url or not player_url:
        return player_url
    
    if '/all_comps/' in player_url:
        return player_url
    
    try:
        parts = player_url.split('/')
        player_id = parts[5]
        player_slug = parts[6].split('-Stats')[0] if '-Stats' in parts[6] else parts[6]
        all_comps_url = f"https://fbref.com/en/players/{player_id}/all_comps/{player_slug}-Stats---All-Competitions"
        return all_comps_url
    except:
        return player_url

def extract_career_history(driver, player_url, player_name, player_slug):
    """Extracts complete career history of the player"""
    print(f"🔍 Extracting career history: {player_name}")
    
    career_data = {
        'player_name': player_name,
        'player_slug': player_slug,
        'player_url': player_url,
        'total_seasons': 0,
        'total_clubs': 0,
        'career_years': '',
        'status': 'Success'
    }
    
    all_seasons = []
    
    try:
        if player_url == 'Not found' or not player_url:
            career_data['status'] = 'No URL'
            return career_data, all_seasons
        
        all_comps_url = convert_to_all_comps_url(player_url)
        print(f"   📊 Using all competitions URL: {all_comps_url}")
        
        driver.get(all_comps_url)
        time.sleep(4)
        
        soup = BeautifulSoup(driver.page_source, 'html.parser')
        
        # ========= EXTRACT MAIN CAREER TABLE =========
        career_table = None
        
        stats_tables = soup.find_all('table', id=re.compile(r'stats'))
        
        for table in stats_tables:
            try:
                table_html = str(table)
                df = pd.read_html(io.StringIO(table_html))[0]
                
                has_season = any('Season' in str(col) for col in df.columns)
                has_squad = any('Squad' in str(col) for col in df.columns)
                has_comp = any('Comp' in str(col) for col in df.columns)
                
                if has_season and has_squad and has_comp:
                    career_table = df
                    print(f"   ✅ Found ALL COMPETITIONS career table with {len(df)} rows")
                    break
                    
            except Exception as e:
                continue
        
        if career_table is not None:
            seasons_processed = process_complex_career_table(career_table, player_name, player_url, all_seasons)
            
            career_data['total_seasons'] = len(seasons_processed)
            career_data['total_clubs'] = len(set([s['squad'] for s in seasons_processed if s['squad']]))
            
            if seasons_processed:
                first_season = seasons_processed[0]['season']
                last_season = seasons_processed[-1]['season']
                career_data['career_years'] = f"{first_season} to {last_season}"
            
            print(f"    Career: {career_data['total_seasons']} seasons, {career_data['total_clubs']} clubs")
            
        else:
            print("    No career table found")
            career_data['status'] = 'No career table'
        
        return career_data, all_seasons
        
    except Exception as e:
        print(f"   Error: {player_name} - {str(e)[:100]}")
        career_data['status'] = f'Error: {str(e)[:100]}'
        return career_data, []

def process_complex_career_table(career_table, player_name, player_url, all_seasons):
    """Processes complex multi-index table"""
    seasons_processed = []
    
    try:
        print(f"    Processing complex multi-index table")
        print(f"   Table shape: {career_table.shape}")
        
        # Strategy: Flatten the multi-index to make it more manageable
        if isinstance(career_table.columns, pd.MultiIndex):
            # Create simpler column names
            new_columns = []
            for col in career_table.columns:
                if pd.isna(col[0]) or col[0] == '':
                    new_columns.append(col[1])
                else:
                    new_columns.append(f"{col[0]}_{col[1]}")
            
            career_table.columns = new_columns
            print(f"    Flattened columns: {new_columns[:10]}...")  # Show first 10
        
        # Find key columns by patterns
        column_mapping = find_columns_by_patterns(career_table)
        
        print(f"    Column mapping:")
        for key, value in column_mapping.items():
            print(f"      {key}: {value}")
        
        # Process each row
        for idx, row in career_table.iterrows():
            season_data = extract_from_complex_row(row, column_mapping)
            
            if season_data and season_data['season'] and season_data['season'] != 'Career' and 'Season' not in season_data['season']:
                season_data['player_name'] = player_name
                season_data['player_url'] = player_url  # ADDED: Include player URL
                seasons_processed.append(season_data)
                all_seasons.append(season_data)
        
        return seasons_processed
        
    except Exception as e:
        print(f"    Error processing complex table: {e}")
        import traceback
        traceback.print_exc()
        return []

def find_columns_by_patterns(career_table):
    """Finds columns by specific patterns"""
    mapping = {
        'season': None, 'squad': None, 'competition': None, 'age': None,
        'matches_played': None, 'minutes': None, 'goals': None, 'assists': None
    }
    
    columns_list = list(career_table.columns)
    
    # Strategy 1: Search by specific names
    for i, col in enumerate(columns_list):
        col_str = str(col).lower()
        
        # Basic columns
        if 'season' in col_str:
            mapping['season'] = col
        elif 'squad' in col_str:
            mapping['squad'] = col
        elif 'comp' in col_str and 'country' not in col_str:
            mapping['competition'] = col
        elif 'age' in col_str:
            mapping['age'] = col
        
        # Playing Time stats
        elif 'mp' == col_str or ('playing' in col_str and 'mp' in col_str):
            mapping['matches_played'] = col
        elif 'min' == col_str or ('playing' in col_str and 'min' in col_str and '90s' not in col_str):
            mapping['minutes'] = col
        
        # Performance stats
        elif 'gls' == col_str or ('performance' in col_str and 'gls' in col_str):
            mapping['goals'] = col
        elif 'ast' == col_str or ('performance' in col_str and 'ast' in col_str):
            mapping['assists'] = col
    
    # Strategy 2: If not found by name, use fixed positions based on the structure we saw
    if mapping['season'] is None and len(columns_list) > 0:
        mapping['season'] = columns_list[0]
    if mapping['age'] is None and len(columns_list) > 1:
        mapping['age'] = columns_list[1]
    if mapping['squad'] is None and len(columns_list) > 2:
        mapping['squad'] = columns_list[2]
    if mapping['competition'] is None and len(columns_list) > 4:
        mapping['competition'] = columns_list[4]
    
    # For MP and Min - positions based on the shown structure
    if mapping['matches_played'] is None:
        # Search in the first columns after the basics
        for i, col in enumerate(columns_list):
            col_str = str(col).lower()
            if col_str == 'mp' or col_str.endswith('_mp'):
                mapping['matches_played'] = col
                break
        if mapping['matches_played'] is None and len(columns_list) > 5:
            mapping['matches_played'] = columns_list[5]  # Typical position for MP
    
    if mapping['minutes'] is None:
        for i, col in enumerate(columns_list):
            col_str = str(col).lower()
            if col_str == 'min' or col_str.endswith('_min'):
                mapping['minutes'] = col
                break
        if mapping['minutes'] is None and len(columns_list) > 6:
            mapping['minutes'] = columns_list[6]  # Typical position for Min
    
    if mapping['goals'] is None:
        for i, col in enumerate(columns_list):
            col_str = str(col).lower()
            if col_str == 'gls' or col_str.endswith('_gls'):
                mapping['goals'] = col
                break
    
    if mapping['assists'] is None:
        for i, col in enumerate(columns_list):
            col_str = str(col).lower()
            if col_str == 'ast' or col_str.endswith('_ast'):
                mapping['assists'] = col
                break
    
    return mapping

def extract_from_complex_row(row, column_mapping):
    """Extracts data from a complex row"""
    try:
        # Get basic values
        season_val = str(row[column_mapping['season']]) if column_mapping['season'] in row and pd.notna(row[column_mapping['season']]) else ''
        
        # Filter unwanted rows
        if (not season_val or season_val == 'nan' or season_val == '' or 
            'Season' in season_val or 'Career' in season_val or 'Seasons' in season_val):
            return None
        
        def get_value(col_key):
            col = column_mapping[col_key]
            if col is None or col not in row:
                return ''
            try:
                value = row[col]
                if pd.isna(value):
                    return ''
                # For minutes, keep commas
                if col_key == 'minutes':
                    return str(value)
                # For other numeric values, clean
                return str(value).replace(',', '').strip()
            except:
                return ''
        
        season_data = {
            'season': get_value('season'),
            'squad': get_value('squad'),
            'competition': get_value('competition'),
            'age': get_value('age'),
            'matches_played': get_value('matches_played'),
            'minutes': get_value('minutes'),
            'goals': get_value('goals'),
            'assists': get_value('assists'),
            # Note: player_name and player_url will be added in the calling function
        }
        
        # Validate that we have basic data
        if not season_data['season'] or not season_data['competition']:
            return None
        
        return season_data
        
    except Exception as e:
        print(f"    Error extracting from complex row: {e}")
        return None

def extract_additional_player_info(driver, player_url, player_name):
    """Extracts additional player information"""
    try:
        driver.get(player_url)
        time.sleep(2)
        
        soup = BeautifulSoup(driver.page_source, 'html.parser')
        
        info = {
            'player_name': player_name,
            'current_club': 'Not found',
            'position': 'Not found',
            'age': 'Not found',
            'nationality': 'Not found'
        }
        
        meta_div = soup.find('div', {'id': 'meta'})
        if meta_div:
            meta_text = meta_div.get_text()
            
            club_match = re.search(r'Club:\s*([^\n]+)', meta_text)
            if club_match:
                info['current_club'] = club_match.group(1).strip()
            
            position_match = re.search(r'Position:\s*([^\n▪]+)', meta_text)
            if position_match:
                info['position'] = position_match.group(1).strip()
            
            age_match = re.search(r'Age:\s*([^\s\)]+)', meta_text)
            if age_match:
                info['age'] = age_match.group(1).strip()
            
            national_match = re.search(r'National Team:\s*([^\n]+)', meta_text)
            if national_match:
                nationality_text = national_match.group(1).strip()
                info['nationality'] = re.sub(r'\s+[a-z]{2}$', '', nationality_text)
        
        return info
        
    except:
        return {
            'player_name': player_name,
            'current_club': 'Not found', 
            'position': 'Not found', 
            'age': 'Not found',
            'nationality': 'Not found'
        }

def main():
    """Extracts career history of all players"""
    print(" EXTRACTING PLAYER CAREER HISTORY (ALL COMPETITIONS)")
    print("=" * 70)
    
    try:
        df_existing = pd.read_csv(INPUT_CSV)
        print(f" Loaded {len(df_existing)} players with URLs")
        
        players_to_process = df_existing[
            (df_existing['player_url'] != 'Not found') & 
            (df_existing['player_url'].notna())
        ]
        
        print(f" Players to process: {len(players_to_process)}")
        
    except Exception as e:
        print(f" Error reading CSV: {e}")
        return
    
    driver = setup_selenium_driver()
    
    all_career_summaries = []
    all_seasons_detailed = []
    all_player_info = []
    
    try:
        # Process ALL players
        test_players = players_to_process
        
        total_players = len(test_players)
        
        for index, row in test_players.iterrows():
            player_name = str(row.get('player_name', '')).strip()
            player_slug = str(row.get('player_slug', '')).strip()
            player_url = str(row.get('player_url', '')).strip()
            
            print(f"\n{index + 1}/{total_players}: {player_name}")
            
            # Extract career history
            career_summary, seasons_data = extract_career_history(driver, player_url, player_name, player_slug)
            
            # Extract additional information
            additional_info = extract_additional_player_info(driver, player_url, player_name)
            
            # Combine information
            career_summary.update(additional_info)
            all_career_summaries.append(career_summary)
            all_player_info.append(additional_info)
            all_seasons_detailed.extend(seasons_data)
            
            # Show progress every 10 players
            if (index + 1) % 10 == 0:
                successful = len([c for c in all_career_summaries if c['status'] == 'Success'])
                print(f"\n📊 Progress: {index + 1}/{total_players}, {successful} successful")
                
                # Save temporary checkpoint
                if all_seasons_detailed:
                    df_temp = pd.DataFrame(all_seasons_detailed)
                    df_temp.to_csv(f"players_career_history_CHECKPOINT_{index+1}.csv", index=False, encoding='utf-8')
                    print(f"💾 Checkpoint saved: players_career_history_CHECKPOINT_{index+1}.csv")
            
            # Random delay to avoid blocking
            time.sleep(random.uniform(2, 4))
    
    except KeyboardInterrupt:
        print("\n Process interrupted by user")
        print(" Saving collected data so far...")
    except Exception as e:
        print(f"\n Unexpected error: {e}")
        print(" Saving collected data so far...")
    finally:
        driver.quit()
        print("\n Selenium driver closed")
    
    # Save final results
    if all_career_summaries:
        # Save career summaries
        df_summary = pd.DataFrame(all_career_summaries)
        df_summary.to_csv("players_career_summary_ALLCOMPS.csv", index=False, encoding='utf-8')
        
        # Save additional player information
        df_player_info = pd.DataFrame(all_player_info)
        df_player_info.to_csv("players_additional_info.csv", index=False, encoding='utf-8')
        
        # Save season details (with player_url included)
        if all_seasons_detailed:
            df_seasons = pd.DataFrame(all_seasons_detailed)
            df_seasons.to_csv(OUTPUT_CSV, index=False, encoding='utf-8')
            
            # Show final preview - NOW INCLUDES player_url
            print(f"\n FINAL EXTRACTION SAMPLE (first 10 rows):")
            preview_columns = ['player_name', 'season', 'squad', 'competition', 'matches_played', 'minutes', 'goals', 'assists', 'player_url']
            available_preview = [col for col in preview_columns if col in df_seasons.columns]
            print(df_seasons[available_preview].head(10).to_string(index=False))
        
        # Detailed final summary
        print("\n" + "=" * 70)
        print(" FINAL EXTRACTION SUMMARY")
        print("=" * 70)
        
        total_processed = len(all_career_summaries)
        successful = len([c for c in all_career_summaries if c['status'] == 'Success'])
        failed = total_processed - successful
        
        print(f" Total players processed: {total_processed}")
        print(f" Successful extractions: {successful}")
        print(f" Failed extractions: {failed}")
        print(f" Total seasons extracted: {len(all_seasons_detailed)}")
        
        # Success statistics
        if total_processed > 0:
            success_rate = (successful / total_processed) * 100
            print(f" Success rate: {success_rate:.1f}%")
        
        print(f"\n Files saved:")
        print(f"   - players_career_summary_ALLCOMPS.csv")
        print(f"   - players_additional_info.csv") 
        print(f"   - {OUTPUT_CSV}")
        
        # Show some successful examples
        if successful > 0:
            successful_players = [c for c in all_career_summaries if c['status'] == 'Success']
            print(f"\n Sample of successful extractions:")
            for i, player in enumerate(successful_players[:5]):
                print(f"   {i+1}. {player['player_name']}: {player['total_seasons']} seasons, {player['total_clubs']} clubs")

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from bs4 import BeautifulSoup
import re
import time
import random
import io

# ========= CONFIGURATION =========
INPUT_CSV = "Second_scrap/players_fbref2.csv"
OUTPUT_CSV = "Second_scrap/players_premier_national.csv"


def setup_stealth_driver():
    """Setup Chrome driver with stealth options"""
    options = Options()
    
    # Remove headless for debugging
    # options.add_argument("--headless")
    
    # Basic options
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    
    # Stealth options
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_experimental_option("excludeSwitches", ["enable-automation"])
    options.add_experimental_option('useAutomationExtension', False)
    
    # Realistic settings
    options.add_argument("--window-size=1920,1080")
    options.add_argument("--disable-extensions")
    options.add_argument("--disable-plugins")
    
    # User agent
    user_agent = "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
    options.add_argument(f"user-agent={user_agent}")
    
    try:
        driver = webdriver.Chrome(options=options)
        
        # Remove webdriver property
        driver.execute_script("Object.defineProperty(navigator, 'webdriver', {get: () => undefined})")
        
        return driver
    except Exception as e:
        print(f" Error setting up driver: {e}")
        return None

def convert_to_national_team_url(player_url):
    """Convierte la URL normal a URL de selección nacional"""
    if 'Not found' in player_url or not player_url:
        return player_url
    
    if '/nat_tm/' in player_url:
        return player_url
    
    try:
        parts = player_url.split('/')
        player_id = parts[5]
        player_slug = parts[6]
        
        # Remove any existing suffix and add national team suffix
        if '-Stats' in player_slug:
            base_slug = player_slug.split('-Stats')[0]
        else:
            base_slug = player_slug
        
        national_team_url = f"https://fbref.com/en/players/{player_id}/nat_tm/{base_slug}-National-Team-Stats"
        return national_team_url
    except Exception as e:
        print(f"    Error converting to national team URL: {e}")
        return player_url

def extract_national_team_data(driver, player_url, player_name, player_slug):
    """Extrae datos de selección nacional"""
    print(f" Extracting national team history: {player_name}")
    
    career_data = {
        'player_name': player_name,
        'player_slug': player_slug,
        'player_url': player_url,
        'has_national_team': 'No',
        'total_seasons_national': 0,
        'total_matches_national': 0,
        'total_goals_national': 0,
        'total_assists_national': 0,
        'national_team': 'None',
        'national_team_years': '',
        'status': 'Success'
    }
    
    all_seasons = []
    
    try:
        if player_url == 'Not found' or not player_url:
            career_data['status'] = 'No URL'
            return career_data, all_seasons
        
        national_team_url = convert_to_national_team_url(player_url)
        print(f"    Using national team URL: {national_team_url}")
        
        # Navigate to the page
        driver.get(national_team_url)
        time.sleep(random.uniform(3, 6))
        
        # Check if we got blocked
        page_source = driver.page_source
        if "Just a moment" in page_source or "Cloudflare" in page_source:
            print("    Cloudflare detected, waiting longer...")
            time.sleep(15)
            page_source = driver.page_source
        
        # Check if page loaded successfully
        if "Page Not Found" in page_source or "404" in driver.title:
            print(f"    Page not found (404)")
            career_data['status'] = 'Page not found'
            return career_data, all_seasons
        
        # Parse the page
        soup = BeautifulSoup(page_source, 'html.parser')
        
        # Look for national team tables
        national_table = find_national_team_table_only(soup)
        
        if national_table:
            seasons_processed = process_national_team_table(national_table, player_name, player_url, all_seasons)
            if seasons_processed:
                return update_career_data(career_data, seasons_processed, all_seasons)
            else:
                career_data['status'] = 'No valid seasons'
        else:
            print(f"    No national team table found")
            career_data['status'] = 'No table found'
        
        return career_data, all_seasons
        
    except Exception as e:
        print(f"    Error: {player_name} - {str(e)}")
        career_data['status'] = f'Error: {str(e)[:100]}'
        return career_data, []

def find_national_team_table_only(soup):
    """Find ONLY national team tables"""
    
    # Method 1: Look for standard stats national team table by exact ID
    table = soup.find('table', id='stats_standard_nat_tm')
    if table:
        print("    Found NATIONAL TEAM table by exact ID: stats_standard_nat_tm")
        return table
    
    # Method 2: Look for tables with national team in parent context
    all_tables = soup.find_all('table')
    for table in all_tables:
        parent = table.find_parent()
        while parent:
            parent_html = str(parent)
            if 'National Team' in parent_html and 'Standard Stats' in parent_html:
                print("    Found NATIONAL TEAM table by parent context")
                return table
            parent = parent.find_parent()
    
    return None

def update_career_data(career_data, seasons_processed, all_seasons):
    """Update career data with national team information"""
    career_data['has_national_team'] = 'Yes'
    career_data['total_seasons_national'] = len(seasons_processed)
    
    # Calculate totals
    total_matches = 0
    total_goals = 0
    total_assists = 0
    for season in seasons_processed:
        if season['matches_played'] and season['matches_played'].isdigit():
            total_matches += int(season['matches_played'])
        if season['goals'] and season['goals'].isdigit():
            total_goals += int(season['goals'])
        if season['assists'] and season['assists'].isdigit():
            total_assists += int(season['assists'])
    
    career_data['total_matches_national'] = total_matches
    career_data['total_goals_national'] = total_goals
    career_data['total_assists_national'] = total_assists
    
    # Get national team info
    if seasons_processed:
        first_team = seasons_processed[0]['squad']
        career_data['national_team'] = first_team
        
        first_season = seasons_processed[0]['season']
        last_season = seasons_processed[-1]['season']
        career_data['national_team_years'] = f"{first_season} to {last_season}"
    
    print(f"    National Team: {career_data['national_team']}")
    print(f"    Seasons: {career_data['total_seasons_national']}, Matches: {career_data['total_matches_national']}, Goals: {career_data['total_goals_national']}, Assists: {career_data['total_assists_national']}")
    
    return career_data, all_seasons

def process_national_team_table(national_table, player_name, player_url, all_seasons):
    """Procesa SOLO la tabla de selección nacional con mapeo CORREGIDO"""
    seasons_processed = []
    
    try:
        # Convert table to DataFrame
        df = pd.read_html(io.StringIO(str(national_table)))[0]
        print(f"    Processing NATIONAL TEAM table with {len(df)} rows")
        
        # Handle multi-index columns
        if isinstance(df.columns, pd.MultiIndex):
            new_columns = []
            for col in df.columns:
                if pd.isna(col[0]) or col[0] == '':
                    new_columns.append(col[1])
                else:
                    new_columns.append(f"{col[0]}_{col[1]}")
            df.columns = new_columns
        
        columns_list = list(df.columns)
        print(f"    NATIONAL TEAM Columns: {columns_list}")
        
        # Find the correct column indices for national team tables
        # CORRECTED: We need the TOTAL goals/assists, not per 90
        column_mapping = find_national_team_columns_corrected(columns_list)
        print(f"    Corrected column mapping: {column_mapping}")
        
        # Process each row
        for idx, row in df.iterrows():
            season_data = extract_from_national_team_row(row, column_mapping, player_url)
            
            if season_data and season_data['season'] and is_national_team_data(season_data):
                season_data['player_name'] = player_name
                seasons_processed.append(season_data)
                all_seasons.append(season_data)
        
        print(f"    Processed {len(seasons_processed)} valid NATIONAL TEAM seasons")
        return seasons_processed
        
    except Exception as e:
        print(f"    Error processing NATIONAL TEAM table: {e}")
        import traceback
        traceback.print_exc()
        return []

def find_national_team_columns_corrected(columns_list):
    """Find the correct columns for national team tables - CORRECTED FOR TOTALS"""
    mapping = {
        'season': None, 'squad': None, 'competition': None, 'age': None,
        'matches_played': None, 'minutes': None, 'goals': None, 'assists': None
    }
    
    # National team table structure:
    # Season, Age, Squad, Comp, LgRank, MP, Starts, Min, 90s, Gls, Ast, G+A, G-PK, PK, PKatt, CrdY, CrdR, Gls/90, Ast/90, ...
    # We want the TOTAL goals (column 9) and TOTAL assists (column 10), not the per90 ones
    
    for i, col in enumerate(columns_list):
        col_str = str(col).lower()
        
        # Basic info columns
        if 'season' in col_str:
            mapping['season'] = i
        elif 'squad' in col_str or 'team' in col_str:
            mapping['squad'] = i
        elif 'comp' in col_str:
            mapping['competition'] = i
        elif 'age' in col_str:
            mapping['age'] = i
        
        # Stats columns - we want the TOTALS, not per90
        elif col_str == 'mp' or col_str.endswith('_mp'):
            mapping['matches_played'] = i
        elif col_str == 'min' or col_str.endswith('_min'):
            mapping['minutes'] = i
        elif (col_str == 'gls' or col_str.endswith('_gls')) and '90' not in col_str:
            # Make sure it's not the per90 column
            mapping['goals'] = i
        elif (col_str == 'ast' or col_str.endswith('_ast')) and '90' not in col_str:
            # Make sure it's not the per90 column
            mapping['assists'] = i
    
    # Fallback: Use typical positions in national team tables for TOTALS
    # Season(0), Age(1), Squad(2), Comp(3), MP(5), Min(7), Gls(9), Ast(10)
    if mapping['season'] is None and len(columns_list) > 0:
        mapping['season'] = 0
    if mapping['age'] is None and len(columns_list) > 1:
        mapping['age'] = 1
    if mapping['squad'] is None and len(columns_list) > 2:
        mapping['squad'] = 2
    if mapping['competition'] is None and len(columns_list) > 3:
        mapping['competition'] = 3
    if mapping['matches_played'] is None and len(columns_list) > 5:
        mapping['matches_played'] = 5
    if mapping['minutes'] is None and len(columns_list) > 7:
        mapping['minutes'] = 7
    if mapping['goals'] is None and len(columns_list) > 9:
        mapping['goals'] = 9
    if mapping['assists'] is None and len(columns_list) > 10:
        mapping['assists'] = 10
    
    return mapping

def is_national_team_data(season_data):
    """Verify that this is actually national team data"""
    squad = season_data.get('squad', '').lower()
    competition = season_data.get('competition', '').lower()
    
    # National team indicators
    national_indicators = [
        'world cup', 'euro', 'copa america', 'africa cup of nations',
        'asian cup', 'gold cup', 'confederations cup', 'wcq', 'qualifying',
        'friendlies', 'nation', 'international'
    ]
    
    # Check if it looks like national team data
    is_national = any(indicator in squad or indicator in competition 
                     for indicator in national_indicators)
    
    return is_national

def extract_from_national_team_row(row, column_mapping, player_url):
    """Extrae datos de una fila de selección nacional con mapeo correcto"""
    try:
        def get_value(col_idx):
            if col_idx is None or col_idx >= len(row):
                return ''
            value = row[col_idx]
            if pd.isna(value):
                return ''
            # Clean the value - remove commas from numbers
            clean_value = str(value).replace(',', '').strip()
            return clean_value
        
        season_val = get_value(column_mapping['season'])
        
        # Skip header and summary rows
        if (not season_val or season_val == 'nan' or 
            'Season' in season_val or 'Career' in season_val or
            'Seasons' in season_val):
            return None
        
        # Only process rows with valid season values (years)
        if not (season_val.isdigit() or '-' in season_val):
            return None
        
        season_data = {
            'season': season_val,
            'squad': get_value(column_mapping['squad']),
            'competition': get_value(column_mapping['competition']),
            'age': get_value(column_mapping['age']),
            'matches_played': get_value(column_mapping['matches_played']),
            'minutes': get_value(column_mapping['minutes']),
            'goals': get_value(column_mapping['goals']),
            'assists': get_value(column_mapping['assists']),
            'player_url': player_url,
            'player_name': ''
        }
        
        # Basic validation
        if not season_data['squad']:
            return None
            
        return season_data
        
    except Exception as e:
        return None

def main():
    """Extrae el historial de selección nacional de TODOS los jugadores"""
    print(" EXTRACTING PLAYER NATIONAL TEAM HISTORY - ALL PLAYERS")
    print("=" * 70)
    
    try:
        df_existing = pd.read_csv(INPUT_CSV)
        print(f" Loaded {len(df_existing)} players with URLs")
        
        players_to_process = df_existing[
            (df_existing['player_url'] != 'Not found') & 
            (df_existing['player_url'].notna())
        ]
        
        print(f" Players to process: {len(players_to_process)}")
        
    except Exception as e:
        print(f" Error reading CSV: {e}")
        return
    
    driver = setup_stealth_driver()
    if not driver:
        print(" Failed to setup browser driver")
        return
    
    all_national_summaries = []
    all_national_seasons = []
    
    try:
        # Process ALL players
        total_players = len(players_to_process)
        
        for index, row in players_to_process.iterrows():
            player_name = str(row.get('player_name', '')).strip()
            player_slug = str(row.get('player_slug', '')).strip()
            player_url = str(row.get('player_url', '')).strip()
            
            print(f"\n{index + 1}/{total_players}: {player_name}")
            
            # Extract national team history
            national_summary, national_seasons = extract_national_team_data(driver, player_url, player_name, player_slug)
            
            all_national_summaries.append(national_summary)
            all_national_seasons.extend(national_seasons)
            
            # Progress update every 10 players
            if (index + 1) % 10 == 0:
                with_national_team = len([n for n in all_national_summaries if n['has_national_team'] == 'Yes'])
                print(f"\n📊 Progress: {index + 1}/{total_players}, {with_national_team} with national team data")
                
                # Save checkpoint
                if all_national_seasons:
                    df_temp = pd.DataFrame(all_national_seasons)
                    df_temp.to_csv(f"national_team_checkpoint_{index+1}.csv", index=False, encoding='utf-8')
                    print(f"💾 Checkpoint saved: national_team_checkpoint_{index+1}.csv")
            
            # Random delay to avoid blocking
            time.sleep(random.uniform(5, 8))
    
    except KeyboardInterrupt:
        print("\n Process interrupted by user")
        print(" Saving collected data so far...")
    except Exception as e:
        print(f"\n Unexpected error: {e}")
        print(" Saving collected data so far...")
    finally:
        if driver:
            driver.quit()
            print("\n Browser driver closed")
    
    # Save final results
    if all_national_summaries:
        df_summary = pd.DataFrame(all_national_summaries)
        df_summary.to_csv("players_national_team_summary.csv", index=False, encoding='utf-8')
        
        if all_national_seasons:
            df_seasons = pd.DataFrame(all_national_seasons)
            df_seasons.to_csv(OUTPUT_CSV, index=False, encoding='utf-8')
            
            print(f"\n FINAL NATIONAL TEAM SAMPLE DATA (First 10 rows):")
            print(df_seasons.head(10).to_string(index=False))
        
        # Final summary
        total_processed = len(all_national_summaries)
        with_national_team = len([n for n in all_national_summaries if n['has_national_team'] == 'Yes'])
        
        print(f"\n FINAL RESULTS:")
        print(f"    Total players processed: {total_processed}")
        print(f"    Players with national team data: {with_national_team}")
        print(f"    Players without national team data: {total_processed - with_national_team}")
        print(f"    Total national team seasons extracted: {len(all_national_seasons)}")
        
        if total_processed > 0:
            success_rate = (with_national_team / total_processed) * 100
            print(f"    Success rate: {success_rate:.1f}%")
        
        print(f"\n Files saved:")
        print(f"   - players_national_team_summary.csv")
        print(f"   - {OUTPUT_CSV}")
    else:
        print(" No data extracted")

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import re
import time
import random

# ========= CONFIGURATION =========

URLS_CSV = "Second_scrap/Injuries/transfer_url_premier_final.csv" #Csv with player url

# OUTPUT: CSV con todas las lesiones detalladas
INJURIES_CSV = "Second_scrap/Injuries/players_injuries_detailed.csv"

# Batch ( Test 5 first)
PROCESS_SIZE = 1500  # Change to 1500 to process all later

HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/141.0.0.0 Safari/537.36"
}

def fix_encoding(text):
    """Fix encoding issues in text"""
    if not text or text == 'Not found':
        return text
    
    encoding_fixes = {
        'Ãƒ': '', 'â€': '', 'Ã³': 'ó', 'Ã©': 'é', 'Ã': 'í', 
        'Ã±': 'ñ', 'Ãº': 'ú', 'Ã¡': 'á', 'Ã¨': 'è', 'Ãª': 'ê',
        'Ã§': 'ç', 'Ã£': 'ã', 'Ãµ': 'õ', 'Ã¶': 'ö', 'Ã¼': 'ü',
        'Ã¯': 'ï', 'Ã«': 'ë', 'Ã¬': 'ì', 'Ã': 'à'
    }
    
    fixed_text = text
    for wrong, correct in encoding_fixes.items():
        fixed_text = fixed_text.replace(wrong, correct)
    
    return fixed_text

def extract_detailed_injuries(player_url, player_name):
    """
    Extract all injuries detailed for player, each injury is a row
    """
    print(f" Extracting detailed injuries for: {player_name}")
    
    # make a list to save all the injuries per player
    all_injuries = []
    
    try:
        # Construct the url for injury page
        injury_url = player_url.replace('/profil/', '/verletzungen/')
        print(f"    Injury URL: {injury_url}")
        
        response = requests.get(injury_url, headers=HEADERS)
        if response.status_code != 200:
            print(f"    HTTP Error: {response.status_code}")
            return all_injuries
        
        soup = BeautifulSoup(response.content, "html.parser")
        
        # Look for all injury tables
        tables = soup.find_all('table', {'class': 'items'})
        print(f"    Found {len(tables)} tables")
        
        # Principal injury table?
        injury_table = None
        for table in tables:
            header = table.get_text()
            if 'Season' in header and 'Injury' in header and 'from' in header:
                injury_table = table
                print(f"    Found injury table")
                break
        
        if not injury_table:
            print(f"    No injury data found for {player_name}")
            return all_injuries
        
        # Find the tbody of the table
        tbody = injury_table.find('tbody')
        if not tbody:
            print(f"    No tbody found in injury table")
            return all_injuries
        
        # Extract all rows for injuries
        rows = tbody.find_all('tr')
        print(f"    Found {len(rows)} injury records")
        
        for row in rows:
            cells = row.find_all('td')
            if len(cells) >= 6:  # Check if we have all columns
                
                # Extraer datos de cada celda
                season = cells[0].get_text(strip=True) if len(cells) > 0 else 'N/A'
                injury_type = cells[1].get_text(strip=True) if len(cells) > 1 else 'N/A'
                from_date = cells[2].get_text(strip=True) if len(cells) > 2 else 'N/A'
                until_date = cells[3].get_text(strip=True) if len(cells) > 3 else 'N/A'
                days_text = cells[4].get_text(strip=True) if len(cells) > 4 else 'N/A'
                games_missed_text = cells[5].get_text(strip=True) if len(cells) > 5 else 'N/A'
                
                # Extrcat club or games_missed_text (if exist)
                club = 'N/A'
                if games_missed_text:
                    # Format "Club X" o "Club - X"
                    club_match = re.search(r'([A-Za-z\s]+)(?:\d+|$)', games_missed_text)
                    if club_match:
                        club = club_match.group(1).strip()
                
                # Extract number of games lost
                games_num = 0
                if games_missed_text:
                    games_match = re.search(r'(\d+)', games_missed_text)
                    if games_match:
                        games_num = int(games_match.group(1))
                
                # Extrct number of days
                days_num = 0
                if days_text:
                    days_match = re.search(r'(\d+)', days_text)
                    if days_match:
                        days_num = int(days_match.group(1))
                
                # create stats
                injury_record = {
                    'player_name': fix_encoding(player_name),
                    'player_url': player_url,
                    'season': fix_encoding(season),
                    'injury_type': fix_encoding(injury_type),
                    'from_date': from_date,
                    'until_date': until_date,
                    'days': days_num,
                    'days_text': days_text,
                    'games_missed': games_num,
                    'games_missed_text': games_missed_text,
                    'club_at_injury': fix_encoding(club)
                }
                
                all_injuries.append(injury_record)
                
                # Debug: ahow first injury
                if len(all_injuries) <= 2:
                    print(f"      • {season}: {injury_type} - {days_num} days, {games_num} games missed")
        
        print(f"   Extracted {len(all_injuries)} injuries for {player_name}")
        return all_injuries
        
    except Exception as e:
        print(f"    Error extracting injuries: {e}")
        return all_injuries

def extract_player_summary(player_url, player_name):
    """
    Extrae información resumida del jugador (opcional, para tener datos básicos)
    """
    print(f" Extracting summary info for: {player_name}")
    
    player_summary = {
        'player_name': fix_encoding(player_name),
        'player_url': player_url,
        'player_id': 'N/A',
        'full_name': 'N/A',
        'position': 'N/A',
        'current_club': 'N/A',
        'status': 'Success'
    }
    
    try:
        if player_url == 'Not found':
            player_summary['status'] = 'No URL'
            return player_summary
        
        # Extraer player_id de la URL
        player_id = player_url.split('/')[-1]
        player_summary['player_id'] = player_id
        
        response = requests.get(player_url, headers=HEADERS)
        if response.status_code != 200:
            player_summary['status'] = f'HTTP Error: {response.status_code}'
            return player_summary
        
        soup = BeautifulSoup(response.content, "html.parser")
        
        # Full name
        name_element = soup.select_one('h1[class="data-header__headline-wrapper"]')
        if name_element:
            full_name = name_element.text.split('\n')[-1].strip()
            player_summary['full_name'] = fix_encoding(full_name)
        
        # Current club
        club_element = soup.select_one('span[class="data-header__club"]')
        if club_element:
            club_text = club_element.text.split('\n')[-1].strip()
            if club_text and club_text != 'Retired':
                player_summary['current_club'] = fix_encoding(club_text)
        
        # Position from info grid
        data_grid = soup.find('div', class_='info-table')
        if data_grid:
            grid_text = data_grid.get_text()
            pos_match = re.search(r'Position:\s*([^\n]+)', grid_text)
            if pos_match:
                player_summary['position'] = fix_encoding(pos_match.group(1).strip())
        
        return player_summary
        
    except Exception as e:
        print(f"    Summary error: {e}")
        player_summary['status'] = f'Error: {str(e)[:100]}'
        return player_summary

def main():
    """Extrae TODAS las lesiones detalladas para cada jugador"""
    print(" DETAILED INJURY DATA EXTRACTION")
    print("=" * 60)
    
    try:
        # Cargar URLs de jugadores
        df_urls = pd.read_csv(URLS_CSV)
        print(f" Loaded {len(df_urls)} players from CSV")
        
        # Filtrar jugadores con URLs válidas
        valid_players = df_urls[df_urls['transfermarkt_url'] != 'Not found']
        print(f" Found {len(valid_players)} players with valid URLs")
        
        # Tomar solo PROCESS_SIZE para prueba
        sample_players = valid_players.head(PROCESS_SIZE)
        print(f" Processing {len(sample_players)} players for testing")
        
        # Mostrar jugadores a procesar
        print("\n PLAYERS TO PROCESS:")
        for i, (idx, row) in enumerate(sample_players.iterrows(), 1):
            print(f"   {i}. {row['player_name']}")
        
        # Listas para guardar resultados
        all_injuries = []      # Todas las lesiones detalladas
        all_summaries = []      # Resúmenes de jugadores (opcional)
        
        for index, row in sample_players.iterrows():
            player_name = row['player_name']
            player_url = row['transfermarkt_url']
            
            print(f"\n{'='*60}")
            print(f"[{index + 1}/{len(sample_players)}] {player_name}")
            print(f"{'='*60}")
            
            # 1. Extraer resumen del jugador (opcional)
            summary = extract_player_summary(player_url, player_name)
            all_summaries.append(summary)
            
            # 2. Extraer todas las lesiones detalladas
            player_injuries = extract_detailed_injuries(player_url, player_name)
            
            if player_injuries:
                all_injuries.extend(player_injuries)
                print(f"    Added {len(player_injuries)} injuries")
            else:
                print(f"    No injuries found")
            
            # Delay entre requests
            if index + 1 < len(sample_players):
                delay = random.uniform(3, 5)
                print(f" Waiting {delay:.1f} seconds...")
                time.sleep(delay)
        
        # ========= GUARDAR RESULTADOS =========
        
        # 1. Guardar todas las lesiones detalladas
        if all_injuries:
            df_injuries = pd.DataFrame(all_injuries)
            
            # Reordenar columnas para mejor legibilidad
            injury_columns = [
                'player_name', 'season', 'injury_type', 'from_date', 'until_date',
                'days', 'days_text', 'games_missed', 'games_missed_text',
                'club_at_injury', 'player_url'
            ]
            
            # Solo incluir columnas que existen
            final_injury_cols = [col for col in injury_columns if col in df_injuries.columns]
            df_injuries = df_injuries[final_injury_cols]
            
            df_injuries.to_csv(INJURIES_CSV, index=False, encoding='utf-8')
            print(f"\n Detailed injuries saved to: {INJURIES_CSV}")
            print(f"   Total injury records: {len(df_injuries)}")
        
        # 2. (Opcional) Guardar resúmenes de jugadores
        if all_summaries:
            df_summaries = pd.DataFrame(all_summaries)
            summaries_csv = INJURIES_CSV.replace('.csv', '_summary.csv')
            df_summaries.to_csv(summaries_csv, index=False, encoding='utf-8')
            print(f" Player summaries saved to: {summaries_csv}")
        
        # ========= RESUMEN FINAL =========
        print("\n" + "=" * 60)
        print(" EXTRACTION SUMMARY")
        print("=" * 60)
        
        total_players = len(all_summaries)
        players_with_injuries = len(set([i['player_name'] for i in all_injuries]))
        total_injuries = len(all_injuries)
        
        print(f" Players processed: {total_players}")
        print(f" Players with injuries: {players_with_injuries}")
        print(f" Total injury records: {total_injuries}")
        
        if total_injuries > 0:
            avg_injuries = total_injuries / players_with_injuries if players_with_injuries > 0 else 0
            total_days = sum([i.get('days', 0) for i in all_injuries])
            total_games = sum([i.get('games_missed', 0) for i in all_injuries])
            
            print(f" Average injuries per player: {avg_injuries:.1f}")
            print(f" Total days missed: {total_days}")
            print(f" Total games missed: {total_games}")
        
        # Mostrar muestra de lesiones
        if all_injuries:
            print(f"\n SAMPLE INJURY RECORDS (first 5):")
            sample_df = pd.DataFrame(all_injuries[:5])
            columns_to_show = ['player_name', 'season', 'injury_type', 'days', 'games_missed', 'club_at_injury']
            available_columns = [col for col in columns_to_show if col in sample_df.columns]
            print(sample_df[available_columns].to_string(index=False))
        
    except Exception as e:
        print(f" Main error: {e}")

if __name__ == "__main__":
    main()

    # ... (el resto de tu código: imports, funciones fix_encoding, identify_columns, extract_detailed_injuries, etc.) ...

def main():
    """Extrae TODAS las lesiones detalladas, continuando desde donde se quedó."""
    print(" DETAILED INJURY DATA EXTRACTION (CON MODO REANUDACIÓN)")
    print("=" * 60)

    # --- 1. CARGAR LO QUE YA TENEMOS (SI EXISTE) ---
    processed_players = set()
    try:
        df_existing = pd.read_csv(INJURIES_CSV)
        # Asumimos que la columna 'player_name' existe en tu output
        if 'player_name' in df_existing.columns:
            processed_players = set(df_existing['player_name'].unique())
            print(f" Archivo de salida encontrado. {len(processed_players)} jugadores ya procesados.")
        else:
            print(" Archivo de salida encontrado, pero no tiene columna 'player_name'. Se procesarán todos.")
    except FileNotFoundError:
        print(" Archivo de salida no encontrado. Se procesarán todos los jugadores desde cero.")
    except Exception as e:
        print(f" Error al leer archivo de salida: {e}. Se procesarán todos los jugadores.")

    # --- 2. CARGAR LISTA MAESTRA DE JUGADORES ---
    try:
        df_urls = pd.read_csv(URLS_CSV)
        print(f" Cargados {len(df_urls)} jugadores desde {URLS_CSV}")

        name_col, url_col = identify_columns(df_urls)
        if not name_col or not url_col:
            print(" No se pudieron identificar las columnas necesarias.")
            return

        # Filtrar filas con URLs válidas
        valid_players = df_urls[df_urls[url_col].notna()]
        valid_players = valid_players[valid_players[url_col] != 'Not found']
        print(f" {len(valid_players)} jugadores con URLs válidas en la lista maestra.")

    except Exception as e:
        print(f" Error al cargar la lista maestra: {e}")
        return

    # --- 3. FILTRAR PARA QUEDARSE SOLO CON LOS PENDIENTES ---
    players_to_process = valid_players[~valid_players[name_col].isin(processed_players)]
    print(f" Jugadores pendientes por procesar: {len(players_to_process)}")

    # Si no hay nadie pendiente, ¡misión cumplida!
    if players_to_process.empty:
        print(" ¡Todos los jugadores ya han sido procesados! No hay nada que hacer.")
        return

    # --- 4. CONFIGURAR EL RESTO (LISTAS, BUCLE, ETC.) ---
    all_injuries = []  # Solo para las nuevas lesiones
    all_summaries = [] # Solo para los nuevos resúmenes

    # ... (aquí va el resto de tu bucle principal, iterando sobre 'players_to_process' en lugar de 'valid_players') ...
    for index, row in players_to_process.iterrows():
        player_name = row[name_col]
        player_url = row[url_col]

        print(f"\n{'='*60}")
        print(f"[{index + 1}/{len(players_to_process)}] Procesando: {player_name}")
        print(f"{'='*60}")

        # 1. Resumen del jugador
        summary = extract_player_summary(player_url, player_name)
        all_summaries.append(summary)

        # 2. Lesiones detalladas
        player_injuries = extract_detailed_injuries(player_url, player_name)
        if player_injuries:
            all_injuries.extend(player_injuries)
            print(f"    Añadidas {len(player_injuries)} lesiones")
        else:
            print(f"    No se encontraron lesiones para este jugador.")

        # Delay entre requests
        if index + 1 < len(players_to_process):
            delay = random.uniform(3, 5)
            print(f" Esperando {delay:.1f} segundos...")
            time.sleep(delay)

    # --- 5. GUARDAR LOS NUEVOS RESULTADOS (AÑADIENDO A LOS EXISTENTES) ---
    if all_injuries:
        df_new_injuries = pd.DataFrame(all_injuries)
        # Reordenar columnas si es necesario (igual que antes)
        injury_columns = [
            'player_name', 'season', 'injury_type', 'from_date', 'until_date',
            'days', 'days_text', 'games_missed', 'games_missed_text',
            'club_at_injury', 'player_url'
        ]
        final_injury_cols = [col for col in injury_columns if col in df_new_injuries.columns]
        df_new_injuries = df_new_injuries[final_injury_cols]

        # AÑADIR (modo 'a') al CSV existente, sin incluir el encabezado
        df_new_injuries.to_csv(INJURIES_CSV, mode='a', header=not pd.io.common.file_exists(INJURIES_CSV), index=False, encoding='utf-8')
        print(f"\n {len(df_new_injuries)} nuevas lesiones añadidas a: {INJURIES_CSV}")

    # (Opcional) Guardar también los nuevos resúmenes de la misma manera
    # ...

    # --- 6. RESUMEN FINAL ---
    print("\n" + "=" * 60)
    print(" RESUMEN DE ESTA EJECUCIÓN")
    print("=" * 60)
    print(f" Jugadores procesados ahora: {len(all_summaries)}")
    print(f" Nuevos jugadores con lesiones: {len(set([i['player_name'] for i in all_injuries])) if all_injuries else 0}")
    print(f" Nuevos registros de lesiones: {len(all_injuries)}")
    print(f" Proceso de reanudación completado.")
```
